```python
#data cleaning 
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

# covert date colum
```

```python
df.info()
```

```python
#numerical column 
df['Customer_Rating'] = df['Customer_Rating'].fillna(
    df['Customer_Rating'].mean()
)

#categorial cloumn 
df['Region'] = df['Region'].fillna(
    df['Region'].mode()[0]
)
```

```python
#numpy use 
sales = df['Sales'].to_numpy()

print("Mean :",np.mean(sales))
print("Mean :",np.median(sales))
print("Maximum :",np.max(sales))
print("Minimum :",np.min(sales))
```

```python
#pandas 

df['Sales'].sum()
df['Sales'].mean()
df['Sales'].median()
df['Sales'].mode()
df['Sales'].min()
df['Sales'].max()
df['Sales'].std()
df['Sales'].count()
df['Sales'].value_counts()
```

```python
#filtering 
df[df['Sales'] > 50000]

df[df['Region'] == "Ahmedabad"]

df[(df['Sales'] > 50000) & (df['Profit'] > 10000)]
```

```python
#sorting 
df.sort_values(by='Sales')

df.sort_values(by='Sales', ascending=False)
```

```python
#gorupby 
df.groupby('Region')['Sales'].sum()

df.groupby('Product')['Profit'].mean()

df.groupby('Category')['Sales'].sum()
```

```python
# multiple cal 
df.groupby('Region').agg({
    'Sales': ['sum', 'mean'],
    'Profit': ['sum', 'mean'],
    'Quantity': 'sum'
})
```

```python
#line chart sale vadhe che ochu thay
#date vs Sales

plt.figure(figsize=(10,5))

plt.plot(df['Order_Date'],df['Sales'])

plt.title("Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#kaya product vadhare bar chart
#group by 
product_sales = df.groupby('Product')['Sales'].sum()

plt.figure(figsize=(10,5))

plt.bar(product_sales.index,product_sales.values)

plt.title("Product Wise Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")

plt.show()
```

```python
#sales distribution
plt.figure(figsize=(10,5))

plt.hist(df['Sales'],bins=10)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
#category sales sjare 
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(8,5))

plt.pie(
    category_sales,
    labels=category_sales.index,
    autopct='%1.1f%%'
)

plt.title("Category Wise Sales")
plt.show()
```

```python
#scatter plot 
plt.figure(figsize=(8,5))

plt.scatter(df['Sales'],df['Profit'])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
#outerline jovie 
plt.figure(figsize=(8,5))

plt.boxplot(df['Sales'].dropna())

plt.title("Sales Outlier Detection")
plt.ylabel("Sales")

plt.show()
```

```python
#seaborn code ocho satical chart banav use thay

sns.set_theme(style="whitegrid")
```

```python
#ketla order kaya product 
plt.figure(figsize=(10,5))

sns.countplot(data=df,x='Product')

plt.title("Product Order Count")
plt.xticks(rotation=45)
plt.show()
```

```python
#ketla order kaya product 
plt.figure(figsize=(10,5))

sns.barplot(data=df,x='Region',y='Sales')

plt.title("region wise Average Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
#product and sales distribution

plt.figure(figsize=(10,5))

sns.boxplot(data=df,x='Product',y='Sales')

plt.title("Product Wise Sales Distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
#numeri vach correlation 

plt.figure(figsize=(10,6))

corr = df.select_dtypes(include='number').corr()

sns.heatmap(corr,annot=True,cmap='coolwarm')

plt.title("Correlation heatmao")
plt.show()
```
