```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt 


df = pd.read_csv("SampleSuperstore.csv")

df

df["Order Date"] = pd.to_datetime(df['Order Date'])

print(df.head())
```

```python
# montly sales trand 

df["Month"] = df["Order Date"].dt.month

sales = df.groupby('Month')["Sales"].sum()

plt.figure(figsize=(10,5))

plt.plot(sales.index,sales.values,marker="o",color="blue",linewidth=2)


plt.title("Monthly Sales Trand")
plt.xlabel("Month")

plt.ylabel("Sales")

plt.show()
```

```python
#sales and region 

region = df.groupby('Region')["Sales"].sum()

plt.figure(figsize=(8,5))

plt.bar(region.index,region.values,color="green",width=0.5)

plt.title("Sales by Region")
plt.xlabel("Region")
plt.xlabel("Sales")

#text lakhvo hoy to
for i in range(len(region)):
    plt.text(i,region.values[i],
    round(region.values[i],2),ha="center")

plt.show()
```

```python
profit = df.groupby('Sub-Category')["Profit"].sum()

top = profit.sort_values(ascending=False).head(10)

plt.figure(figsize=(8,5))

plt.barh(top.index,top.values,color="orange",height=0.5)


plt.title("Tops 10 Category by profit")
plt.xlabel("Profit")
plt.xlabel("Sub-Category")

# #text lakhvo hoy to
# for i in range(len(region)):
#     plt.text(i,region.values[i],
#     round(region.values[i],2),ha="center")

plt.show()
```

```python
category = df.groupby('Category')["Sales"].sum()

explode = [0,0,0]
explode[category.values.argmax()] = 0.1

plt.figure(figsize=(6,6))

plt.pie(category.values,labels=category.index,autopct='%1.1f%%',explode=explode)

plt.title("Sales by Category")

plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.pie(category.values,labels=category.index,autopct='%1.1f%%',explode=explode,wedgeprops={'width':0.4})

plt.title("Sales Dount Chart")
plt.show()
```
