```python
# Session 7.9 — Mini NumPy Case Study
# હવે અગાઉના તમામ Concepts ને એક વાસ્તવિક Mini Project માં જોડીએ.
# Project: Retail Product Sales Analysis
# Business Problem: એક Store ને દરેક Product નું Revenue, કુલ વેચાણ અને Average Price સમજવું છે. Dataset માં Missing Values અને અસામાન્ય Prices પણ હોઈ શકે છે.

# English Definition: A NumPy case study is a practical exercise that applies array creation, data cleaning, calculations, filtering, and summary statistics to solve a defined analytical problem.

# ગુજરાતી વ્યાખ્યા: NumPy Case Study એટલે એક નિશ્ચિત Business Problem ને ઉકેલવા માટે Array Creation, Data Cleaning, Mathematical Operations, Filtering અને Statistical Functions નો ઉપયોગ કરીને Analysis કરવાની Practical Activity.
```

```python
# Step 1 — Dataset બનાવો

import numpy as np 

products = np.array([
     "Laptop",
    "Mouse",
    "Keyboard",
    "Monitor",
    "Webcam",
    "Speaker"
])

prices = np.array([
    50000.0,
    800.0,
    1500.0,
    12000.0,
    np.nan,
    2500.0
])

quantities = np.array([2,10,5,3,4,6])

print("Products : ",products)
print("Prices : ",prices)
print("Quantities : ",quantities)

# અહીં ત્રણ Arrays છે:
# - products: Product Names.
# - prices: દરેક Product ની Price.
# - quantities: વેચાયેલા Units ની સંખ્યા.
# ત્રણેય Arrays ની Length સમાન છે. એટલે Index 0 પરના Product, Price અને Quantity એકબીજા સાથે સંબંધિત છે.

```

```python
# Step 2 — Dataset Inspect કરો

print("Products shape",products.shape)
print("Prices shape:", prices.shape)
print("Quantities shape:", quantities.shape)

print("Missing prices : ",np.isnan(prices))
print("Missing Count : ",np.isnan(prices).sum()) # total 1 null value

# આથી Missing Price ધરાવતા Product ની સ્થિતિ જાણી શકાય છે.
```

```python
# Step 3 — Missing Data Handle કરો
# આ Demo માટે નક્કી કરીએ કે Missing Price ધરાવતા Product નું Revenue હાલ ગણવામાં નહીં આવે. તેને અંદાજથી Price આપવાને બદલે Missing તરીકે જ રાખીશું.

valid_price = np.isfinite(prices)

print("Valid Price products :",products[valid_price]) # jeni value null/mising aene hal count nahi kari ye
print("Valid prices :",prices[valid_price])
```

```python
# Step 4 — Revenue ગણો 

#Revenue = Price * Quantity 

revenue = prices * quantities

print("Revenue :",revenue)

# Missing Price ધરાવતા Webcam માટે Revenue NaN છે.
```

```python
# Step 5 — Total Revenue શોધો

total_revene = np.nansum(revenue)

print("Total revenue : ",total_revene)

# અહીં Total Revenue ₹1,66,500 છે, જેમાં Missing Price ધરાવતા Webcam નું Revenue સામેલ નથી.
# Business Note: આ Partial Total છે. વાસ્તવિક Report માં Missing Product Revenue નો ઉલ્લેખ કરવો જોઈએ, જેથી Management તેને સંપૂર્ણ Revenue સમજી ન બેસે.
```

```python
# Step 6 — સૌથી વધુ Revenue ધરાવતો Product શોધો

valid_revene = np.isfinite(revenue)

valid_indices = np.where(valid_revene)[0]

best_index = valid_indices[np.argmax(revenue[valid_revene])]

print("Best Product : ",products[best_index])
print("Best Revenue : ",revenue[best_index])

# આ રીતે Revenue ના આધારે સૌથી વધુ કમાણી આપતો Product શોધી શકાય છે.
```

```python
# Step 7 — Filtering કરો
# ₹10,000 કરતાં વધુ Revenue ધરાવતા Products શોધો:

high_revenue = np.isfinite(revenue) & (revenue > 10000)

print(products[high_revenue])
print(revenue[high_revenue])

# અહીં Boolean Indexing અને Logical AND નો ઉપયોગ કર્યો છે.
```
