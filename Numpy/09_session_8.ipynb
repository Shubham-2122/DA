```python
# Session 7.8 — Useful Utility Functions
# આ Session Data Cleaning અને Data Validation માટે ખાસ મહત્વનું છે. Data Analyst ને ફક્ત ગણતરી જ નહીં, પણ ખોટી અથવા અસામાન્ય Values ઓળખતાં પણ આવડવું જોઈએ.
# 7.8.1 np.unique()
# English Definition: np.unique() returns the unique values in an array, optionally along with counts or indices.

# ગુજરાતી: Array માં રહેલી Duplicate Values ને દૂર કરીને Unique Values મેળવવા માટે વપરાય છે.

import numpy as np

a = np.array([10,20,10,30,20,40])  

print(np.unique(a)) # same value nahi aave

print(np.unique(a,return_counts=True)) # counting value ketli var che
```

```python
# 7.8.2 np.sort() અને np.argsort()
# Definition: np.sort() returns sorted values, while np.argsort() returns the indices that would sort the values.

# ગુજરાતી: sort() Values ને ક્રમમાં ગોઠવે છે, જ્યારે argsort() Values ને ગોઠવવા માટે જરૂરી Indices આપે છે.

a = np.array([40,10,30,20])

print(np.sort(a)) # uniqe aape order sort kair dese
print(np.argsort(a))  # product name relation sachvo hoy to
```

```python
# 7.8.3 np.nonzero()
# English Definition: np.nonzero() returns the indices of non-zero elements.

# ગુજરાતી: Zero સિવાયની Values ક્યાં આવેલી છે તેનાં Indices મેળવવા માટે વપરાય છે.

a = np.array([0,10,0,20,30])
print(np.nonzero(a)) # zero sivay ni value ne indices karva
```

```python
# 7.8.4 np.any() અને np.all()

marks = np.array([60,75,30,90])

print(np.any(marks < 35)) # ek value sachi hoy true
print(np.all(marks >=35)) # all value true hovi joei ye

# np.any() — ઓછામાં ઓછી એક Condition સાચી હોય તો True.
# np.all() — બધી Conditions સાચી હોય તો જ True.
```

```python
# 7.8.5 np.isnan(), np.isinf() અને np.isfinite()
# Definitions:
# - np.isnan() checks for NaN values.
# - np.isinf() checks for positive or negative infinity.
# - np.isfinite() checks for values that are neither NaN nor infinite.
# ગુજરાતી સમજ:
# - np.isnan() — Value NaN છે કે નહીં તે તપાસે છે.
# - np.isinf() — Value Positive અથવા Negative Infinity છે કે નહીં તે તપાસે છે.
# - np.isfinite() — Value સામાન્ય Finite Number છે કે નહીં તે તપાસે છે.

a = np.array([10.0,np.nan,np.inf,-np.inf,50.0])

print(np.isnan(a)) # only null value true
print(np.isinf(a)) # null but positive and nagetive both truw
print(np.isfinite(a)) # number hoy true baki false

```

```python
# 7.8.6 np.clip()
# English Definition: np.clip() limits array values to a specified minimum and maximum range.

# ગુજરાતી: આપેલી Minimum અને Maximum મર્યાદા બહારની Values ને તે મર્યાદાની અંદર લાવવા માટે વપરાય છે.

marks = np.array([20,40,60,80,120])

print(np.clip(marks,0,100)) # min and max inside value lava mate use thay

# અહીં 120 ને 100 સુધી મર્યાદિત કરવામાં આવ્યું.
# સાવચેતી: np.clip() Data ને મર્યાદામાં લાવે છે, પરંતુ મૂળ Data ખોટો હતો કે સાચો તે નક્કી કરતું નથી. જો 120 ખરેખર Data Entry Error હોય તો સુધારવાની યોગ્ય રીત પસંદ કરો.
```

```python
# # 7.8.7 NaN Handling
# # NaN નો અર્થ: Not a Number.

# # English Definition: NaN represents an undefined or missing numerical result and requires special handling in numerical analysis.

# # ગુજરાતી: Numerical Data માં કોઈ Value ઉપલબ્ધ ન હોય અથવા ગણતરીથી નિર્ધારિત ન થઈ શકે ત્યારે NaN જેવી Value જોવા મળી શકે છે.

a = np.array([10.0, 20.0, np.nan, 40.0])

print(np.mean(a)) 
print(np.nanmean(a))
print(np.nanstd(a))
print(np.nansum(a))

# મહત્વનો તફાવત:
# Function	વર્તન
# np.mean()	   NaN હોય તો સામાન્ય રીતે પરિણામ NaN બને છે
# np.nanmean()	Mean ગણતી વખતે NaN અવગણે છે 
# np.nanstd()	Standard Deviation ગણતી વખતે NaN અવગણે છે
# np.nansum()	Sum ગણતી વખતે NaN ને અવગણે છે

# Trainer Note: NaN ને અવગણવું દરેક વખતે યોગ્ય નથી. Missing Data નું કારણ સમજવું જરૂરી છે. જો આખા Data માં કોઈ Valid Value જ ન હોય તો np.nanmean() જેવી Function Warning સાથે NaN આપી શકે છે.
```
