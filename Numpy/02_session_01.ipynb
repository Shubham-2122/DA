```python
# Session 7.1 — Introduction to NumPy
# 7.1.1 Python List vs NumPy Array
# Python List માં અલગ-અલગ પ્રકારના Objects રાખી શકાય છે. NumPy Array ખાસ કરીને સમાન પ્રકારના Numerical Data પર કાર્યક્ષમ ગણતરી કરવા માટે બનાવવામાં આવી છે.

import numpy as np 

prices_list = [100,200,300] # list ma merge karse 

prices_array = np.array([100,200,300]) # numpy all element ma sum karse

print(prices_list + [10])
print(prices_array + 10)

```

```python
# 7.1.2 Vectorization શું છે?
#vectorization : all elenent deffent python loop lakhi ya vagar aakha array ma operataion karvu

import numpy as np

marks = np.array([50,30,60,70])

updated_marks = marks + 5 

print(updated_marks)

# Python List માં પણ List Comprehension અથવા અન્ય રીતથી આ કામ થઈ શકે છે, પરંતુ NumPy નું Vectorized Approach Numerical Workloads માં ઘણીવાર વધુ કાર્યક્ષમ હોય છે
```

```python
# 7.1.3 Memory Efficiency

# memeory efficiency - data store and how to work and how use memory

import numpy as np 
import sys

number_list = [10,20,30,40]
number_array = np.array([10,20,30,40] , dtype = np.int64)

print("List object size :",sys.getsizeof(number_list)) # shallow size integer objets
print("array data size : ",number_array.nbytes) # array element data
print("Array object size :",sys.getsizeof(number_array)) # array object no shallow size

```
