```python
# Session 7.6 — Indexing & Slicing
# 7.6.1 Indexing શું છે?
# English Definition: Indexing is the process of accessing a specific element of an array using its position.

# ગુજરાતી: Array માં ચોક્કસ સ્થાન પર રહેલી Value મેળવવાની પ્રક્રિયાને Indexing કહે છે.

# Python અને NumPy માં Indexing સામાન્ય રીતે 0 થી શરૂ થાય છે
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print(a[0])
print(a[2])
print(a[-1])
```

```python
# 7.6.2 2D Array Indexing
a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(a[0,1]) # 0 row and 1 col number
print(a[2,2]) # 2 row and 2 col number
print(a[1]) # all 1 row and col
```

```python
# 7.6.3 Slicing
# English Definition: Slicing extracts a portion of an array using a start position, stop position, and optional step.

# ગુજરાતી: Array માંથી ચોક્કસ Range ની Values મેળવવા માટે Slicing વપરાય છે.
a = np.array([10, 20, 30, 40, 50, 60])

print(a[1:4]) # index0 statrt
print(a[:3]) # end index
print(a[::2]) # jump 2
print(a[::-1]) # revser
```

```python
# 7.6.4 Fancy Indexing
# English Definition: Fancy indexing selects array elements using a list or array of integer indices.

# ગુજરાતી: અનેક ચોક્કસ Index આપીને પસંદગીની Values મેળવવાની પદ્ધતિને Fancy Indexing કહે છે.
a = np.array([100, 200, 300, 400, 500])

print(a[[0, 2, 4]])
```

```python
# 7.6.5 Boolean Indexing
# English Definition: Boolean indexing selects elements where a Boolean condition evaluates to True.

# ગુજરાતી: આપેલી Condition સાચી હોય તે Elements પસંદ કરવાની પદ્ધતિને Boolean Indexing કહે છે.
marks = np.array([35, 55, 80, 42, 90])

print(marks[marks >= 50]) # vlaue share thase

# આ Data Cleaning અને Filtering માટે ખૂબ જ ઉપયોગી છે.
```

```python
# 7.6.6 np.where()
# English Definition: np.where() can return the indices where a condition is true or choose values from two alternatives according to a condition.

# ગુજરાતી: Condition સાચી હોય તે Index શોધવા અથવા Condition પ્રમાણે બે Values માંથી એક પસંદ કરવા માટે વપરાય છે.

marks = np.array([35, 55, 80, 42, 90])

print(np.where(marks >= 50)) # index number aapse
print(np.where(marks >= 50, "Pass", "Fail")) # output fail ya pass
```

```python
# Session 7.6 Assessment
# વિદ્યાર્થીઓને નીચેનો Array આપો:
# 1) પ્રથમ Element મેળવો.
# 2) છેલ્લાં ત્રણ Elements મેળવો.
# 3) દરેક બીજા Element મેળવો.
# 4) Index 1, 3 અને 5 ની Values મેળવો.
# 5) 20 કરતાં મોટી Values મેળવો.
# 6) np.where() વડે 20 કે તેથી મોટી Values માટે "High" અને બાકીની માટે "Low" લખો.

a = np.array([5, 10, 15, 20, 25, 30, 35])

print(a[0])
print(a[4:7])
print(a[1::2])
print(np.where(a>=20,"high","low"))
```
