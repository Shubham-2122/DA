```python
# Session 7.2 — Creating NumPy Arrays
# આ Session માં તમે Array બનાવવાની મહત્વની Methods શીખશો. દરેક Method નો ઉપયોગ ક્યારે કરવો તે સમજવું જરૂરી છે.

# 7.2.1 np.array() : python list or tuple inside numpy array create use

import numpy as np

a = np.array([10,20,30,40]) # 1d array
b = np.array([[10,20,30],[40,50,60]]) # 2d array

print(a)
print(b)

```

```python
# 7.2.2 np.arange() : start value and end value inside step vlaue create use
# શરૂઆતની Value થી અંતની Value પહેલાં સુધી ચોક્કસ Step મુજબ Values બનાવવા માટે તેનો ઉપયોગ થાય છે.


print(np.arange(1,10,2)) # start , end , jump
print(np.arange(1,20,3)) # start ,stop,step
```

```python
# 7.2.3 np.linspace()

# any two number evenly spaced value between start and end point
# બે Values વચ્ચે સમાન અંતર ધરાવતી ચોક્કસ સંખ્યામાં Values બનાવવા માટે વપરાય છે.

print(np.linspace(0,10,5)) # total value joei ne space aape che

print(np.linspace(1,20,6)) # third value dived for space


# arange() vs linspace()
# - arange() માં Step Size આપો છો.
# - linspace() માં કેટલી Values જોઈએ તે આપો છો.
```

```python
# 7.2.4 np.logspace() : aama (10^1, 10^2, 10^3)

print(np.logspace(1,3,3))
print(np.logspace(2,2,3))
```

```python
# 7.2.5 np.zeros() અને np.ones() 
# mare pase dimestion hase aen zero lakhi badhe 0 
# and ones lakhi dimestion badhe one aavse 

print(np.zeros(4))
print(np.zeros((3,3))) # 3 row and 3 col

print(np.ones(5)) # 5 col

print(np.ones((3,4))) # 3 row and 4 col
```

```python
# 7.2.6 np.empty() : merroy allocate kare but value ne 9 paku aaptu nathi

a = np.empty(4)
print(a)

# સાવચેતી: np.empty() ને Zero Array સમજશો નહીં. કોઈ Value પર આધાર રાખતાં પહેલાં દરેક Element માં Value Assign કરો.
```

```python
# 7.2.7 np.eye() અને np.identity()

print(np.eye(3)) # પણ 3 × 3 Matrix બનાવે છે, અને Rows તથા Columns અલગ આપીને Rectangular Matrix પણ બનાવી શકાય છે.
print(np.identity(3)) #3 × 3 Identity Matrix બનાવે છે.
```

```python
# 7.2.8 Random Arrays 

# random data testing,simulations and smaple data 

import numpy as np

print(np.random.rand(3)) 
print(np.random.randn(3))
print(np.random.randint(1,10,size=5))
print(np.random.choice([10,20,30],size=4))

# rand() : 0 thi 1 uniform random values 
# randn() : standard normal distribution under value 
# randint() :  integer range under integers value
# choice() : radom selection 

# randinit(1,10) : 1 aallow che starting and 10 nathi allow
```

```python
np.random.seed(42)
print(np.random.randint(1,100,size=5))

np.random.seed(42)
print(np.random.randint(1, 100, size=5))
```

```python
rng = np.random.default_rng(42)
print(rng.integers(1, 100, size=5))
```
