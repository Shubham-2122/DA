```python
# Session 7.7 — Array Manipulation
# 7.7.1 Array Manipulation શું છે?
# English Definition: Array manipulation refers to changing an array's structure or combining, splitting, inserting, or removing its elements.

# ગુજરાતી: Array ની રચના બદલવી, Arrays જોડવા, ભાગ પાડવા અથવા Elements ઉમેરવા અને દૂર કરવાની પ્રક્રિયાને Array Manipulation કહે છે.
# 7.7.2 np.concatenate()
# Definition: Concatenation joins arrays along an existing axis.

# ગુજરાતી: આપેલા Axis પર બે કે વધુ Arrays ને જોડવા માટે વપરાય છે.
import numpy as np

a = np.array([1,2,3])
b = np.array([4,5,6])

print(np.concatenate((a,b)))
```

```python
# 7.7.3 np.vstack(), np.hstack() અને np.dstack()

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.vstack((a, b)))
print(np.hstack((a, b)))
print(np.dstack((a, b)))

# Function	ગુજરાતી સમજ
# np.vstack()	Arrays ને Vertical દિશામાં Stack કરે છે
# np.hstack()	Arrays ને Horizontal દિશામાં Stack કરે છે
# np.dstack()	Arrays ને ત્રીજા Axis પર Stack કરે છે
```

```python
# 7.7.4 np.split() અને np.array_split()

a = np.array([10, 20, 30, 40, 50, 60])

print(np.split(a, 3))
print(np.array_split(a, 4))

# np.split() — સમાન ભાગોમાં Split શક્ય ન હોય તો સામાન્ય રીતે Error આપે છે.
# np.array_split() — સમાન ભાગોમાં વહેંચી ન શકાય ત્યારે અલગ-અલગ Size ના ભાગો આપી શકે છે.
```

```python
# 7.7.5 np.insert(), np.append() અને np.delete()

a = np.array([10, 20, 30])

print(np.insert(a, 1, 15)) # index parlakhi ne add thase
print(np.append(a, 40)) # after index add thase
print(np.delete(a, 1)) # index delete karse
```

```python
# 7.7.6 Copy vs View
# English Definition: A copy owns separate array data, whereas a view provides a different way to access the same underlying data when sharing is possible.

# ગુજરાતી: Copy માં Data ની અલગ નકલ બને છે; View માં મૂળ Data સાથે Memory શેર થઈ શકે છે.

a = np.array([10, 20, 30, 40])

view = a.view()
copy = a.copy()

view[0] = 99

print("Original:", a) # add karse view index par
print("Copy:", copy) # copy karse

# અહીં View માં ફેરફાર થતાં મૂળ Array પણ બદલાયો, જ્યારે Copy બદલાઈ નહીં.

# Slicing પણ ઘણીવાર View આપે છે:

a = np.array([10, 20, 30, 40])

part = a[1:3]
part[0] = 999

print(a)

```
