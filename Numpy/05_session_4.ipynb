```python
# Session 7.4 — Array Operations
# 7.4.1 Arithmetic Operations
# English Definition: Arithmetic operations perform numerical calculations element by element on compatible NumPy arrays.

# ગુજરાતી: Compatible Shape ધરાવતા Arrays પર ગણિતીય Operations દરેક Element પર લાગુ કરી શકાય છે.

import numpy as np

a = np.array([10,20,30])
b = np.array([2,4,5])

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b) # floor division 
print(a**b) # number che aen time lakiya hoy multiple kare 
print(a%b)
```

```python
# 7.4.2 Comparison Operators
# બે Values ની સરખામણી કરીને True અથવા False આપતા Operators.

a = np.array([10,20,30,40])

print(a>20)
print(a ==20)
print(a!=30)
print(a<=30)
```

```python
# 7.4.3 Logical Operations
# બે કે વધુ Conditions ને AND/OR વડે જોડવા અથવા Condition ને ઉલટાવવા માટે Logical Functions વપરાય છે.

marks = np.array([35,55,75,90])

print(np.logical_and(marks >= 40,marks < 80))
print(np.logical_or(marks < 40,marks>=90))
print(np.logical_not(marks >=40))
```

```python
# # 7.4.4 Broadcasting

# nglish Definition: Broadcasting allows NumPy to perform element-wise operations on arrays of compatible shapes without explicitly repeating data.

# ગુજરાતી: Broadcasting એટલે Compatible Shapes ધરાવતા Arrays પર Data ને સ્પષ્ટ રીતે Repeat કર્યા વગર Operations કરવાની સુવિધા.

a = np.array([10,20,30])
print(a+5)
```

```python
# all element sum karse
a = np.array([[1,2,3],[4,5,6]])
b=np.array([10,30,30])

print(a+b)

# મહત્વનો નિયમ: Dimensions ને જમણી બાજુથી સરખાવો. દરેક Dimension સમાન હોવી જોઈએ અથવા તેમાંથી એક Dimension 1 હોવી જોઈએ. અન્યથા Broadcasting Error આવી શકે છે.
```

```python
# 7.4.5 Element-wise vs Aggregate Operations

a= np.array([10,20,30])

print(a*2)
print(np.sum(a))
print(np.mean(a)) # total sum / total element
```
