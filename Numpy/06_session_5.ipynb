```python
# Session 7.5 — Mathematical & Statistical Functions
# આ Session Data Analyst માટે ખૂબ જ ઉપયોગી છે. NumPy ની મદદથી તમે Sales, Profit, Marks, Prices અને અન્ય Numerical Data ની ગણતરી કરી શકો છો.
# 7.5.1 Mathematical Functions
# English Definition: NumPy mathematical functions perform numerical operations on individual values or entire arrays.

# ગુજરાતી: NumPy Mathematical Functions એક Value અથવા Array ના દરેક Element પર ગણિતીય ગણતરી કરે છે.

import numpy as np

a= np.array([4,9,16])

print(np.sqrt(a))
print(np.exp(a))

print(np.log(a))
print(np.log10(a))

# Function	    ઉપયોગ	        ઉદાહરણ
# np.add()	    Addition	    np.add(2, 3) → 5
# np.subtract()	Subtraction	    np.subtract(5, 2) → 3
# np.multiply()	Multiplication	np.multiply(4, 3) → 12
# np.divide()	Division	    np.divide(10, 2) → 5
# np.power()	Power	        np.power(2, 3) → 8
# np.mod()	    Remainder	    np.mod(10, 3) → 1
# np.sqrt()	    Square Root	    np.sqrt(16) → 4
# np.exp()	    e^x   	        np.exp(1) → આશરે 2.718
# np.log()	    Natural Log	    np.log(np.e) → 1
# np.log10()	Base-10 Log	    np.log10(100) → 2
```

```python
prices = np.array([100, 400, 900])

print(np.sqrt(prices))
```

```python
# 7.5.2 Trigonometric Functions
# English Definition: Trigonometric functions calculate mathematical relationships involving angles.

# ગુજરાતી: Angles સાથે જોડાયેલા Sine, Cosine અને Tangent જેવા ગણિતીય Functions ને Trigonometric Functions કહે છે.

# NumPy ના Trigonometric Functions સામાન્ય રીતે Angles ને Radians માં લે છે

import numpy as np

ag = np.array([0,30,45,60,90])

radians = np.radians(ag)

print(np.sin(radians))
print(np.cos(radians))
print(np.tan(radians))

# અહીં np.radians() Degrees ને Radians માં ફેરવે છે.

# મહત્વના Functions:
# - np.sin(x) — Sine.
# - np.cos(x) — Cosine.
# - np.tan(x) — Tangent.
# - np.arcsin(x) — Inverse Sine, જે Radians માં Angle આપે છે.
# - np.degrees(x) — Radians ને Degrees માં ફેરવે છે.
# - np.radians(x) — Degrees ને Radians માં ફેરવે છે

print(np.degrees(np.pi))
print(np.radians(180))
```

```python
# 7.5.3 Rounding Functions
# English Definition: Rounding functions modify numerical values according to rounding rules or move values toward specified integer boundaries.

# ગુજરાતી: સંખ્યાને નજીકની Value અથવા પૂર્ણાંક તરફ ફેરવવા માટે Rounding Functions વપરાય છે.

a = np.array([2.3, 2.5, 2.8, -2.3, -2.8])

print(np.round(a))
print(np.floor(a))
print(np.ceil(a))

# np.round() — નજીકની Rounding Value આપે છે. ચોક્કસ અડધી Value માટે NumPy સામાન્ય રીતે nearest even તરફ Round કરે છે.
# np.floor() — Value કરતાં નાનો અથવા સમાન સૌથી નજીકનો Integer.
# np.ceil() — Value કરતાં મોટો અથવા સમાન સૌથી નજીકનો Integer.
```

```python
# 7.5.4 Statistical Functions
# English Definition: Statistical functions summarize numerical data using measures such as central tendency, spread, totals, and percentiles.

# ગુજરાતી: Data નો સરેરાશ, મધ્ય Value, ફેલાવો, કુલ સરવાળો અને અન્ય Summary Measures શોધવા માટે Statistical Functions વપરાય છે.

marks = np.array([50, 60, 70, 80, 90])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
print("Variance:", np.var(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Sum:", np.sum(marks))
print("Cumulative Sum:", np.cumsum(marks))
print("Product:", np.prod(marks))
print("75th Percentile:", np.percentile(marks, 75))


# Function	Definition અને ગુજરાતી અર્થ
# np.mean()	Arithmetic Average — સરેરાશ
# np.median()	મધ્યમાં આવતી Value
# np.std()	Standard Deviation — Data નો ફેલાવો
# np.var()	Variance — ફેલાવાનું વર્ગ આધારિત માપ
# np.min()	સૌથી નાની Value
# np.max()	સૌથી મોટી Value
# np.sum()	બધી Values નો સરવાળો
# np.cumsum()	Running અથવા Cumulative Sum
# np.prod()	બધી Values નો ગુણાકાર
# np.percentile()	આપેલા Percentile સુધીની Value
```

```python
# Standard Deviation અને Variance
# NumPy માં np.std() અને np.var() Default રૂપે Population Standard Deviation અને Population Variance ગણે છે, એટલે ddof=0.

# જો Sample Variance અથવા Sample Standard Deviation જોઈએ તો ddof=1 વાપરો.
data = np.array([10, 20, 30, 40, 50])

print(np.var(data))
print(np.var(data, ddof=1))

print(np.std(data))
print(np.std(data, ddof=1))
```
