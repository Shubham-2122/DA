```python
# Session 7.3 — Array Attributes & Properties
# 7.3.1 મુખ્ય Attributes

#Array ના Attributes તેની રચના, Size, Dimensions, Data Type અને Memory વિશે માહિતી આપે છે.

import numpy as np

a = np.array([[10,20,30],[40,50,60]])

print("Array :\n",a)
print("ndim :",a.ndim)
print("shape :",a.shape)
print("size :",a.size)
print("dtype : ",a.dtype)
print("itemsize : ",a.itemsize)
print("nbytes :",a.nbytes)
```

```python
# 7.3.2 reshape()

# Elements ની સંખ્યા સમાન રાખીને Array ને નવી Rows અને Columns ની રચનામાં ફેરવવા માટે વપરાય છે.

a = np.arange(1,7)

print(a.reshape(2,3)) # two row and 3 col

print(a.reshape(3,2)) # three row and 2 col

print(a.reshape(2,-1)) # -1 baki na auto matically set karse
```

```python
# 7.3.3 ravel() vs flatten()

# બંને 2D અથવા Multidimensional Array ને 1D સ્વરૂપમાં ફેરવે છે. ravel() શક્ય હોય ત્યારે મૂળ Data સાથે જોડાયેલ View આપે છે; flatten() અલગ Copy આપે છે

a= np.array([[1,2],[3,4]])
print(a)

r= a.ravel()
f= a.flatten()

print(r) # 1d ma convert karvu
print(f) # 1d ma convert karvu
```

```python
# 7.3.4 resize()
# NumPy માં np.resize() અને ndarray.resize() બંને છે, પરંતુ બંનેનું વર્તન અલગ છે
# np.resize() નવી Shape ધરાવતો Array આપે છે. જો વધુ Elements જોઈએ તો મૂળ Data ફરીથી વાપરી શકે છે. 
# જ્યારે a.resize() મૂળ Array નો Size બદલી શકે છે

a = np.array([1,2,3])

print(np.resize(a,(2,3)))
# અહીં Data Repeat થયું છે. આ Missing Data ભરવાની પદ્ધતિ નથી
```

```python
# 7.3.5 Transpose — .T અને np.transpose()

# 2D Matrix માં Rows ને Columns અને Columns ને Rows માં ફેરવવાની પ્રક્રિયાને Transpose કહે છે.

a = np.array([[1,2,3],[4,5,6]])

print(a.T)
print(np.transpose(a))
```
