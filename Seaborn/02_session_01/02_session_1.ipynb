```python
# 2. Matplotlib vs Seaborn
# Feature	Matplotlib	Seaborn
# મુખ્ય ઉપયોગ	General-purpose plotting	Statistical visualization
# Default appearance	Basic plotting style	Statistical charts માટે તૈયાર styling
# Pandas DataFrame	Support કરે છે	સીધો ઉપયોગ ખૂબ સરળ છે
# Statistical plots	બનાવી શકાય છે, ઘણી વખત વધુ code જોઈએ	ઘણી statistical plots માટે built-in functions
# Correlation heatmap	Manually બનાવી શકાય	sns.heatmap()
# Regression visualization	અલગથી implement કરી શકાય	sns.regplot(), sns.lmplot()
# Customization	ખૂબ flexible	Matplotlib સાથે flexible
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Seaborn version:", sns.__version__)
```

```python
# 4. Step 2 — Sample dataset બનાવો
# હાલ આપણે નાના student dataset થી શરૂઆત કરીશું, જેથી દરેક output સરળતાથી સમજાય.
import pandas as pd

data = {
    "Student": [
        "Amit", "Neha", "Ravi",
        "Pooja", "Karan", "Meera",
        "Jay", "Priya"
    ],
    "Python": [85, 92, 70, 88, 65, 95, 78, 82],
    "SQL": [78, 88, 75, 90, 68, 92, 80, 85],
    "Department": [
        "IT", "IT", "Computer", "Computer",
        "IT", "Computer", "IT", "Computer"
    ]
}

df = pd.DataFrame(data)
df
print(df)
```

```python
# Dataset ની basic information તપાસો:

print(df.head())
print(df.info())
print(df.describe())
```

```python
# 5. Step 3 — Matplotlib માં chart બનાવો
# સૌપ્રથમ Matplotlib માં Python marks નો chart બનાવીએ.

plt.figure(figsize=(8,5))

plt.bar(df["Student"],df["Python"])

plt.title("Student Python Marks")
plt.xlabel("Student")
plt.ylabel("Marks")

plt.show()
```

```python
# 6. Step 4 — Seaborn માં એ જ chart બનાવો
# je aapre matploite thodu code vadhre lakhvpade ae reuse thase

sns.barplot(
    data=df,
    x="Student",
    y="Python"
)

plt.title("Student python Marks")

plt.show()


# data=df — કયા DataFrame માંથી data લેવો છે.
# x="Student" — X-axis પર student names.
# y="Python" — Y-axis પર Python marks.
# sns.barplot() — categories પ્રમાણે numerical values નો statistical bar plot બનાવે છે; default estimator mean છે.
```

```python
# 7. Step 5 — Seaborn Themes
# Seaborn માં themes એટલે chart નો overall visual style — background, grid અને axes ની appearance.

#darkgrid
#whitegrid
#ticks

# ત્રણેય themes નો practical

import matplotlib.pyplot as plt
import seaborn as sns

themes =  ["darkgrid", "whitegrid", "ticks"]

for theme in themes:
    sns.set_theme(style=theme)

    fig,ax = plt.subplots(figsize=(7,4))

    sns.barplot(
        data=df,
        x="Student",
        y="Python",
        ax=ax
    )

    ax.set_title(f"Theme: {theme}")
    ax.set_ylim(0, 100)

    plt.show()
```

```python
# 8. Step 6 — Figure-level vs Axes-level functions
# આ Seaborn નું એક મહત્વનું concept છે.
# Axes-level functions
# આ functions સામાન્ય રીતે એક axes પર એક plot બનાવે છે અને existing Matplotlib axes સાથે સરળતાથી કામ કરે છે.

# ઉદાહરણો:
# - sns.scatterplot()
# - sns.barplot()
# - sns.histplot()
# - sns.boxplot()
# - sns.regplot()

fig, ax = plt.subplots(figsize=(7, 4))

sns.scatterplot(
    data=df,
    x="Python",
    y="SQL",
    hue="Department",
    ax=ax
)

ax.set_title("Python vs SQL Marks")
plt.show()

# અહીં ax=ax થી Seaborn ને જણાવીએ છીએ કે chart બનાવતી વખતે કયા Matplotlib axes નો ઉપયોગ કરવો.
```

```python
# figure-level functions
# આ functions આખી figure ને manage કરે છે. ખાસ કરીને multiple panels અથવા અલગ subsets માટે ઉપયોગી છે.

# ઉદાહરણો:
# - sns.relplot()
# - sns.catplot()
# - sns.displot()
# - sns.lmplot()
# - sns.pairplot()

sns.relplot(
    data=df,
    x="Python",
    y="SQL",
    hue="Department",
    kind="scatter"
)

plt.show()
```

```python
# આમાં relplot() પોતે figure અને axes બનાવે છે; તેથી સામાન્ય રીતે ax=ax આપતા નથી.
# Feature	Axes-level	Figure-level
# Example	sns.scatterplot()	sns.relplot()
# Figure control	Existing axes પર plot	Figure-level layout manage કરે
# Subplots/facets	Matplotlib વડે જાતે ગોઠવી શકાય	Built-in faceting support
# ax= argument	સામાન્ય રીતે ઉપલબ્ધ	સામાન્ય રીતે ઉપયોગ થતો નથી
```

```python
# 9. Step 7 — Theme અને context સાથે chart બનાવો
# Seaborn માં theme ઉપરાંત context વડે text અને chart elements નું scale બદલી શકાય છે.

sns.set_theme(
    style="whitegrid",
    context="notebook",
    palette="deep"
)

fig, ax = plt.subplots(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Python",
    y="SQL",
    hue="Department",
    s=100,
    ax=ax
)

ax.set_title("Student Marks Relationship")
plt.show()

# style="whitegrid" — background અને grid style.
# context="notebook" — notebook માટે અનુકૂળ size.
# palette="deep" — Seaborn ની predefined color palette.
# hue="Department" — department પ્રમાણે colors અને legend.
# s=100 — scatter markers નું size.
```
