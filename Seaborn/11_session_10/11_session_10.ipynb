```python
# Session 10.10 — Advanced Seaborn Styling
# આજે આપણે Seaborn charts ને professional, readable અને presentation-ready બનાવતાં શીખીશું. અત્યાર સુધી આપણે Countplot, Histplot, Boxplot, Pairplot, Heatmap અને Regression plots બનાવ્યા. હવે આ charts ની theme, colors, font sizes અને overall appearance ને નિયંત્રિત કરીશું.

# આ session માં આપણે શીખીશું:
# 1. sns.set_theme() અને sns.set()
# 2. Styles — darkgrid, whitegrid, dark, white, ticks
# 3. Color palettes — deep, muted, colorblind, bright, pastel, dark
# 4. Context — paper, notebook, talk, poster
# 5. sns.color_palette() અને sns.palplot()
# 6. Chart title, labels, fonts અને layout ને customize કરવું
# 7. એક જ theme નો ઉપયોગ કરીને multiple charts ને consistent બનાવવું.
```

```python
# 1. Setup અને Dataset
# VS Code માં Jupyter Notebook ખોલો અને નીચેનો code ચલાવો.

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Student": [
        "Amit", "Neha", "Ravi", "Pooja", "Karan",
        "Meera", "Jay", "Priya", "Rahul", "Nisha"
    ],
    "Department": [
        "IT", "IT", "Computer", "Computer", "IT",
        "Computer", "IT", "Computer", "IT", "Computer"
    ],
    "Python": [85, 92, 70, 88, 65, 95, 78, 82, 90, 86],
    "SQL": [78, 88, 75, 90, 68, 92, 80, 85, 84, 89]
}

df = pd.DataFrame(data)

print(df.head())
```

```python
# 2. sns.set_theme() — Global Theme
# Seaborn માં set_theme() દ્વારા default styling નક્કી કરી શકાય છે.

sns.set_theme(
    style="whitegrid",
    context="notebook",
    palette="deep"
)

sns.barplot(
    data=df,
    x="Department",
    y="Python",
    hue="Department"
)

plt.title("Average Python Marks by Department")
plt.show()

# Parameter	કામ
# style	Background અને grid lines ની શૈલી
# context	Text અને chart elements નું relative size
# palette	Colors ની default scheme
```

```python
# 3. Styles — Chart નો Background બદલવો
# Seaborn માં પાંચ સામાન્ય styles છે:
# - darkgrid
# - whitegrid
# - dark
# - white
# - ticks
# દરેક style નો હેતુ અલગ છે. ઉદાહરણ તરીકે, whitegrid માં white background સાથે grid lines આવે છે, જે numerical values વાંચવામાં મદદ કરે છે.

```

```python
# Practical — Styles ની સરખામણી
# દરેક code block અલગ ચલાવો.
import seaborn as sns

sns.set_theme(style="darkgrid")

sns.barplot(
    data=df,
    x="Department",
    y="Python",
    hue="Department",
    legend=False
)

plt.title("Darkgrid Style")
plt.show()
```

```python
sns.set_theme(style="whitegrid")

sns.barplot(
    data=df,
    x="Department",
    y="Python",
    hue="Department",
    legend=False
)

plt.title("Whitegrid Style")
plt.show()
```

```python
sns.set_theme(style="ticks")

sns.barplot(
    data=df,
    x="Department",
    y="Python",
    hue="Department",
    legend=False
)

sns.despine()
plt.title("Ticks Style")
plt.show()
```

```python
# 4. Color Palettes — Charts માટે યોગ્ય રંગો
# Palette એટલે રંગોની તૈયાર scheme. રંગો પસંદ કરતી વખતે readability અને accessibility ધ્યાનમાં રાખવી જોઈએ.
# Palette	ઉપયોગ
# deep	સામાન્ય purpose માટે balanced colors
# muted	હળવા અને ઓછા તેજસ્વી રંગો
# colorblind	રંગ ઓળખવામાં મુશ્કેલી ધરાવતા લોકો માટે વધુ યોગ્ય પસંદગી
# bright	વધારે તેજસ્વી રંગો
# pastel	હળવા રંગો
# dark	ઘાટા રંગો
sns.palplot(sns.color_palette("deep"))
plt.show()

sns.palplot(sns.color_palette("muted"))
plt.show()

sns.palplot(sns.color_palette("colorblind"))
plt.show()

sns.palplot(sns.color_palette("pastel"))
plt.show()
```

```python
# Chart માં palette લાગુ કરો
sns.set_theme(style="whitegrid", palette="colorblind")

sns.scatterplot(
    data=df,
    x="Python",
    y="SQL",
    hue="Department",
    s=100
)

plt.title("Python vs SQL — Colorblind Palette")
plt.show()
```

```python
# 5. Context — Chart અને Text નું કદ
# context chart ના વિવિધ visual elements નું relative size નિયંત્રિત કરે છે.
# Context	ઉપયોગ
# paper	નાનાં charts અને papers માટે
# notebook	Jupyter Notebook માં સામાન્ય analysis
# talk	Presentation અથવા classroom માં બતાવવા
# poster	મોટા poster અથવા દૂરથી દેખાવા માટે
sns.set_theme(
    style="whitegrid",
    context="notebook",
    palette="deep"
)

sns.scatterplot(
    data=df,
    x="Python",
    y="SQL",
    hue="Department",
    s=100
)

plt.title("Notebook Context")
plt.show()
```

```python
# હવે ફક્ત context બદલીને જુઓ:

sns.set_theme(
    style="whitegrid",
    context="talk",
    palette="deep"
)

sns.scatterplot(
    data=df,
    x="Python",
    y="SQL",
    hue="Department",
    s=100
)

plt.title("Talk Context")
plt.show()
```

```python
# 6. sns.set() અને sns.set_theme() વચ્ચેનો ફરક
# જૂના tutorials માં ઘણીવાર sns.set() જોવા મળે છે.

sns.set(style="whitegrid", palette="deep")

sns.set_theme(
    style="whitegrid",
    context="notebook",
    palette="deep"
)
```

```python
# 7. Complete Professional Styling Example
# હવે theme, palette, title, labels અને layout ને એકસાથે લાગુ કરીએ.

sns.set_theme(
    style="whitegrid",
    context="talk",
    palette="colorblind"
)

fig, ax = plt.subplots(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Python",
    y="SQL",
    hue="Department",
    style="Department",
    s=130,
    ax=ax
)

ax.set_title(
    "Python vs SQL Performance",
    fontsize=18,
    fontweight="bold",
    pad=15 # title and chart between space
)

ax.set_xlabel("Python Marks", fontsize=13)
ax.set_ylabel("SQL Marks", fontsize=13)

ax.legend(title="Department")
plt.tight_layout()
plt.show()
```
