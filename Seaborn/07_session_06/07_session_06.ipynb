```python
# Session 10.6 — Jointplot (Relationship Analysis)
# આજે આપણે Seaborn ની sns.jointplot() શીખીશું. આનો ઉપયોગ બે numerical variables વચ્ચેનો સંબંધ સમજવા માટે થાય છે. સાથે જ બંને variables નું distribution પણ જોઈ શકીએ છીએ.

# ઉદાહરણ તરીકે:
# - વિદ્યાર્થીના Python અને SQL marks વચ્ચે સંબંધ છે કે નહીં?
# - Advertising ખર્ચ વધે ત્યારે Sales વધે છે કે નહીં?
# - Height અને Weight વચ્ચે કેવો સંબંધ છે?
# - બે numerical columns માં data ક્યાં વધારે એકત્રિત થયો છે?

# એટલે jointplot() એક જ visualization માં relationship + બંને variables નું distribution બતાવી શકે છે
```

```python
# 2. Setup અને Dataset
# VS Code માં તમારી Jupyter Notebook ખોલીને નીચેનો code ચલાવો. જો libraries ઇન્સ્ટોલ ન હોય તો Notebook cell માં પહેલાં આ ચલાવો:
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

data = {
    "Student": [
        "Amit", "Neha", "Ravi", "Pooja", "Karan",
        "Meera", "Jay", "Priya", "Rahul", "Nisha",
        "Dev", "Riya"
    ],
    "Department": [
        "IT", "IT", "Computer", "Computer",
        "IT", "Computer", "IT", "Computer",
        "IT", "Computer", "IT", "Computer"
    ],
    "Python": [85, 92, 70, 88, 65, 95, 78, 82, 90, 86, 73, 96],
    "SQL": [78, 88, 75, 90, 68, 92, 80, 85, 84, 89, 72, 94],
    "Maths": [80, 91, 72, 85, 66, 93, 76, 84, 88, 87, 75, 95]
}

df = pd.DataFrame(data)

print(df.head())

# Python — X-axis માટે વાપરીશું.
# SQL — Y-axis માટે વાપરીશું.
# Department — જરૂર પડે ત્યારે groups અલગ પાડવા માટે વાપરીશું.
```

```python
# 3. Basic Jointplot — Scatter પ્રકાર
# હવે Python અને SQL marks વચ્ચેનો સંબંધ જોઈશું.

sns.jointplot(
    data=df,
    x="Python",
    y="SQL",
    kind="scatter" #મધ્યમાં scatter plot બનાવવો
)

plt.show()
```

```python
# 1. Scatter
sns.jointplot(data=df, x="Python", y="SQL", kind="scatter")
plt.show()

# 2. Hexbin
sns.jointplot(data=df, x="Python", y="SQL", kind="hex")
plt.show()

# 3. KDE
sns.jointplot(data=df, x="Python", y="SQL", kind="kde")
plt.show()

# 4. 2D Histogram
sns.jointplot(data=df, x="Python", y="SQL", kind="hist")
plt.show()

# 5. Regression
sns.jointplot(data=df, x="Python", y="SQL", kind="reg")
plt.show()
```

```python
# 5. Jointplot ને Customize કરવો
g = sns.jointplot(
    data=df,
    x="Python",
    y="SQL",
    kind="scatter",
    height=7, #આખા figure ની ઊંચાઈ અને પહોળાઈ નિયંત્રિત કરે છે
    ratio=4, #મધ્યના plot અને marginal plots નો size ratio
    space=0.15, #મધ્યના plot અને marginal plots વચ્ચેનું અંતર
    color="teal",
    marginal_kws={"bins": 6} #Marginal histograms માં 6 bins રાખવા
)

g.set_axis_labels("Python Marks", "SQL Marks")
plt.show()
```

```python
# 6. hue થી Department પ્રમાણે સરખામણી
# હવે IT અને Computer department ના વિદ્યાર્થીઓને અલગ રંગમાં જોઈશું.

sns.jointplot(
    data=df,
    x="Python",
    y="SQL",
    hue="Department",
    kind="scatter",
    height=7
)

plt.show()
```

```python
# 7. Industry Practical — Advertising ખર્ચ અને Sales
# આવા charts business analysis માં ઉપયોગી છે. ઉદાહરણ તરીકે, કોઈ કંપની વિવિધ campaigns પર advertising ખર્ચ કરે છે અને દરેક campaign પછી sales નોંધે છે.
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

business_data = {
    "Advertising": [5, 8, 10, 12, 15, 18, 20, 22, 25, 30],
    "Sales":       [20, 25, 28, 35, 38, 42, 48, 50, 58, 65]
}

sales_df = pd.DataFrame(business_data)

sns.jointplot(
    data=sales_df,
    x="Advertising",
    y="Sales",
    kind="reg",
    height=7,
    color="green"
)

plt.show()
```
