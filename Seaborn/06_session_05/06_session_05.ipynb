```python
# # Session 10.5 — Pairplot (Multivariate Analysis)
# # આજે આપણે Seaborn નું sns.pairplot() શીખીશું. Data Science માં ઘણી વખત dataset માં એક કરતાં વધુ numerical columns હોય છે. દરેક column માટે અલગ chart બનાવવાને બદલે, Pairplot દ્વારા આપણે અનેક columns વચ્ચેના relationships એકસાથે જોઈ શકીએ છીએ.
# Pairplot એક scatter plot matrix બનાવે છે.

# તેમાં:
# - Diagonal plots: દરેક numerical feature નું distribution બતાવે છે.
# - Off-diagonal plots: બે numerical features વચ્ચેનો સંબંધ બતાવે છે.
# - hue: Category પ્રમાણે observations ને અલગ રંગથી દર્શાવે છે.
# ઉદાહરણ તરીકે, વિદ્યાર્થીના Python, SQL અને Maths marks વચ્ચેનો સંબંધ એક જ Pairplot માં જોઈ શકાય છે.
# 2. Learning objectives
# આ session પછી તમે:
# - sns.pairplot() નો ઉપયોગ કરી શકશો.
# - hue, vars, diag_kind અને kind parameters સમજી શકશો.
# - Numerical features વચ્ચેના patterns જોઈ શકશો.
# - Iris dataset પર Multivariate Analysis કરી શકશો.
# - Pairplot પરથી યોગ્ય observations લખી શકશો.
```

```python
# 3. Practical 1 — Student Dataset બનાવો
# VS Code અથવા Jupyter Notebook માં નીચેનો code ચલાવો

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

data = {
    "Student": [
        "Amit", "Neha", "Ravi", "Pooja", "Karan",
        "Meera", "Jay", "Priya", "Rahul", "Nisha",
        "Dev", "Riya"
    ],
    "Department": [
        "IT", "IT", "Computer", "Computer", "IT",
        "Computer", "IT", "Computer", "IT", "Computer",
        "IT", "Computer"
    ],
    "Python": [85, 92, 70, 88, 65, 95, 78, 82, 90, 86, 73, 96],
    "SQL": [78, 88, 75, 90, 68, 92, 80, 85, 84, 89, 72, 94],
    "Maths": [80, 91, 72, 85, 66, 93, 76, 84, 88, 87, 75, 95]
}

df = pd.DataFrame(data)

print(df.head())
```

```python
# 4. Practical 2 — Basic Pairplot

sns.pairplot(
    data=df,
    vars=["Python","SQL","Maths"]
)

plt.show()
```

```python
# 5. Practical 3 — hue Parameter
# હવે IT અને Computer department ને અલગ રંગથી દર્શાવીએ.

sns.pairplot(
    data=df,
    vars=["Python", "SQL", "Maths"],
    hue="Department" # colro diffrent batavse
)

plt.show()
```

```python
# 6. Practical 4 — kind Parameter
# # Pairplot ના off-diagonal plots નો પ્રકાર બદલી શકાય છે.

# A. Scatter Plot
sns.pairplot(
    data=df,
    vars=["Python", "SQL", "Maths"],
    kind="scatter" #observations ને dots તરીકે દર્શાવે છે. આ default પ્રકાર છે.
)

plt.show()
```

```python
# B. KDE Plot

sns.pairplot(
    data=df,
     vars=["Python", "SQL", "Maths"],
    kind="kde" # density estimates
)

plt.show()
```

```python
# C. Regression Plot

sns.pairplot(
    data=df,
    vars=["Python", "SQL", "Maths"],
    kind="reg" #scatter points સાથે regression line દર્શાવે છે
)

plt.show()
```

```python
# 7. Practical 5 — diag_kind Parameter
# Diagonal plots માં કયો પ્રકાર દેખાડવો તે પસંદ કરી શકાય છે.

sns.pairplot(
    data=df,
    vars=["Python", "SQL", "Maths"],
    hue="Department",
    diag_kind="hist" #diagonal પર histograms
)

plt.show()

# diag_kind="hist" — દરેક feature નું histogram.
# diag_kind="kde" — દરેક feature નું smooth density estimate.
```

```python
# 8. Practical 6 — Iris Dataset પર Industry-Style Analysis
# Iris dataset Machine Learning માં ખૂબ જાણીતો sample dataset છે. તેમાં ફૂલના ચાર numerical measurements અને species ની category હોય છે

iris = sns.load_dataset("iris")

print(iris.head())
print(iris.info())
```

```python
sns.pairplot(
    data=iris,
    hue="species",
    vars=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ],
    diag_kind="hist"
)

plt.show()

# 9. મહત્વપૂર્ણ Parameters — Quick Reference
# Parameter	ઉપયોગ
# data	DataFrame આપવો
# vars	Plot કરવા માટે numerical columns પસંદ કરવાં
# hue	Category પ્રમાણે અલગ રંગ દર્શાવવા
# kind	Off-diagonal plots નો પ્રકાર
# diag_kind	Diagonal પર histogram અથવા KDE પસંદ કરવું
# corner=True	Duplicate upper-triangle plots છુપાવવા
# height	દરેક subplot ની ઊંચાઈ નિયંત્રિત કરવી
# plot_kws	Scatter અથવા regression plots ની styling નિયંત્રિત કરવી
```

```python
sns.pairplot(
    data=iris,
    hue="species",
    corner=True,
    height=2.5
)

plt.show()
```
