```python
# Session 10.8 — Relational Plots (Scatter, Line, Regressions)
# આજે આપણે Seaborn માં Relational Plots શીખીશું. તેનો મુખ્ય હેતુ બે કે તેથી વધુ variables વચ્ચેનો સંબંધ graph દ્વારા સમજવાનો છે.

# ઉદાહરણ તરીકે:
# - Study hours વધે તો વિદ્યાર્થીના marks માં શું ફેરફાર થાય છે?
# - દર મહિને Sales વધે છે કે ઘટે છે?
# - અલગ-अलग departments ના results કેવી રીતે સરખાવી શકાય?
# - Product price અને sales quantity વચ્ચે કેવો સંબંધ છે?
# આ session માં આપણે ત્રણ મુખ્ય functions શીખીશું:
# - sns.scatterplot() — બે numerical variables વચ્ચેનો સંબંધ.
# - sns.lineplot() — values માં ક્રમ પ્રમાણે થતા ફેરફાર અથવા trend.
# - sns.relplot() — relational plots બનાવવા માટેનું figure-level interface, જેમાં scatter અથવા line plot અને અલગ panels બનાવી શકાય છે.
```

```python
# 1. Setup અને Dataset
# VS Code માં Jupyter Notebook ખોલો. જરૂરી libraries import કરો:

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

data = {
    "Student": [
        "Amit", "Neha", "Ravi", "Pooja", "Karan",
        "Meera", "Jay", "Priya", "Rahul", "Nisha",
        "Dev", "Riya"
    ],
    "Department": [
        "IT", "IT", "Computer", "Computer",
        "IT", "Computer", "IT", "Computer",
        "IT", "Computer", "IT", "Computer"
    ],
    "StudyHours": [2, 7, 3, 6, 2, 8, 4, 5, 6, 5, 3, 9],
    "Python": [65, 92, 70, 88, 68, 95, 78, 82, 90, 86, 73, 96],
    "SQL": [68, 88, 75, 90, 70, 92, 80, 85, 84, 89, 72, 94]
}

df = pd.DataFrame(data)

print(df.head())
```

```python
# 2. sns.scatterplot() — Scatter Plot
# Scatter plot માં દરેક observation ને એક point તરીકે બતાવવામાં આવે છે. બે numerical variables વચ્ચેનો pattern જોવા માટે આ સૌથી સામાન્ય charts પૈકીનો એક છે.

# name chart bane che
sns.scatterplot(
    data=df,
    x="StudyHours", # x axis
    y="Python" # y axios
)

plt.title("Study Hours vs Python Marks")
plt.xlabel("Study Hours")
plt.ylabel("Python Marks")
plt.show()
```

```python
# 3. hue, style અને size parameters
# Scatter plot માં વધારાના variables ને રંગ, marker style અથવા point size દ્વારા દર્શાવી શકાય છે.

sns.scatterplot(
    data=df,
    x="StudyHours",
    y="Python",
    hue="Department", # color change
    style="Department", # style marker badlay che
    size="SQL",
    sizes=(50, 200) # 
)

plt.title("Student Performance by Department")
plt.show()
```

```python
# 4. sns.lineplot() — Line Plot
# Line plot સામાન્ય રીતે time અથવા કોઈ meaningful sequence પ્રમાણે values કેવી રીતે બદલાય છે તે બતાવવા માટે ઉપયોગી છે.

# ઉદાહરણ: જાન્યુઆરીથી જૂન સુધી company sales કેવી રીતે બદલાઈ.

sales_data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 14000, 18000, 22000, 25000]
}

sales_df = pd.DataFrame(sales_data)

sns.lineplot(
    data=sales_df,
    x="Month",
    y="Sales",
    marker="o" # o symbols batave che
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
# Multiple groups સાથે Line Plot

monthly_data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"] * 2,
    "Department": ["IT"] * 6 + ["Computer"] * 6,
    "Sales": [
        12000, 15000, 14000, 18000, 22000, 25000,
        10000, 13000, 16000, 17000, 19000, 23000
    ]
}

monthly_df = pd.DataFrame(monthly_data)

sns.lineplot(
    data=monthly_df,
    x="Month",
    y="Sales",
    hue="Department",
    marker="o"
)

plt.title("Monthly Sales by Department")
plt.show()
```

```python
# # 5. sns.relplot() — Figure-level Relational Plot
# # relplot() એ figure-level function છે. તેનો ઉપયોગ scatter અથવા line plot બનાવવા અને જરૂર પડે ત્યારે data ને અલગ panels માં વહેંચવા માટે થાય છે.
# Function	પ્રકાર	ઉપયોગ
# sns.scatterplot()	Axes-level	એક axes પર scatter plot
# sns.lineplot()	Axes-level	એક axes પર line plot
# sns.relplot()	Figure-level	Scatter/line plots અને અલગ panels

```

```python
# Example 1: relplot() સાથે Scatter
sns.relplot(
    data=df,
    x="StudyHours",
    y="Python",
    hue="Department",
    kind="scatter",
    height=5,
    aspect=1.3
)

plt.show()
```

```python
# Example 2: relplot() સાથે Line
sns.relplot(
    data=monthly_df,
    x="Month",
    y="Sales",
    hue="Department",
    kind="line", # line
    marker="o", #symbols
    height=5,
    aspect=1.4
)

plt.show()
```

```python
# 6. col અને row — અલગ panels બનાવવા
# ધારો કે તમે દરેક department માટે અલગ chart જોવા માંગો છો. relplot() માં col અથવા row નો ઉપયોગ કરી શકો છો.
sns.relplot(
    data=df,
    x="StudyHours",
    y="Python",
    col="Department",
    kind="scatter",
    height=4,
    aspect=1
)

plt.show()

# આ code IT અને Computer માટે અલગ panels બનાવશે.
# - col="Department" — દરેક department માટે અલગ column panel.
# - row="Department" — દરેક department માટે અલગ row panel.
# - hue="Department" — એક જ panel માં department પ્રમાણે રંગ.
```
