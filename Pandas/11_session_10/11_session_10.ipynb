```python
# Session 8.10 — Pandas Merging & Joining DataFrames
# આજે આપણે શીખીશું કે બે અથવા વધુ DataFrames ને એકસાથે કેવી રીતે જોડવા. Data Science માં આ ખૂબ જ મહત્વનું Topic છે, કારણ કે વાસ્તવિક ડેટા ઘણી વખત અલગ-અલગ Tables અથવા Files માં હોય છે.

# ઉદાહરણ તરીકે, કોલેજમાં Student ની માહિતી એક Excel Sheet માં છે અને તેના Marks બીજી Sheet માં છે. બંનેને Student ID ના આધારે જોડીને Final Result તૈયાર કરી શકાય છે.
# આજના Learning Objectives
#  pd.merge() નો ઉપયોગ

#  Inner, Left, Right અને Outer Join

#  Common Key દ્વારા DataFrames જોડવા

#  join() અને Multiple DataFrames

#  Student Marks Report નો Practical Project

# 1. Merging અને Joining શું છે?
# English definition: Merging and joining are techniques used to combine data from two or more DataFrames using common columns or indexes.

# ગુજરાતીમાં: બે અથવા વધુ DataFrames માં રહેલી સંબંધિત માહિતી Common Column અથવા Index ના આધારે એક DataFrame માં જોડવાની પ્રક્રિયાને Merging અથવા Joining કહે છે.

# ઉદાહરણ:

# Students DataFrame
# StudentID	Name
# 101	        Amit
# 102	        Neha
# 103	        Ravi

# Marks DataFrame
# StudentID	Marks
# 101	        85
# 102 	    92
# 104 	    78

# બંને DataFrames માં StudentID Common Column છે. તેને આધારે આપણે Name અને Marks એક જ Table માં લાવી શકીએ છીએ.
```

```python
# 2. Practical Dataset બનાવીએ
# VS Code માં તમારી Jupyter Notebook ખોલો અને આ Code Run કરો.

import pandas as pd

students = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Name": ["Amit", "Neha", "Ravi"],
    "City": ["Ahmedabad", "Surat", "Rajkot"]
})

marks = pd.DataFrame({
    "StudentID": [101, 102, 104],
    "Marks": [85, 92, 78]
})

print("Students DataFrame:")
print(students)

print("\nMarks DataFrame:")
print(marks)

# અહીં:
# - students માં ત્રણ વિદ્યાર્થીઓની વિગતો છે.
# - marks માં ત્રણ Student ID ના Marks છે.
# - Student ID 103 ના Marks ઉપલબ્ધ નથી.
# - Student ID 104 નું નામ અને City ઉપલબ્ધ નથી.
# આ પરિસ્થિતિથી આપણે અલગ-અલગ Join સમજશું.
```

```python
# 3. pd.merge() — બે DataFrames જોડવા
# સૌથી પહેલાં સામાન્ય Merge કરીએ.

result = pd.merge(
    students,
    marks,
    on="StudentID"
)
print(result)


# અહીં ફક્ત એવા Student ID આવ્યા છે જે બંને DataFrames માં હાજર છે.

# pd.merge() ના મહત્વના Parameters:
# Parameter	ઉપયોગ
# left	પ્રથમ DataFrame
# right	બીજો DataFrame
# on	બંનેમાં Common Column
# how	કયા પ્રકારનો Join કરવો
# suffixes	સમાન નામના Columns માટે અલગ Suffix
# left_on	પ્રથમ DataFrame માં Join Column
# right_on	બીજા DataFrame માં Join Column

# યાદ રાખો: pd.merge(students, marks, on="StudentID") માં Default how="inner" હોય છે.
```

```python
# 4. Inner Join — બંનેમાં Common Records
# Inner Join ફક્ત એવા Records આપે છે જેમની Join Key બંને DataFrames માં મળે.

inner_result = pd.merge(
    students,
    marks,
    on="StudentID",
    how="inner"
)

print(inner_result)

# ફક્ત બંને Tables માં હાજર Common Keys જ Result માં આવશે.
```

```python
# 5. Left Join — ડાબી બાજુના બધા Records
# Left Join માં પ્રથમ DataFrame એટલે કે Left Table ના બધા Records રહે છે. બીજી Table માં Match ન મળે તો સંબંધિત Columns માં NaN આવે છે.

left_result = pd.merge(

    students,
    marks,
    on="StudentID",
    how="left"
)

print(left_result)

# Student 103 ની માહિતી જળવાઈ રહે છે, પરંતુ Marks મળતા નથી.

# Real-world ઉપયોગ: કોલેજના બધા વિદ્યાર્થીઓની યાદી રાખવી હોય, ભલે કેટલાકના Marks હજુ દાખલ ન થયા હોય.
```

```python
# 6. Right Join — જમણી બાજુના બધા Records
# Right Join માં બીજી DataFrame એટલે કે Right Table ના બધા Records રહે છે.

right_result =  pd.merge(
    students,
    marks,
    on="StudentID",
    how="right"

)

print(right_result)

# Student 104 ના Marks છે, પરંતુ Students Table માં તેની વિગતો નથી.
```

```python
# 7. Outer Join — બંને Tables ના બધા Records
# Outer Join બંને DataFrames ના બધા Keys રાખે છે. Match મળે ત્યાં વિગતો જોડાય છે, ન મળે ત્યાં Missing Values આવે છે.

outer_result = pd.merge(
    students,
    marks,
    on="StudentID",
    how="outer"
)

print(outer_result)

# આથી કોઈપણ Table માં રહેલો Student ID ગુમાતો નથી.
```

```python
# ચારેય Joins વચ્ચેનો તફાવત
# Join Type	Result માં કયા Records રહેશે?
# inner	બંનેમાં Match થતા Records
# left	Left Table ના બધા Records
# right	Right Table ના બધા Records
# outer	બંને Tables ના બધા Records

# યાદ રાખવાની રીત: Inner = Common, Left = Left All, Right = Right All, Outer = Both All.

```

```python
# 8. જ્યારે Common Column ના નામ અલગ હોય
# ક્યારેક એક DataFrame માં StudentID અને બીજીમાં ID નામ હોય છે. ત્યારે on ની જગ્યાએ left_on અને right_on વાપરી શકાય.

student_info = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Name": ["Amit", "Neha", "Ravi"]
})

exam_info = pd.DataFrame({
    "ID": [101, 102, 103],
    "Marks": [85, 92, 76]
})

result = pd.merge(
    student_info,
    exam_info,
    left_on="StudentID",
    right_on="ID",
    how="inner"
)

print(result)

# અહીં StudentID અને ID એક જ પ્રકારની ઓળખ દર્શાવે છે, પરંતુ Column ના નામ અલગ છે.

# જો પરિણામમાં બંને ID Columns ની જરૂર ન હોય તો વધારાનો Column કાઢી શકાય
result = result.drop(columns=["ID"])

print(result)
```

```python
# 9. સમાન નામના Columns — suffixes
# ધારો કે બંને DataFrames માં City નામનો Column છે. Merge કર્યા પછી Pandas તેમને અલગ ઓળખ આપવા માટે Suffix લગાવે છે.

students_city = pd.DataFrame({
    "StudentID": [101, 102],
    "City": ["Ahmedabad", "Surat"]
})

exam_city = pd.DataFrame({
    "StudentID": [101, 102],
    "City": ["Ahmedabad Center", "Surat Center"]
})

result = pd.merge(
    students_city,
    exam_city,
    on="StudentID",
    suffixes=("_student", "_exam")
)

print(result)
```

```python
# 10. DataFrame.join() — Index ના આધારે જોડવું
# merge() સામાન્ય રીતે Columns દ્વારા જોડવા માટે વપરાય છે. join() ખાસ કરીને Index આધારિત Joining માટે અનુકૂળ છે.

student_info = pd.DataFrame({
    "Name": ["Amit", "Neha", "Ravi"]
}, index=[101, 102, 103])

student_marks = pd.DataFrame({
    "Marks": [85, 92, 76]
}, index=[101, 102, 103])

result = student_info.join(student_marks)

print(result)

# અહીં Index માં Student ID રાખવામાં આવ્યો છે. તેથી join() બંને DataFrames ના Index ના આધારે વિગતો જોડે છે
```

```python
# 11. બે કરતાં વધુ DataFrames Merge કરવા
# વાસ્તવિક Project માં Student Details, Marks અને Attendance અલગ Tables માં હોઈ શકે છે.

students = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Name": ["Amit", "Neha", "Ravi"]
})

marks = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Marks": [85, 92, 76]
})

attendance = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Attendance": [90, 95, 80]
})

final_report = pd.merge(
    students,
    marks,
    on="StudentID",
    how="inner"
)

final_report = pd.merge(
    final_report,
    attendance,
    on="StudentID",
    how="inner"
)

print(final_report)
```

```python
# 12. મહત્વની બાબત — Duplicate Keys થી સાવચેત રહો
# જો એક જ Student ID કોઈ DataFrame માં એક કરતાં વધુ વાર આવે, તો Merge કર્યા પછી એકથી વધુ Rows બની શકે છે.

students = pd.DataFrame({
    "StudentID": [101, 102],
    "Name": ["Amit", "Neha"]
})

marks = pd.DataFrame({
    "StudentID": [101, 101, 102],
    "Subject": ["Python", "SQL", "Python"],
    "Marks": [85, 90, 92]
})

result = pd.merge(
    students,
    marks,
    on="StudentID",
    how="inner"
)

print(result)

result = pd.merge(
    students,
    marks,
    on="StudentID",
    validate="one_to_many"
)

print(result)
```
