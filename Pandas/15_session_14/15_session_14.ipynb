```python
# 3. Step 2 — Sample dataset બનાવો
# આપણે 10 records બનાવશું. તેમાં જાણીજોઈને extra spaces, uppercase/lowercase, missing values, invalid date અને એક duplicate record રાખ્યો છે, જેથી cleaning ની practice થઈ શકે.

import pandas as pd
import numpy as np

data = {
    "StudentID": [101, 102, 103, 104, 105,
                  106, 107, 108, 109, 108],

    "Name": [
        " Amit Shah ",
        "NEHA PATEL",
        "Ravi Joshi",
        "pooja mehta",
        "KARAN DESAI",
        "Mira Shah",
        "Dev Patel",
        "Isha Joshi",
        None,
        "Isha Joshi"
    ],

    "Department": [
        "IT", "CS", "IT", "CS", "IT",
        "CS", "IT", "CS", "IT", "CS"
    ],

    "StudyHours": [2, 4, 3, 6, None, 5, 1, 7, 4, 7],

    "Attendance": [70, 85, 75, 95, 80, None, 60, 98, 88, 98],

    "Marks": [55, 75, 65, 90, 60, 85, 35, 96, 72, 96],

    "AdmissionDate": [
        "2025-06-01",
        "2025-06-02",
        "2025-06-03",
        "2025-06-04",
        "2025-06-05",
        "2025-06-06",
        "2025-06-07",
        "2025-06-08",
        "wrong-date",
        "2025-06-08"
    ]
}

df = pd.DataFrame(data)
df
print(df.head())
print("Dataset Shape:", df.shape)

# સમજણ:
# - pd.DataFrame(data) dictionary ને table માં ફેરવે છે.
# - head() પ્રથમ પાંચ records બતાવે છે.
# - shape માં (rows, columns) મળે છે.
```

```python
# 4. Step 3 — Data ને inspect કરો (EDA)
# Data clean કરતાં પહેલાં તેની સ્થિતિ સમજવી જરૂરી છે.

# first 5 records 
display(df.head())

# last 5 records
display(df.tail())

# cloumns , data types and non-null values
df.info()

# numerical columns no statistical summary
display(df.describe())

# mising values ni count 
print(df.isnull().sum())
```

```python
# 5. Step 4 — Data Cleaning
# 5.1 Duplicate records દૂર કરો

print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates().copy()

print("Shape after removing duplicates:", df.shape)
```

```python
print(df["StudentID"].duplicated().sum())
```

```python
# 5.2 Names અને Department સાફ કરો

df["Name"] = (
    df["Name"]
    .astype("string")
    .str.strip()
    .str.title()
)

df["Department"] = (
    df["Department"]
    .astype("string")
    .str.strip()
    .str.upper()
)

display(df[["StudentID", "Name", "Department"]])
```
