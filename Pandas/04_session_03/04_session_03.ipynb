```python
# 3. Practical માટે dataset બનાવો
# આપણે એક જ dataset નો ઉપયોગ આખા session દરમિયાન કરીશું.

import pandas as pd

data = {
   "StudentID": [101, 102, 103, 104, 105, 106, 107],
    "Name": ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira", "Jay"],
    "Department": ["IT", "CS", "IT", "CS", "IT", "CS", "IT"],
    "Marks": [85, 92, 76, 88, 67, 95, 80],
    "Attendance": [90, 95, 80, 85, 70, 98, 75],
    "City": [
        "Ahmedabad", "Surat", "Rajkot", "Ahmedabad",
        "Surat", "Rajkot", "Ahmedabad"
    ]
}

df = pd.DataFrame(data)

# અહીં:
# - 7 rows એટલે 7 વિદ્યાર્થીઓ.
# - 6 columns એટલે દરેક વિદ્યાર્થીની 6 પ્રકારની માહિતી.
# - Index 0 થી 6 છે.
# - StudentID એક column છે; તે હાલ DataFrame નો index નથી.

print(df)
```

```python
# 5. loc[] — Label આધારિત Selection
# English Definition: loc[] is a label-based indexer used to select rows and columns by their index labels and column names.

# ગુજરાતી વ્યાખ્યા: loc[] વડે DataFrame માંથી row ના index label અને column ના નામનો ઉપયોગ કરીને data પસંદ કરી શકાય છે.

print(df.loc[2]) # loction ma dataframe index row number ae full defatils malse
```

```python
# Practical 2 — ચોક્કસ row અને column પસંદ કરો

# df.loc[row_label,column_label] stntax

print(df.loc[2,"Name"]) # spacify name
print(df.loc[2,"Marks"]) # spacify marks
```

```python
# Practical 3 — અનેક rows અને columns

print(df.loc[0:3,["Name","Marks"]]) # slicing and and colum name accses
```

```python
# Practical 4 — Boolean condition સાથે loc

print(df.loc[df["Marks"] >= 85,["Name","Marks"]])

# અહીં પહેલાં condition તપાસાય છે અને પછી ફક્ત પસંદ કરેલા columns બતાવાય છે.
```

```python
# 6. iloc[] — Position આધારિત Selection
# English Definition: iloc[] is an integer-position-based indexer used to select rows and columns by their numerical positions.

# ગુજરાતી વ્યાખ્યા: iloc[] વડે row અને column ની position એટલે કે ક્રમાંકનો ઉપયોગ કરીને data પસંદ કરી શકાય છે.

# loc માં labels અને iloc માં positions મહત્વપૂર્ણ છે.

print(df.iloc[0])
```

```python
# 7. Boolean Filtering — Condition પ્રમાણે records મેળવવા
# English Definition: Boolean filtering selects rows based on a condition that evaluates to True or False.

# ગુજરાતી વ્યાખ્યા: દરેક row માટે condition સાચી છે કે ખોટી તે તપાસીને ફક્ત સાચી condition ધરાવતી rows પસંદ કરવાની પ્રક્રિયાને Boolean Filtering કહે છે.

# Practical 8 — 80 કે તેથી વધુ marks ધરાવતા વિદ્યાર્થીઓ

result = df[df["Marks"] > 80]
print(result)
```

```python
# Practical 9 — 80 કરતાં વધુ marks

print(df[df["Marks"] >= 80][["Name","Marks"]]) # and colum joei ye aetle male
```

```python
# Practical 10 — Text condition 
#only It deparament 

print(df[df["Department"]== "IT"])
```

```python
# only ahemedabad city na
print(df[df["City"] == "Ahmedabad"])
```

```python
# 8. Multiple Conditions — AND, OR, NOT
# Pandas માં એક કરતાં વધુ conditions વાપરી શકાય છે.

# | Operator | અર્થ | Pandas માં |
# |---|---|---|
# | AND | બંને conditions સાચી હોવી જોઈએ | & |
# | OR | ઓછામાં ઓછી એક condition સાચી હોવી જોઈએ | | |
# | NOT | condition નું ઉલટું | ~ |

# Pandas માં multiple conditions જોડતી વખતે દરેક condition ને parentheses () માં લખો.
# Practical 11 — AND condition
# Question: IT વિભાગના અને 80 કે તેથી વધુ marks ધરાવતા વિદ્યાર્થીઓ શોધો.

result = df[
    (df["Department"] == "IT") &
    (df["Marks"] >= 80)
]

print(result[["Name","Department","Marks"]])

# બંને conditions સાચી હોય ત્યારે જ row સામેલ થાય છે.
```

```python
# Practical 12 — OR condition
# Question: IT વિભાગના અથવા 90 કે તેથી વધુ marks ધરાવતા વિદ્યાર્થીઓ શોધો.

result = df[
    (df["Department"] == "IT") | 
    (df["Marks"] >= 90)
]

print(result[["Name","Department","Marks"]])
```

```python
# Practical 13 — NOT condition
# Question: IT વિભાગ સિવાયના વિદ્યાર્થીઓ પસંદ કરો.


result = df[~(df["Department"] == "IT")]

print(result[["Name","Department"]])
```

```python
# 9. isin() અને between() — સરળ filtering
# આ બંને methods વાસ્તવિક dataset માં ઘણી વાર ઉપયોગી થાય છે.
# isin() — અનેક values માંથી પસંદગી
# English Definition: isin() checks whether each value belongs to a specified collection of values.

# ગુજરાતી સમજણ: કોઈ column માં આપેલી અનેક values માંથી એક value હોય તો તે row પસંદ કરવા isin() ઉપયોગી છે.

result = df[df["Department"].isin(["IT", "CS"])]

print(result[["Name", "Department"]])
```

```python
result = df[df["City"].isin(["Surat", "Rajkot"])]

print(result[["Name", "City"]])
```

```python
# between() — Range પ્રમાણે filtering
# English Definition: between() identifies values that fall within a specified range.

# ગુજરાતી સમજણ: કોઈ value નિર્ધારિત minimum અને maximum ની વચ્ચે છે કે નહીં તે તપાસવા between() વપરાય છે. Default રીતે બંને endpoints સામેલ થાય છે.

result = df[df["Marks"].between(75, 90)]

print(result[["Name", "Marks"]])
```

```python
# 10. query() — Condition ને સરળ રીતે લખવી
# English Definition: DataFrame.query() selects rows using a query expression evaluated against the DataFrame's column names.

# ગુજરાતી વ્યાખ્યા: DataFrame ના column names નો ઉપયોગ કરીને readable expression લખી rows filter કરવા query() વપરાય છે.

# Practical 14 — Marks પ્રમાણે filtering

result = df.query("Marks >= 85")

print(result[["Name", "Marks"]])

# આ પહેલાંના Boolean filtering જેવા જ records આપે છે.
```

```python
# Practical 15 — Multiple conditions

result = df.query(
    "Department == 'IT' and Marks >= 80"
)

print(result[["Name", "Department", "Marks"]])
```

```python
# Practical 16 — Variables નો ઉપયોગ
# ધારો કે minimum marks વારંવાર બદલવા પડે છે.
minimum_marks = 85

result = df.query("Marks >= @minimum_marks")

print(result[["Name", "Marks"]])

# ક્યારે query() વાપરવું?
# - Condition readable રાખવી હોય.
# - સરળ filtering expressions લખવા.
# - Python variable ને expression માં વાપરવો હોય.
```

```python
# 11. Real-World Case Study — College Result Analysis
# તમારી class માં વિદ્યાર્થીઓને આ business questions આપો:
# 1. 90 કે તેથી વધુ marks ધરાવતા વિદ્યાર્થીઓ કોણ છે?
# 2. IT વિભાગના અને attendance 80 કે તેથી વધુ ધરાવતા વિદ્યાર્થીઓ કોણ છે?
# 3. Ahmedabad માં રહેતા વિદ્યાર્થીઓનાં નામ અને marks શું છે?
# 4. 75 થી 90 marks વચ્ચેના વિદ્યાર્થીઓ કોણ છે?
# 5. CS વિભાગ સિવાયના વિદ્યાર્થીઓ કોણ છે?
# નીચેનો code આ પ્રશ્નોના જવાબ આપે છે.

# Q1: Marks >= 90
q1 = df[df["Marks"] >= 90]
print(q1[["Name", "Marks"]])

# Q2: IT and Attendance >= 80
q2 = df[
    (df["Department"] == "IT") &
    (df["Attendance"] >= 80)
]
print(q2[["Name", "Attendance"]])

# Q3: Ahmedabad students
q3 = df.loc[
    df["City"] == "Ahmedabad",
    ["Name", "Marks"]
]
print(q3)

# Q4: Marks between 75 and 90
q4 = df[df["Marks"].between(75, 90)]
print(q4[["Name", "Marks"]])

# Q5: Students outside CS
q5 = df.query("Department != 'CS'")
print(q5[["Name", "Department"]])
```
