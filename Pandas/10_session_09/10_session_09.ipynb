```python

# # Session 8.9 — Pandas માં Date-Time Handling
# # આજે આપણે શીખીશું કે Pandas માં તારીખો અને સમય સાથે કેવી રીતે કામ કરવું. Data Science માં આ ખૂબ ઉપયોગી છે, કારણ કે Sales Report, Student Admission, Attendance, Banking Transactions અને Website Analytics જેવા ડેટામાં તારીખોનો ઉપયોગ વારંવાર થાય છે.

# આજના Learning Objectives
#  String ને Date-Time માં Convert કરવું

#  Year, Month, Day અને Weekday કાઢવા

#  Date Range મુજબ Data Filter કરવો

#  NaT, pd.date_range() અને Time Difference સમજવું

#  Month-wise Sales Analysis કરવું

# 1. Date-Time Handling શું છે?
# English definition: Date-Time Handling is the process of converting, extracting, filtering, and analyzing dates and times in a dataset.

# ગુજરાતીમાં: Dataset માં રહેલી તારીખોને યોગ્ય Date-Time format માં ફેરવવી, તેમાંથી Year/Month/Day કાઢવા, તારીખ પ્રમાણે ડેટા Filter કરવો અને સમય આધારિત Analysis કરવું તેને Date-Time Handling કહે છે.

# ઉદાહરણ તરીકે, Online Shopping માંથી આપણે જાણી શકીએ કે:
# - જાન્યુઆરી મહિનામાં કેટલી Sales થઈ?
# - કઈ તારીખે સૌથી વધુ Orders આવ્યા?
# - છેલ્લા 30 દિવસમાં કેટલું Revenue થયું?
# - કયા દિવસે સૌથી વધુ Orders આવ્યા?
# 2. Practical Dataset બનાવીએ
# VS Code માં તમારી Jupyter Notebook ખોલો અને નીચેનો કોડ Run કરો.

import pandas as pd
import numpy as np

data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005, 1006],
    "Customer": ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira"],
    "OrderDate": [
        "2026-01-05",
        "2026-01-15",
        "2026-02-10",
        "not available",
        "2026-03-20",
        None
    ],
    "Amount": [1200, 2500, 800, 1800, 3200, 1500]
}

df = pd.DataFrame(data)

print(df)
```

```python
# અહીં OrderDate માં બે પ્રકારની સમસ્યાઓ છે:
# - "not available" — ખોટી તારીખની Value.
# - None — Missing Value.
# આગળ આપણે આ બંનેને Handle કરીશું.
# 3. pd.to_datetime() — String ને Date-Time માં Convert કરવું
# ઘણી વખત CSV અથવા Excel માં તારીખો Text/String તરીકે આવે છે. તારીખ આધારિત Analysis કરતાં પહેલાં તેને Date-Time format માં Convert કરવી જરૂરી છે.

df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)

print(df)
print(df["OrderDate"].dtype) 
```

```python
# Code	અર્થ
# pd.to_datetime()	Value ને Date-Time માં Convert કરે છે
# errors="coerce"	Invalid તારીખને NaT બનાવે છે
# dtype	Column નો Data Type બતાવે છે

# NaT નો અર્થ Not a Time થાય છે. તે Date-Time Column માં Missing અથવા Invalid તારીખ દર્શાવે છે.

# Expected result માં OrderDate નો dtype સામાન્ય રીતે datetime64[ns] હશે, અને invalid/missing તારીખો NaT તરીકે દેખાશે.

# ખાસ ધ્યાન: Date-Time માં Convert કર્યા વગર તારીખોને String તરીકે Compare કરવાથી ખોટા પરિણામો મળી શકે છે

```

```python
# 4. તારીખમાંથી Year, Month અને Day કાઢવા
# Pandas માં .dt accessor નો ઉપયોગ Date-Time Column માંથી અલગ-અલગ ભાગ કાઢવા માટે થાય છે.

df["Year"] = df["OrderDate"].dt.year
df["Month"] = df["OrderDate"].dt.month
df["Day"] = df["OrderDate"].dt.day
print(df[["OrderDate","Year","Month","Day"]])

# ઉદાહરણ તરીકે, 2026-03-20 માટે:
# - Year → 2026
# - Month → 3
# - Day → 20
# NaT ધરાવતી Row માટે આ Values Missing રહેશે.
```

```python
# ઉપયોગી .dt properties અને methods
# Property / Method	ઉપયોગ	ઉદાહરણ
# .dt.year	વર્ષ કાઢે	2026
# .dt.month	મહિનો નંબર	3
# .dt.day	તારીખનો દિવસ	20
# .dt.day_name()	અઠવાડિયાનો દિવસ	Friday
# .dt.month_name()	મહિનાનું નામ	March
# .dt.dayofweek	Monday = 0 થી Sunday = 6	4
# .dt.quarter	વર્ષનો Quarter	1
# .dt.is_month_start	મહિનાનો પ્રથમ દિવસ છે?	True/False
# .dt.is_month_end	મહિનાનો છેલ્લો દિવસ છે?	True/False

# Practical:
df["DayName"] = df["OrderDate"].dt.day_name()
df["MonthName"] = df["OrderDate"].dt.month_name()
df["Quarter"] = df["OrderDate"].dt.quarter

print(
    df[["OrderDate", "DayName", "MonthName", "Quarter"]]
)
```

```python
# 5. Date પ્રમાણે Data Filter કરવો
# ધારો કે આપણે ફક્ત February 2026 પછીના Orders જોવા માંગીએ છીએ.
# A. ચોક્કસ તારીખ પછીના Orders

feb_orders = df[df["OrderDate"] >= "2026-02-01"]

print(feb_orders[["OrderID", "Customer", "OrderDate", "Amount"]])
```

```python
# B. Date Range ની વચ્ચેના Orders

date_orders = df[
    df["OrderDate"].between(
        "2026-01-01",
        "2026-03-31"
    )
]

print(date_orders)

# between() આપેલી શરૂઆતની અને અંતિમ તારીખ બંનેને સામેલ કરે છે.
```

```python
# C. તારીખ પ્રમાણે Sort કરવું
sorted_df = df.sort_values(
    by="OrderDate",
    ascending=True,
    na_position="last"
)

print(sorted_df)

# ascending=True — જૂની તારીખ પહેલાં.
# ascending=False — નવી તારીખ પહેલાં.
# na_position="last" — Missing તારીખ છેલ્લે.
```

```python
# 6. pd.date_range() — તારીખોની શ્રેણી બનાવવી
# કેટલીક વખત દરેક દિવસની તારીખ, દરેક અઠવાડિયાની તારીખ અથવા દરેક મહિનાની શરૂઆતની તારીખ બનાવવાની જરૂર પડે છે.
dates = pd.date_range(
    start="2026-01-01",
    end="2026-01-07",
    freq="D"
)

print(dates)
```

```python
# freq="D" એટલે Daily — દરરોજની તારીખ.

# અન્ય ઉપયોગી Frequency:
# Frequency	અર્થ
# "D"	દરરોજ
# "W"	દર અઠવાડિયે
# "MS"	દરેક મહિનાની શરૂઆત
# "ME"	દરેક મહિનાનો અંત, જો તમારી Pandas Version તેને Support કરતી હોય

# ઉદાહરણ — મહિનાની શરૂઆતની તારીખો:


month_dates = pd.date_range(
    start="2026-01-01",
    periods=6,
    freq="MS"
)

print(month_dates)
```

```python
# . બે તારીખો વચ્ચેનો તફાવત કાઢવો
# ધારો કે Order થયાની તારીખથી 1 એપ્રિલ 2026 સુધી કેટલા દિવસ થયા તે જાણવા માંગીએ છીએ.
reference_date = pd.Timestamp("2026-04-01")

df["DaysSinceOrder"] = (
    reference_date - df["OrderDate"]
).dt.days

print(
    df[["OrderID", "OrderDate", "DaysSinceOrder"]]
)

# આમાં:
# - pd.Timestamp() — ચોક્કસ તારીખ/સમયની Value બનાવે છે.
# - reference_date - df["OrderDate"] — બે તારીખો વચ્ચેનો તફાવત આપે છે.
# - .dt.days — તફાવતને દિવસોમાં બતાવે છે.
# Missing તારીખ હોય તો DaysSinceOrder પણ Missing આવશે. અહીં Reference Date ઉદાહરણ માટે નિશ્ચિત રાખી છે, જેથી દરેક વખતે Output સમાન રહે.
```

```python
# 8. Real-world Project — Month-wise Sales Analysis
# આ Data Science માં ખૂબ સામાન્ય કામ છે: દરેક મહિને કેટલું Revenue થયું તે શોધવું.

valid_df = df.dropna(subset=["OrderDate"]).copy()

monthly_sales = (
    valid_df
    .set_index("OrderDate")
    .resample("MS")["Amount"]
    .sum()
)

print(monthly_sales)
```
