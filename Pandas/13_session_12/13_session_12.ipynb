```python
# Session 8.12 — Pandas GroupBy & Aggregations
# આજે આપણે Data Science માં ખૂબ જ મહત્વપૂર્ણ Topic શીખીશું: ડેટાને Groups માં વહેંચીને તેનું Analysis કરવું.

# ઉદાહરણ તરીકે, કોલેજમાં દરેક Department માં કેટલા Students છે, દરેક Department ના સરેરાશ Marks કેટલા છે અને કયા Department ના સૌથી વધુ Marks છે — આ બધું groupby() દ્વારા શોધી શકાય છે.
# આજના Learning Objectives
#  groupby() થી Data Group કરવો

#  sum(), mean(), count(), min() અને max()

#  agg() વડે એકસાથે અનેક Calculations

#  value_counts() અને size() સમજવા

#  Department-wise અને Product-wise Analysis

# 1. GroupBy શું છે?
# English definition: GroupBy is a process of splitting data into groups based on one or more columns and applying calculations to each group.

# ગુજરાતીમાં: કોઈ એક અથવા વધુ Columns ની સમાન Values ધરાવતી Rows ને Group માં વહેંચીને દરેક Group પર અલગ Calculation કરવી તેને GroupBy કહે છે.

# ઉદાહરણ તરીકે, નીચેના Student Dataset માં Department પ્રમાણે Groups બનાવી શકીએ છીએ.
```

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira"],
    "Department": ["IT", "IT", "CS", "CS", "IT", "CS"],
    "Marks": [85, 90, 78, 92, 74, 88],
    "Fees": [50000, 50000, 45000, 45000, 50000, 45000]
}

df = pd.DataFrame(data)

print(df)
# આ DataFrame માં 6 Students છે અને બે Departments છે: IT અને CS.
```

```python
# 2. groupby() — Department પ્રમાણે Group કરવું

grouped = df.groupby("Department")

print(grouped["Marks"].mean())

# સમજીએ:
# - groupby("Department") — Department પ્રમાણે Groups બનાવે છે.
# - ["Marks"] — Marks Column પસંદ કરે છે.
# - .mean() — દરેક Group ના સરેરાશ Marks કાઢે છે.

# ગણતરી:
# - CS: (78 + 92 + 88) / 3 = 86
# - IT: (85 + 90 + 74) / 3 = 83
```

```python
# 3. મહત્વની Aggregation Methods
# Aggregation એટલે અનેક Values ને એક Summary Value માં ફેરવવી.
# Method	અર્થ	ઉદાહરણ
# sum()	કુલ સરવાળો	Department ના કુલ Fees
# mean()	સરેરાશ	સરેરાશ Marks
# count()	Missing Values સિવાયની Values ગણવી	કેટલા Marks દાખલ છે
# size()	Group માં કુલ Rows ગણવી	કુલ Student Records
# min()	સૌથી નાની Value	ઓછામાં ઓછા Marks
# max()	સૌથી મોટી Value	સૌથી વધુ Marks
# median()	મધ્યસ્થ Value	મધ્યસ્થ Fees
# A. કુલ Marks — sum()

print(df.groupby("Department")["Marks"].sum())
```

```python
# B. સરેરાશ Marks — mean()

print(df.groupby("Department")["Marks"].mean())
```

```python
# C. સૌથી ઓછા અને વધુ Marks
print("Min marks")
print(df.groupby("Department")["Marks"].min())
print("Max marks")
print(df.groupby("Department")["Marks"].max())
```

```python
# 4. count() અને size() વચ્ચેનો તફાવત
# આ બંને Methods ઘણી વાર ગૂંચવણ ઊભી કરે છે.

df.groupby("Department")["Marks"].count()
# આ દરેક Department માં Marks Column ની Missing Values સિવાયની Values ગણે છે.

df.groupby("Department").size()
# આ દરેક Department માં કુલ Rows ગણે છે, ભલે કોઈ Column માં Missing Value હોય.
```

```python
# આ તફાવત સમજવા માટે Missing Marks ધરાવતો Student ઉમેરો:

df_missing = pd.DataFrame({
    "Department": ["IT", "IT", "CS", "CS"],
    "Marks": [85, None, 90, 75]
})

print("Count:") #none nathi count thatu
print(df_missing.groupby("Department")["Marks"].count())

print("\nSize:") # size ma che
print(df_missing.groupby("Department").size())
```

```python
# 5. agg() — એકસાથે અનેક Calculations
# માનો કે દરેક Department માટે કુલ Marks, સરેરાશ Marks, સૌથી વધુ Marks અને Students ની સંખ્યા એકસાથે જોઈએ છે.

summary = df.groupby("Department").agg({
    "Marks": ["sum", "mean", "min", "max", "count"],
    "Fees": ["mean", "sum"]
})

print(summary)
```

```python
# આ એક જ Result માં અનેક Calculations આપે છે.

# પરંતુ Output માં Multi-level Column Names આવે છે, જેમ કે ("Marks", "mean"). વધુ વાંચી શકાય એવો Report બનાવવા Named Aggregation વાપરી શકાય છે.

summary = df.groupby("Department").agg(
    Total_Marks=("Marks", "sum"),
    Average_Marks=("Marks", "mean"),
    Highest_Marks=("Marks", "max"),
    Lowest_Marks=("Marks", "min"),
    Student_Count=("Name", "count"),
    Average_Fees=("Fees", "mean")
).reset_index()

print(summary)
```

```python
# 6. એક કરતાં વધુ Columns પ્રમાણે GroupBy
# ધારો કે Department અને Gender બંને પ્રમાણે Analysis કરવું છે.

students = pd.DataFrame({
    "Department": ["IT", "IT", "CS", "CS", "IT", "CS"],
    "Gender": ["Male", "Female", "Male", "Female", "Male", "Female"],
    "Marks": [85, 90, 78, 92, 74, 88]
})

result = students.groupby(
    ["Department", "Gender"]
)["Marks"].mean()

print(result)
```

```python
# જો તમારે Result ને સામાન્ય DataFrame તરીકે જોવો હોય તો:

result = students.groupby(
    ["Department", "Gender"],
    as_index=False
).agg(
    Average_Marks=("Marks", "mean"),
    Total_Students=("Marks", "count")
)

print(result)
```

```python
# 7. value_counts() — Category કેટલી વાર આવી?
# value_counts() એ Category ની દરેક Value કેટલી વાર આવી છે તે ઝડપથી ગણવા માટે ઉપયોગી છે.

print(df["Department"].value_counts())
```

```python
# ઉદાહરણ તરીકે, Dataset માં કેટલા Students IT અને કેટલા CS Department માં છે તે તરત જાણી શકાય છે.

# Percentages જોઈએ તો:

print(df["Department"].value_counts(normalize=True) * 100)
```

```python
# અહીં દરેક Department નો Dataset માં કેટલો ટકા હિસ્સો છે તે મળશે.

# value_counts() અને groupby().size() વચ્ચેનો સામાન્ય તફાવત:
# - value_counts() — એક Column ની Values કેટલી વાર આવે છે તે ગણવા માટે ઝડપી રીત.
# - groupby().size() — એક અથવા વધુ Columns દ્વારા બનેલા દરેક Group માં Rows ગણવા માટે.
```

```python
# 8. Real-world Project — Product-wise Sales Analysis
# હવે Student Data ને બદલે Online Store નું Dataset લઈએ.

sales = pd.DataFrame({
    "Product": [
        "Laptop", "Mobile", "Laptop",
        "Mobile", "Tablet", "Laptop",
        "Tablet", "Mobile"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics"
    ],
    "City": [
        "Ahmedabad", "Surat", "Surat",
        "Ahmedabad", "Rajkot", "Ahmedabad",
        "Surat", "Rajkot"
    ],
    "Quantity": [2, 5, 1, 3, 4, 1, 2, 6],
    "Revenue": [100000, 75000, 50000, 45000, 60000, 50000, 30000, 90000]
})

print(sales)
```

```python
# A. દરેક Product નું Total Revenue

product_sales = sales.groupby("Product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Average_Revenue=("Revenue", "mean")
).reset_index()

print(product_sales)
```

```python
# B. City પ્રમાણે Sales

city_sales = sales.groupby("City").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Quantity=("Quantity", "sum")
).reset_index()

print(city_sales)
```

```python
# C. સૌથી વધુ Revenue ધરાવતા Product ને શોધવો

product_sales = product_sales.sort_values(
    by="Total_Revenue",
    ascending=False
)

print(product_sales)
```

```python
# D. Result Excel માં Export કરવો

with pd.ExcelWriter("Sales_Analysis.xlsx") as writer:
    product_sales.to_excel(
        writer,
        sheet_name="Product Sales",
        index=False
    )

    city_sales.to_excel(
        writer,
        sheet_name="City Sales",
        index=False
    )

print("Excel Report Created")
```

```python
# 9. GroupBy નો આંતરિક Flow
# GroupBy ને ત્રણ મુખ્ય પગલાંમાં સમજવું સરળ છે.
# 1. Split

# Department પ્રમાણે Rows ને અલગ Groups માં વહેંચો.

# 2. Apply

# દરેક Group પર sum(), mean(), count() જેવી Calculation કરો.

# 3. Combine

# બધા Groups ના પરિણામને એક Summary માં જોડો.


# આને સામાન્ય રીતે Split–Apply–Combine Pattern કહેવામાં આવે છે.
# 10. સામાન્ય ભૂલો અને Tips
# સમસ્યા	ઉકેલ
# groupby() પછી ફક્ત Group Object દેખાય	તેની સાથે .sum(), .mean() જેવી Method વાપરો
# ખોટો Average મળે	Numeric Column અને Missing Values ચકાસો
# Group Column Index બની જાય	reset_index() અથવા as_index=False વાપરો
# count() અને Row Count અલગ આવે	યાદ રાખો કે count() Missing Values ગણતું નથી
# agg() પછી Column Names અઘરા લાગે	Named Aggregation વાપરો
# Revenue નો સરવાળો ખોટો લાગે	Duplicate Records અથવા Dataset ની Granularity તપાસો

# એક મહત્વની બાબત: દરેક Group નો Sum કાઢતાં પહેલાં ખાતરી કરો કે દરેક Row શું દર્શાવે છે. જો એક જ Order અનેક Rows માં હોય, તો સીધો Sum કરવાથી Revenue બે વાર ગણાઈ શકે છે.
```
