```python
# Session 8.8 — Pandas String Operations
# આજે આપણે Pandas માં Text Data એટલે કે String Values સાથે કામ કરવાનું શીખીશું. વાસ્તવિક Data Science Projects માં Names, Emails, Cities, Product Names અને Department જેવી માહિતીમાં ઘણી વખત Capital Letters, વધારાની Spaces અથવા ખોટી Spelling હોય છે. આ Data ને Clean અને Transform કરવા માટે Pandas String Operations ઉપયોગી છે.

# 8.8.1 — Text Case બદલવો
# str.lower(), str.upper(), str.title()

# 8.8.2 — Text Clean અને Replace કરવો
# str.strip(), str.replace()

# 8.8.3 — Text શોધવો અને Split કરવો
# str.contains(), str.split(), str.len()

# 1. String Operations એટલે શું?
# English Definition: String operations are methods used to manipulate, clean, search, and transform text data.

# ગુજરાતીમાં: Text Data ને બદલવા, સાફ કરવા, શોધવા અથવા અલગ ભાગોમાં વહેંચવા માટે ઉપયોગમાં લેવાતી પ્રક્રિયાઓને String Operations કહેવાય છે.

# ઉદાહરણો:
# - "SHUBHAM" ને "shubham" બનાવવું.
# - " Ahmedabad " માંથી વધારાની Spaces દૂર કરવી.
# - Email માંથી Domain અલગ કરવો.
# - "Information Technology" ધરાવતા Records શોધવા.
# - "Shubham Jadhav" ને First Name અને Last Name માં વહેંચવું.
```

```python
import pandas as pd
import numpy as np

data = {
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": [
        "  amit shah ",
        "NEHA PATEL",
        "Ravi Joshi",
        "pooja mehta",
        "KARAN DESAI",
        None
    ],
    "Email": [
        "amit@gmail.com",
        "NEHA@GMAIL.COM",
        "ravi@yahoo.com",
        "pooja@gmail.com",
        "karan@yahoo.com",
        None
    ],
    "City": [
        " ahmedabad ",
        "SURAT",
        "Rajkot",
        "Ahmedabad",
        " surat ",
        None
    ],
    "Department": [
        "Information Technology",
        "Computer Science",
        "information technology",
        "Computer Science",
        "IT",
        None
    ]
}

df = pd.DataFrame(data)
print(df)
df
```

```python
# 3. str.lower() — Text ને Lowercase કરવો
# બધા English Letters ને નાના અક્ષરોમાં ફેરવવા માટે આ Method વપરાય છે.

df["Name_Lower"] = df["Name"].str.lower()

print(df[["Name","Name_Lower"]]) # all letter lower case ma

# Real-world ઉપયોગ: Email Addresses ને એકસરખા Case માં લાવવા માટે ઉપયોગી છે.
# df["Email_Clean"] = df["Email"].str.lower()
```

```python
# 4. str.upper() — Text ને Uppercase કરવો
# બધા English Letters ને મોટા અક્ષરોમાં ફેરવવા માટે:

df["City_Upper"] = df["City"].str.upper()

print(df[["City", "City_Upper"]])
```

```python
# 5. str.title() — દરેક શબ્દનો પહેલો અક્ષર Capital કરવો

df["Name_Title"] = df["Name"].str.title()

print(df[["Name", "Name_Title"]])
```

```python
# 6. str.strip() — વધારાની Spaces દૂર કરવી
# Data Entry અથવા Excel માંથી Data Import કરતી વખતે શબ્દની શરૂઆત અથવા અંતે વધારાની Spaces હોઈ શકે છે.

df["City_Clean"] = df["City"].str.strip()

print(df[["City", "City_Clean"]])

# strip(), lstrip() અને rstrip() નો તફાવત
# Method	કાર્ય
# str.strip()	શરૂઆત અને અંતની Spaces દૂર કરે
# str.lstrip()	શરૂઆતની Spaces દૂર કરે
# str.rstrip()	અંતની Spaces દૂર કરે
```

```python
# 7. str.replace() — Text બદલવો
# ધારો કે Department Column માં "Information Technology" ને "IT" તરીકે લખવું છે.

df["Department_Clean"] = df["Department"].str.replace(
    "Information Technology",
    "IT",
    regex=False
)

print(df[["Department", "Department_Clean"]]) # capital and small deffrenec change nahi thay

df["Department_Clean"] = df["Department"].str.replace(
    "information technology",
    "IT",
    case=False,
    regex=False
)

print(df[["Department", "Department_Clean"]])
```

```python
# 8. str.contains() — Text શોધવો
# જો Department માં "Computer" શબ્દ ધરાવતા બધા Students શોધવા હોય તો

result = df[
    df["Department"].str.contains(
        "Computer",
        case=False,
        na=False
    )
]

print(result[["Name", "Department"]])

# સમજણ:
# - "Computer" શોધવાનો શબ્દ છે.
# - case=False એટલે Uppercase અને Lowercase નો ભેદ અવગણવો.
# - na=False એટલે Missing Values માટે False ગણવું.
# - પરિણામમાં માત્ર શરત સંતોષતી Rows આવશે.
```

```python
# Email માં Gmail વાપરતા Students શોધો

gmail_students = df[
    df["Email"].str.contains(
        "@gmail.com",
        case=False,
        na=False,
        regex=False
    )
]

print(gmail_students[["Name", "Email"]])

# અહીં regex=False રાખવાથી "." સહિતના Characters ને સામાન્ય Text તરીકે ગણવામાં આવે છે.

# વાસ્તવિક ઉપયોગ: Customer Database માં ચોક્કસ Email Domain, Product Name અથવા Department શોધવા માટે આ Method ઉપયોગી છે.
```

```python
# 9. str.split() — String ને ભાગોમાં વહેંચવી
# ધારો કે "Shubham Jadhav" માંથી First Name અને Last Name અલગ કરવા છે.

names = pd.Series(["Shubham Jadhav", "Amit Shah", "Neha Patel"])

split_names = names.str.split(" ",n=1)

print(split_names)
```

```python
# અલગ Columns બનાવો
df[["FirstName", "LastName"]] = (
    df["Name"]
    .str.strip()
    .str.split(r"\s+", n=1, expand=True)
)

print(df[["Name", "FirstName", "LastName"]])

# અહીં:
# - expand=True પરિણામને અલગ Columns માં ફેરવે છે.
# - n=1 એટલે વધુમાં વધુ એક Split થાય છે.
# - એક જ શબ્દ હોય તો બીજો ભાગ Missing હોઈ શકે છે.
# જો નામોમાં Middle Name હોય અથવા નામમાં અનેક ભાગ હોય તો Data પ્રમાણે વધુ મજબૂત Logic જરૂરી બની શકે છે.
```

```python
# 10. બીજા ઉપયોગી String Methods
# Method	શું કરે છે?
# str.len()	Text ની લંબાઈ ગણે છે
# str.startswith("A")	Text "A" થી શરૂ થાય છે કે નહીં તે તપાસે છે
# str.endswith(".com")	Text .com થી પૂરો થાય છે કે નહીં તે તપાસે છે
# str.contains("IT")	Text માં "IT" છે કે નહીં તે તપાસે છે
# str.split("@")	@ ના આધારે Text ને ભાગોમાં વહેંચે છે

# Name ની લંબાઈ
df["NameLength"] = df["Name"].str.len()

# Name A થી શરૂ થતું હોય તેવા Students
starts_with_a = df[
    df["Name"].str.strip().str.startswith("A", na=False)
]

# Email .com થી પૂરો થતો હોય તેવા Records
com_emails = df[
    df["Email"].str.endswith(".com", na=False)
]

print(df[["Name", "NameLength"]])
print(starts_with_a[["Name"]])
print(com_emails[["Email"]])
```

```python
# 11. Complete Practical — Student Data Cleaning
# હવે એક જ Pipeline માં અનેક String Operations નો ઉપયોગ કરીએ.

import pandas as pd

data = {
    "Name": ["  AMIT SHAH ", "neha patel", " Ravi  Joshi ", None],
    "Email": ["AMIT@GMAIL.COM", "neha@yahoo.com", "ravi@gmail.com", None],
    "City": [" AHMEDABAD ", "surat", "RAJKOT ", None],
    "Department": [
        "Information Technology",
        "Computer Science",
        "IT",
        None
    ]
}

df = pd.DataFrame(data)

# Dataset ની Copy બનાવો
clean_df = df.copy()

# Names Clean કરો
clean_df["Name"] = (
    clean_df["Name"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)

# Emails Clean કરો
clean_df["Email"] = clean_df["Email"].str.strip().str.lower()

# Cities Clean કરો
clean_df["City"] = (
    clean_df["City"]
    .str.strip()
    .str.title()
)

# Department Names Standardize કરો
clean_df["Department"] = clean_df["Department"].str.replace(
    "Information Technology",
    "IT",
    case=False,
    regex=False
)

clean_df["Department"] = clean_df["Department"].str.replace(
    "Computer Science",
    "CS",
    case=False,
    regex=False
)

print(clean_df)

# Excel માં Export કરો
clean_df.to_excel("cleaned_student_data.xlsx", index=False)
```
