```python
import pandas as pd 

data = {
    "StudentID":[101,102,103,104,105,106],
    "Name":["Amit","Neha","Ravi","Pooja","Karan","Mira"],
    "Department":["IT","CS","IT","CS","IT","CS"],
    "Marks":[85,92,76,88,67,95],
    "Attendance":[90,95,80,85,70,98]
}

df = pd.DataFrame(data)

print(df)
```

```python
# આ dataset માં 6 students અને 5 columns છે. હવે તેને CSV અને Excel બંને format માં save કરીશું, જેથી reading નો practical પણ જાતે કરી શકાય.

df.to_csv("students.csv",index=False) # index ma false kari ye aetle row default value na jay
df.to_excel("students.xlsx",index=False)

print("Files Created Successfully!")

# ndex=False નો અર્થ: Pandas નો default row index અલગ column તરીકે file માં save નહીં થાય.

# જો તમે file બનાવો છો, તો તે તમારા notebook ના current working directory માં save થશે.
```

```python
# 4. pd.read_csv() — CSV file વાંચવી
# English Definition: pd.read_csv() reads a CSV file and loads its contents into a Pandas DataFrame.

# ગુજરાતી સમજણ: તમારી પાસે students.csv file હોય, તો pd.read_csv() વડે તેમાં રહેલો data Python માં DataFrame તરીકે load કરી શકાય છે.

import pandas as pd 

df = pd.read_csv("students.csv")

print(df)
```

```python
# Code માં શું થયું?
# 1. pd.read_csv("students.csv") file વાંચે છે.
# 2. વાંચેલો data df નામના DataFrame માં રાખે છે.
# 3. print(df) સંપૂર્ણ DataFrame દર્શાવે છે.
# મહત્વના read_csv() parameters
# Parameter	ઉપયોગ
# filepath	File નું નામ અથવા path
# usecols	જરૂરી columns પસંદ કરવા
# nrows	    શરૂઆતની ચોક્કસ સંખ્યાની rows વાંચવા
# encoding	Text encoding પસંદ કરવા
# sep	    Values અલગ પાડતું delimiter
# index_col	કયા column ને index તરીકે વાપરવો

#only name and marks colums read karva

df = pd.read_csv("students.csv",usecols=["Name","Marks"])
print(df)
```

```python
# 5. pd.read_excel() — Excel file વાંચવી
# English Definition: pd.read_excel() reads spreadsheet data from an Excel file and returns a Pandas DataFrame.

# ગુજરાતી સમજણ: .xlsx અથવા અન્ય supported Excel file માંથી worksheet નો data Pandas DataFrame માં load કરવા માટે આ function વપરાય છે.

import pandas as pd 

df = pd.read_excel("students.xlsx")

print(df)
```

```python
# 6. head() — શરૂઆતની rows જોવી
# English Definition: head() returns the first few rows of a DataFrame.

# ગુજરાતી સમજણ: Dataset ની શરૂઆતની rows જોઈને data નો format અને columns સમજવા માટે head() વપરાય છે. Default રીતે તે પ્રથમ 5 rows બતાવે છે.

df = pd.read_csv("students.csv")

print(df.head())

# only 5 student read thase 

print(df.head(2)) # head ma number row check kari sakay
```

```python
# 7. tail() — છેલ્લી rows જોવી
# English Definition: tail() returns the last few rows of a DataFrame.

# ગુજરાતી સમજણ: Dataset ના અંતે કયો data છે તે જોવા tail() વપરાય છે. Default રીતે છેલ્લી 5 rows બતાવે છે.

print(df.tail())

# last ma thi chalu thay che 

print(df.tail(2))
```

```python
# 8. info() — Dataset ની રચના સમજવી
# English Definition: DataFrame.info() prints a concise summary of a DataFrame, including its columns, non-null counts, data types, and memory usage.

# ગુજરાતી સમજણ: Dataset માં કેટલા columns છે, દરેક column માં કેટલી non-missing values છે અને તેનો data type કયો છે તે જાણવા info() વપરાય છે.

df = pd.read_csv("students.csv")

df.info()

# કારણ કે info() પોતે summary print કરે છે અને return value None છે. તેથી print(df.info()) કરવાથી summary પછી None પણ દેખાય છે.
```

```python
# 9. describe() — Statistical summary
# English Definition: describe() generates descriptive statistics that summarize the central tendency, dispersion, and distribution of numerical data.

# ગુજરાતી સમજણ: Numerical columns માટે count, mean, standard deviation, minimum, quartiles અને maximum જેવી summary values મેળવવા describe() વપરાય છે.

df = pd.read_csv("students.csv")

print(df.describe())

# count: non-missing numerical values ની સંખ્યા.
# mean: સરેરાશ.
# std: standard deviation, એટલે values માં ફેલાવો કેટલો છે તેનો એક માપ.
# min: સૌથી નાની value.
# 25%: પ્રથમ quartile.
# 50%: median.
# 75%: ત્રીજું quartile.
# max: સૌથી મોટી value.
```

```python
print(df.describe(include="object"))

# unique: અલગ values કેટલી છે.
# top: સૌથી વધુ વાર આવતી value.
# freq: top value કેટલી વાર આવે છે.
```

```python
# 10. CSV માં Export કરવું — to_csv()
# English Definition: DataFrame.to_csv() writes a DataFrame to a CSV file.

# ગુજરાતી સમજણ: Pandas માં data clean અથવા modify કર્યા પછી તેને CSV file તરીકે save કરવા માટે to_csv() વપરાય છે.

import pandas as pd
df = pd.read_csv("students.csv")

#new csv file create karvi

df.to_csv("students_export.csv",index=False)

print("CSV export completed")

# with_index.csv માં DataFrame નો index પણ વધારાના column તરીકે save થશે.
# without_index.csv માં માત્ર મૂળ data columns save થશે
```

```python
# Export પહેલાં data filter કરો

df = pd.read_csv("students.csv")

high_scorers = df[df["Marks"] >= 85]

high_scorers.to_csv("high_scorers.csv",index=False)

print(high_scorers)

# અહીં high_scorers.csv માં માત્ર 85 કે તેથી વધુ marks ધરાવતા વિદ્યાર્થીઓ save થશે. જૂની students.csv file બદલાશે નહીં.
```

```python
# 11. Excel માં Export કરવું — to_excel()
# English Definition: DataFrame.to_excel() writes a DataFrame to an Excel worksheet.

# ગુજરાતી સમજણ: DataFrame ને .xlsx Excel file માં save કરવા માટે to_excel() વપરાય છે.

import pandas as pd 

df = pd.read_csv("students.csv")

df.to_excel("students_export.xlsx",index="False")

print("Excel exporst completed")
```

```python
# એક Excel file માં અનેક worksheets બનાવવી

with pd.ExcelWriter("college_report.xlsx") as writer :
    df.to_excel(writer,sheet_name="All Students",index=False)

    high_scorers = df[df["Marks"] >= 85]

    high_scorers.to_excel(
        writer,
        sheet_name="high_scorers",
        index=False
    )

#     આ એક જ Excel file માં બે sheets બનાવશે:
# 1. All Students
# 2. High Scorers
```

```python
# 12. એક જ practical માં બધા functions નો ઉપયોગ
# આ code તમારી notebook માં એકસાથે run કરો:

import pandas as pd

# 1. Read CSV
df = pd.read_csv("students.csv")

# 2. First rows
print("\n--- HEAD ---")
print(df.head(3))

# 3. Last rows
print("\n--- TAIL ---")
print(df.tail(2))

# 4. Structure
print("\n--- INFO ---")
df.info()

# 5. Statistics
print("\n--- DESCRIBE ---")
print(df.describe())

# 6. Export CSV
df.to_csv("students_final.csv", index=False)

# 7. Export Excel
df.to_excel("students_final.xlsx", index=False)

print("\nBoth exports completed!")
```
