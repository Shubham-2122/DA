```python
# Session 8.11 — Pandas Concatenation & Data Reshaping
# આજે આપણે શીખીશું કે અલગ-અલગ DataFrames ને કેવી રીતે જોડવા અને ડેટાનું Format કેવી રીતે બદલવું. Data Science માં આનો ઉપયોગ Sales Reports, Monthly Data, Survey Results અને Dashboard બનાવવા માટે થાય છે.
# આજના Learning Objectives
#  pd.concat() થી DataFrames જોડવા

#  axis=0 અને axis=1 સમજવા

#  ignore_index=True નો ઉપયોગ

#  pivot() અને pivot_table() દ્વારા Data Reshape કરવો

#  melt() થી Wide Data ને Long Data માં ફેરવવું

# 1. Concatenation શું છે?
# English definition: Concatenation is the process of combining DataFrames along rows or columns.

# ગુજરાતીમાં: બે અથવા વધુ DataFrames ને એકની નીચે એક અથવા બાજુમાં જોડવાની પ્રક્રિયાને Concatenation કહે છે.

# ઉદાહરણ તરીકે, જાન્યુઆરીના Sales Data અને ફેબ્રુઆરીના Sales Data અલગ Files માં છે. બંનેનું Format સરખું હોય તો તેને એક DataFrame માં જોડવા pd.concat() વાપરી શકાય.

```

```python
# 2. pd.concat() — DataFrames ને ઉપર-નીચે જોડવા
# સૌપ્રથમ બે DataFrames બનાવીએ.

import pandas as pd

jan_sales = pd.DataFrame({
    "Product": ["Laptop", "Mobile"],
    "Sales": [50000, 20000]
})

feb_sales = pd.DataFrame({
    "Product": ["Tablet", "Keyboard"],
    "Sales": [15000, 3000]
})

print("January Sales:")
print(jan_sales)

print("\nFebruary Sales:")
print(feb_sales)
```

```python
# હવે બંનેને ઉપર-નીચે જોડીએ.

all_sales = pd.concat(
    [jan_sales, feb_sales]
)

print(all_sales)
```

```python
# ignore_index=True નો ઉપયોગ
# જો આપણે નવી સતત Index Numbering જોઈએ તો:

all_sales = pd.concat(
    [jan_sales, feb_sales],
    ignore_index=True # coutine index add kari hoy to
)

print(all_sales)
```

```python
# 3. axis=0 અને axis=1 વચ્ચેનો તફાવત
# Pandas માં axis બતાવે છે કે કઈ દિશામાં Operation કરવું.
# axis=0

# Row-wise Concatenation

# DataFrames ને ઉપર-નીચે જોડે છે. સામાન્ય રીતે અલગ મહિનાઓ અથવા અલગ બેચના Records જોડવા માટે ઉપયોગી છે.



# axis=1

# Column-wise Concatenation

# DataFrames ને બાજુમાં જોડે છે. સામાન્ય રીતે સમાન Index ધરાવતા Records માટે ઉપયોગી છે.



# axis=1 નો Practical
student_names = pd.DataFrame({
    "Name": ["Amit", "Neha", "Ravi"]
})

student_marks = pd.DataFrame({
    "Marks": [85, 92, 76]
})

result = pd.concat(
    [student_names, student_marks],
    axis=1
)

print(result)
```

```python
# 4. Columns અલગ હોય ત્યારે concat() શું કરે છે?
# બંને DataFrames માં Columns અલગ હોય તો Pandas બધા Columns જાળવે છે અને જ્યાં Value ઉપલબ્ધ ન હોય ત્યાં NaN મૂકે છે

df1 = pd.DataFrame({
    "Name": ["Amit", "Neha"],
    "Python": [85, 90]
})

df2 = pd.DataFrame({
    "Name": ["Ravi", "Pooja"],
    "SQL": [80, 95]
})

result = pd.concat(
    [df1, df2],
    ignore_index=True
)

print(result)

# આનું કારણ એ છે કે પ્રથમ DataFrame માં SQL Column નથી અને બીજા DataFrame માં Python Column નથી.
```

```python
# 5. pivot() — Long Data ને Wide Data માં ફેરવવું
# English definition: pivot() reshapes data by turning unique values from one column into separate columns.

# ગુજરાતીમાં: કોઈ Column ની Values ને નવા Columns માં ફેરવીને Data ને નવા Format માં ગોઠવવાની રીતને Pivot કહે છે.

# ઉદાહરણ તરીકે, Sales Data માં દરેક Product અને Month માટે અલગ Row હોય:

sales = pd.DataFrame({
    "Month": ["Jan", "Jan", "Feb", "Feb"],
    "Product": ["Laptop", "Mobile", "Laptop", "Mobile"],
    "Amount": [50000, 20000, 55000, 25000]
})

print(sales)
```

```python
# આને Pivot કરીએ:

pivot_sales = sales.pivot(
    index="Month",
    columns="Product",
    values="Amount"
)

print(pivot_sales)

# index="Month" — Row માં Month રાખવો.
# columns="Product" — દરેક Product માટે અલગ Column બનાવવો.
# values="Amount" — કઈ Value ભરવી તે નક્કી કરે છે.
```

```python
# 6. pivot_table() — Duplicate Records સાથે Analysis
# જો એક જ Month માં એક જ Product ના અનેક Sales Records હોય તો pivot_table() વાપરી શકાય છે. તે Aggregation દ્વારા અનેક Values ને એક Value માં ફેરવી શકે છે.

sales = pd.DataFrame({
    "Month": ["Jan", "Jan", "Jan", "Feb", "Feb"],
    "Product": ["Laptop", "Laptop", "Mobile", "Laptop", "Mobile"],
    "Amount": [30000, 20000, 20000, 55000, 25000]
})

pivot_result = sales.pivot_table(
    index="Month",
    columns="Product",
    values="Amount",
    aggfunc="sum",
    fill_value=0
)

print(pivot_result)

# Parameter	ઉપયોગ
# index	Row માં કયો Column રાખવો
# columns	કયા Column ની Values ને નવા Columns બનાવવા
# values	કઈ Numeric Value નું Analysis કરવું
# aggfunc="sum"	કુલ સરવાળો
# aggfunc="mean"	સરેરાશ
# aggfunc="count"	Records ની સંખ્યા
# fill_value=0	ખાલી જગ્યાએ 0 મૂકવો
```

```python
# 7. melt() — Wide Data ને Long Data માં ફેરવવું
# English definition: melt() converts a DataFrame from wide format to long format by unpivoting selected columns into rows.

# ગુજરાતીમાં: અનેક અલગ Columns માં રહેલી Values ને એક Value Column માં લાવીને Rows માં ફેરવવાની પ્રક્રિયાને melt() કહે છે.

# આ પ્રકારનું Reshaping Data Visualization અને Analytics માટે ઉપયોગી છે.

# આપણી પાસે આ Data છે:

wide_data = pd.DataFrame({
    "Student": ["Amit", "Neha", "Ravi"],
    "Python": [85, 90, 78],
    "SQL": [80, 95, 88],
    "PowerBI": [75, 85, 92]
})

print(wide_data)
```

```python
#long data concert
long_data = pd.melt(
    wide_data,
    id_vars=["Student"],
    var_name="Subject",
    value_name="Marks"
)

print(long_data)


# id_vars=["Student"] — Student Column ને ઓળખ માટે જાળવે છે.
# var_name="Subject" — મૂળ Column Names ને Subject માં મૂકે છે.
# value_name="Marks" — Marks ની Values ને Marks Column માં મૂકે છે.
```

```python
# Pivot અને Melt વચ્ચેનો તફાવત
# Function	શું કરે છે?
# pivot()	Long Data ને Wide Format માં ફેરવે છે
# pivot_table()	Aggregation સાથે Wide Format બનાવે છે
# melt()	Wide Data ને Long Format માં ફેરવે છે
# concat()	DataFrames ને જોડે છે

```

```python
# 8. Mini Project — Monthly Student Results
# હવે એક Practical Project માં શીખેલી પદ્ધતિઓનો ઉપયોગ કરીએ.

# ધારો કે બે અલગ Batches ના Results છે.

batch_a = pd.DataFrame({
    "StudentID": [101, 102, 103],
    "Name": ["Amit", "Neha", "Ravi"],
    "Python": [85, 90, 78]
})

batch_b = pd.DataFrame({
    "StudentID": [104, 105, 106],
    "Name": ["Pooja", "Karan", "Mira"],
    "Python": [92, 74, 88]
})

all_students = pd.concat(
    [batch_a, batch_b],
    ignore_index=True
)

all_students["Result"] = all_students["Python"].apply(
    lambda marks: "Pass" if marks >= 40 else "Fail"
)

print(all_students)

# આ Project માં:
# 1. બે Batches ના DataFrames concat() થી જોડ્યા.
# 2. Index ફરીથી બનાવ્યો.
# 3. apply() અને lambda થી Pass/Fail Result બનાવ્યો.
# આવી રીતે અલગ Files માં રહેલા વિદ્યાર્થીઓનો એક Final Result તૈયાર કરી શકાય છે.
```
