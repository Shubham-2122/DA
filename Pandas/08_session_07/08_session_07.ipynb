```python
# 2. Practical Dataset બનાવો
# VS Code માં Jupyter Notebook ખોલો અને નીચેનો Code ચલાવો.

import pandas as pd
import numpy as np

data = {
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira"],
    "Department": ["IT", "CS", "IT", "CS", "IT", "CS"],
    "Marks": [85, 92, 76, 88, 67, 95],
    "Attendance": [90, 95, 80, 85, 70, 98],
    "Fees": [15000, 18000, 15000, 18000, 16000, 18000]
}

df = pd.DataFrame(data)

print(df)
```

```python
# 3. નવા Columns બનાવવા
# 3.1 Arithmetic Operations વડે Column બનાવો
# ધારો કે દરેક વિદ્યાર્થીના Marks 100 માંથી છે અને આપણે Percentage Column બનાવવો છે

df["Percentage"] = (df["Marks"] / 100) *100

print(df[["Name","Marks","Percentage"]])
```

```python
# અહીં Marks 100 માંથી હોવાથી Percentage અને Marks ની Numeric Value એકસરખી આવશે.

# હવે ધારો કે Marks 500 માંથી છે:

df["Percentage_500"] = (df["Marks"] / 500) * 100
print(df[["Name","Marks","Percentage_500"]])
```

```python
# 3.2 બે Columns માંથી નવો Column બનાવો
# ધારો કે દરેક વિદ્યાર્થીની બાકી Fees શોધવી છે.

df["TotalFees"] = 20000
df["PendingFees"] = df["TotalFees"] - df["Fees"]

print(df[["Name","TotalFees","Fees","PendingFees"]])

# - TotalFees = કુલ Fees
# - Fees = ચૂકવેલી Fees
# - PendingFees = બાકી Fees
# જો Amit ની કુલ Fees ₹20,000 હોય અને તેણે ₹15,000 ચૂકવ્યા હોય, તો બાકી Fees ₹5,000 થાય.
```

```python
# ""3.3 બે Text Columns જોડીને નવો Column બનાવો

df["FirstName"] = ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira"]
df["LastName"] = ["Shah", "Patel", "Joshi", "Mehta", "Desai", "Shah"]

df["FullName"] = df["FirstName"] + " " + df["LastName"]

print(df[["FirstName","LastName","FullName"]])

# આ રીત Student Registration Systems અને Customer Databases માં ઉપયોગી છે.
```

```python
# 4. apply() Function શું છે?
# English Definition: apply() applies a function along a DataFrame axis or to each value of a Series.

# ગુજરાતીમાં: apply() વડે Series ની દરેક Value અથવા DataFrame ની દરેક Row/Column પર Function લાગુ કરી શકાય છે.
# 4.1 apply() વગર Grade નક્કી કરવો
def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"

df["Grade"] = df["Marks"].apply(get_grade)

print(df[["Name", "Marks", "Grade"]])

# apply(get_grade) દરેક Marks Value માટે Function ચલાવે છે.
```

```python
# 4.2 apply() વડે Attendance Category બનાવો

def attendance_status(attendance):
    if attendance >= 90:
        return "Excellent"
    elif attendance >= 75:
        return "Good"
    else:
        return "Low"

df["AttendanceStatus"] = df["Attendance"].apply(
    attendance_status
)

print(df[["Name","Attendance","AttendanceStatus"]])
```

```python
# 5. lambda Function શું છે?
# English Definition: A lambda function is a small anonymous function defined using the lambda keyword.

# ગુજરાતીમાં: lambda એ નાનું Function છે, જેને અલગ નામ આપ્યા વગર એક જ Expression માટે લખી શકાય છે.

def double(x):
    return x * 2

print(double(5))
```

```python
double = lambda x: x * 2

print(double(5))
```

```python
# 5.1 apply() સાથે lambda
# દરેક વિદ્યાર્થીના Marks માં 5 Bonus Marks ઉમેરવાના છે.

df["BonusMarks"] = df["Marks"].apply(
    lambda x : x +5
)

print(df[["Name","Marks","BonusMarks"]])

# after your give 5 marks add
```

```python
# 5.2 lambda સાથે Conditions

df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 70 else "Fail"
)

print(df[["Name", "Marks", "Result"]])

# આ Code માં:
# - Marks 70 કે તેથી વધુ હોય તો Pass.
# - અન્યથા Fail.
# આ Threshold માત્ર પ્રેક્ટિસ માટે છે; વાસ્તવિક Passing Marks અલગ હોઈ શકે છે.
```

```python
# 5.3 એકથી વધુ Conditions સાથે lambda

df["Grade"] = df["Marks"].apply(
    lambda x: "A" if x >= 90
    else "B" if x >= 80
    else "C" if x >= 70
    else "D"
)

print(df[["Name", "Marks", "Grade"]])

# ટૂંકી અને સરળ Conditions માટે lambda ઉપયોગી છે. ઘણી Conditions અથવા જટિલ Logic હોય તો def વડે Function લખવું વધુ વાંચવા યોગ્ય રહે છે.
```

```python
# 6. Vectorized Operations અને apply() વચ્ચેનો તફાવત
# Pandas માં શક્ય હોય ત્યાં સીધા Column Operations વાપરવા વધુ સરળ અને ઘણી વખત વધુ ઝડપી હોય છે.

df["BonusMarks"] = df["Marks"] + 5

df["BonusMarks"] = df["Marks"].apply(
    lambda x: x + 5
)


# Method	ઉપયોગ
# df["Marks"] + 5	સીધી Numeric ગણતરી માટે
# apply()	Custom Function લાગુ કરવા
# lambda	ટૂંકું Function લખવા
# def	જટિલ Logic અને ફરીથી ઉપયોગ કરી શકાય તેવા Function માટે

```

```python
# 7. apply() માં axis સમજીએ
# DataFrame.apply() માં axis નક્કી કરે છે કે Function Rows પર લાગુ થશે કે Columns પર.

# દરેક Row માં Marks અને Attendance નો સરવાળો
df["MarksPlusAttendance"] = df[["Marks", "Attendance"]].apply(
    lambda row: row["Marks"] + row["Attendance"],
    axis=1
)

print(df[["Name", "Marks", "Attendance", "MarksPlusAttendance"]])

# axis=1: દરેક Row પર Function લાગુ થાય છે.
# axis=0: દરેક Column પર Function લાગુ થાય છે.
```

```python
# 8. applymap() વિશે જાણો
# જૂના Pandas અભ્યાસક્રમોમાં DataFrame.applymap() શીખવવામાં આવતું હતું. નવી Pandas આવૃત્તિઓમાં આ કામ માટે DataFrame.map() ઉપલબ્ધ છે.

# ધારો કે DataFrame ના બધા Text ને Uppercase કરવાના છે:

names_df = pd.DataFrame({
    "FirstName": ["amit", "neha"],
    "Department": ["it", "cs"]
})

result = names_df.map(
    lambda x: x.upper() if isinstance(x, str) else x
)

print(result)
```

```python
# 9. Complete Practical — Student Performance Report
# હવે આપણે શીખેલા Concepts ભેગા કરીને નાનો Project બનાવીએ.


import pandas as pd

data = {
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira"],
    "Marks": [85, 92, 76, 88, 67, 95],
    "Attendance": [90, 95, 80, 85, 70, 98],
    "FeesPaid": [15000, 18000, 15000, 18000, 16000, 18000]
}

df = pd.DataFrame(data)

# 1. બાકી Fees
df["TotalFees"] = 20000
df["PendingFees"] = df["TotalFees"] - df["FeesPaid"]

# 2. Grade
def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"

df["Grade"] = df["Marks"].apply(get_grade)

# 3. Result
df["Result"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 70 else "Fail"
)

# 4. Attendance Status
df["AttendanceStatus"] = df["Attendance"].apply(
    lambda x: "Eligible" if x >= 75 else "Review"
)

# 5. Final Report
print(df[[
    "StudentID", "Name", "Marks", "Grade", "Result",
    "Attendance", "AttendanceStatus", "PendingFees"
]])

# 6. Excel માં Export
df.to_excel("student_performance_report.xlsx", index=False)
```
