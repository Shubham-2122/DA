```python
# Session 8.13 — Exploratory Data Analysis (EDA) with Pandas
# આજે આપણે શીખીશું કે કોઈપણ Dataset પર Machine Learning અથવા Data Science શરૂ કરતાં પહેલાં તેની તપાસ કેવી રીતે કરવી.

# ધારો કે તમને 10,000 Students નો Dataset મળે છે. તેમાં કેટલા Students છે, કયા Columns માં Missing Values છે, Duplicate Records કેટલા છે, Marks નો Average કેટલો છે અને કયા Columns વચ્ચે સંબંધ છે — આ બધું EDA દ્વારા જાણી શકાય છે.
# આજના Learning Objectives
#  EDA શું છે અને શા માટે જરૂરી છે?

#  shape, info(), describe() દ્વારા Dataset સમજવો

#  Missing Values અને Duplicate Records તપાસવા

#  Numerical અને Categorical Columns નું Analysis

#  Correlation અને EDA પરથી Insights મેળવવા

# 1. EDA શું છે?
# English definition: Exploratory Data Analysis (EDA) is the process of inspecting, summarizing, and visualizing data to understand its structure, quality, patterns, and relationships.

# ગુજરાતીમાં: Dataset ની રચના, Data Quality, Missing Values, Patterns અને Columns વચ્ચેના સંબંધોને સમજવા માટે કરવામાં આવતી તપાસને Exploratory Data Analysis (EDA) કહે છે.

# EDA કરવાના મુખ્ય હેતુઓ:
# - Dataset માં શું માહિતી છે તે સમજવું.
# - Missing અને Duplicate Records શોધવા.
# - Data માં રહેલી અસામાન્ય Values ઓળખવી.
# - Columns વચ્ચેના સંબંધો સમજવા.
# - આગળની Data Cleaning અને Machine Learning માટે યોગ્ય નિર્ણયો લેવા.
```

```python
# 2. Practical Dataset બનાવીએ
# આજે આપણે Students ના Marks અને Study Hours નો Dataset લઈશું, જેમાં ઇરાદાપૂર્વક Missing Values અને Duplicate Records પણ રાખ્યાં છે.

import pandas as pd
import numpy as np

data = {
    "StudentID": [101, 102, 103, 104, 105, 106, 106],
    "Name": [
        "Amit", "Neha", "Ravi", "Pooja",
        "Karan", "Mira", "Mira"
    ],
    "Department": [
        "IT", "CS", "IT", "CS", "IT", "CS", "CS"
    ],
    "StudyHours": [2, 4, 3, 6, np.nan, 5, 5],
    "Marks": [55, 75, 65, 90, 60, 85, 85],
    "Attendance": [70, 85, 75, 95, 80, np.nan, np.nan]
}

df = pd.DataFrame(data)

print(df)

# આ Dataset માં:
# - કુલ 7 Rows છે.
# - StudyHours માં એક Missing Value છે.
# - Attendance માં બે Missing Values છે.
# - છેલ્લી બે Rows Duplicate છે.
```

```python
# 3. Dataset ની Basic Information તપાસવી
# A. shape — Rows અને Columns ની સંખ્યા

print(df.shape) # 7 row and 6 colums
```

```python
# B. head() અને tail()

print(df.head()) # first 5 row batave
print(df.tail(3)) # last 3 row batave

# head() ડિફૉલ્ટ રીતે પ્રથમ 5 Rows બતાવે છે.
# tail(3) છેલ્લી 3 Rows બતાવે છે.
```

```python
# C. info() — Data Types અને Non-null Values

df.info()
```

```python
# આ Output માં તમને મળશે:
# - દરેક Column નું નામ.
# - Non-null Values ની સંખ્યા.
# - દરેક Column નો Data Type.
# - કુલ Memory Usage
```

```python
# D. describe() — Statistical Summary
print(df.describe()) # statistical summery count,mean,std,min
```

```python
# Categorical Columns માટે:

print(df.describe(include="object")) 

# આથી Name અને Department જેવા Text Columns માટે Count, Unique Values, સૌથી સામાન્ય Value અને તેની Frequency જોઈ શકાય છે.
```

```python
# 4. Missing Values નું Analysis
# Dataset માં કઈ જગ્યાએ Data ઉપલબ્ધ નથી તે શોધવું EDA નું અગત્યનું પગલું છે.

print(df.isnull().sum())
```

```python
# Missing Values નો Percentage શોધવો

missing_percent = (df.isnull().mean() * 100).round(2)

print(missing_percent)
```

```python
# Missing Values નું Visualization

missing_summary = df.isnull().sum()

print(
    missing_summary[missing_summary > 0]
)
```

```python
# # આથી ફક્ત Missing Values ધરાવતા Columns જ જોવા મળશે.

# # મહત્વપૂર્ણ: EDA કરતી વખતે Missing Values તરત જ Delete ન કરો. પહેલાં સમજવું જોઈએ કે Value શા માટે Missing છે અને તેને Fill કરવી કે Record દૂર કરવો તે તમારા Analysis પર કેવી અસર કરશે.
# 5. Duplicate Records શોધવા
# Duplicate Records હોવાથી ગણતરીઓ ખોટી થઈ શકે છે.

print(df.duplicated())
print("Total duplicates:", df.duplicated().sum())
```

```python
# અહીં સંપૂર્ણ સમાન Rows તપાસવામાં આવે છે. આપણી Sample Data માં છેલ્લી Row Duplicate છે.

# Duplicate Rows જોવા માટે:

duplicates = df[df.duplicated(keep=False)]

print(duplicates)
```

```python
# જો ખરેખર Duplicate Row દૂર કરવી યોગ્ય હોય તો:

clean_df = df.drop_duplicates().copy()

print(clean_df.shape) # duplicates nathi and shap ma aavse

# પરંતુ એક બાબત ધ્યાનમાં રાખો: જો બે અલગ Rows માં Student ID સમાન હોય પણ બાકીની માહિતી અલગ હોય, તો તે જરૂરી નથી કે Duplicate હોય. એટલે Business Rules પ્રમાણે તપાસવું જોઈએ.
```

```python
# 6. Numerical Columns નું Analysis
# Numerical Columns માં StudyHours, Marks અને Attendance આવે છે.
# A. Average, Median અને Standard Deviation

# A. Average, Median અને Standard Deviation

print("Mean Marks:", df["Marks"].mean())
print("Median Marks:", df["Marks"].median())
print("Standard Deviation:", df["Marks"].std())
```

```python
# B. Range તપાસવી
print("Minimum Marks:", df["Marks"].min())
print("Maximum Marks:", df["Marks"].max())
print("Marks Range:", df["Marks"].max() - df["Marks"].min())
```

```python
# C. અલગ-અલગ Numeric Columns ની Summary

numeric_summary = df[
    ["StudyHours", "Marks", "Attendance"]
].describe()

print(numeric_summary)
```

```python
# 7. Categorical Columns નું Analysis
# Categorical Columns માં Department અને Name જેવા Columns આવે છે.
# A. Department પ્રમાણે Student Count

print(df["Department"].value_counts())
```

```python
# B. Department નો Percentage

print(
    df["Department"].value_counts(normalize=True) * 100
)
```

```python
# C. Unique Values તપાસવી

print(df["Department"].unique())
print(df["Department"].nunique())

# unique() — અલગ-અલગ Values બતાવે છે.
# nunique() — કેટલી Unique Values છે તે ગણે છે.
```

```python
# D. Department પ્રમાણે Average Marks

department_analysis = df.groupby("Department").agg(
    Student_Count=("StudentID", "count"),
    Average_Marks=("Marks", "mean"),
    Highest_Marks=("Marks", "max")
)

print(department_analysis)

# આથી IT અને CS Department ના Marks ની તુલના કરી શકાય છે. ધ્યાન રાખો કે Duplicate Records હાજર હોય તો Student Count અને Average પર અસર થઈ શકે છે.
```

```python
# 8. Correlation — બે Numeric Columns વચ્ચેનો સંબંધ
# English definition: Correlation measures the strength and direction of a relationship between numerical variables.

# ગુજરાતીમાં: બે Numerical Variables વચ્ચેનો સંબંધ કેટલો મજબૂત છે અને તે કઈ દિશામાં છે તેનો અંદાજ Correlation દ્વારા મળે છે.

correlation = df[
    ["StudyHours", "Marks", "Attendance"]
].corr()

print(correlation)
```

```python
# 9. EDA પરથી Insights કેવી રીતે મેળવવા?
# હવે આપણા Dataset માંથી થોડાં પ્રશ્નોના જવાબ શોધીએ.
# પ્રશ્ન 1: સૌથી વધુ Marks કોના છે?

top_student = df.loc[df["Marks"].idxmax()]

print(top_student)
```

```python
# આ સૌથી વધુ Marks ધરાવતી Row આપે છે. જો એકથી વધુ વિદ્યાર્થીઓના Marks સૌથી વધુ હોય તો આ રીત પ્રથમ મળેલી Row આપે છે.

# બધા Top Scorers જોવા:

max_marks = df["Marks"].max()

top_students = df[df["Marks"] == max_marks]

print(top_students)
```

```python
# પ્રશ્ન 2: 80 કે તેથી વધુ Marks ધરાવતા વિદ્યાર્થીઓ કેટલા?

high_scorers = df[df["Marks"] >= 80]

print(high_scorers[["Name", "Marks"]])
print("Total:", len(high_scorers))
```

```python
# પ્રશ્ન 3: Department પ્રમાણે સરેરાશ Study Hours

study_analysis = df.groupby("Department").agg(
    Average_Study_Hours=("StudyHours", "mean"),
    Average_Marks=("Marks", "mean")
)

print(study_analysis)

# આથી Study Hours અને Marks ને Department પ્રમાણે તપાસી શકાય છે.
```
