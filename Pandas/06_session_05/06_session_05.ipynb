```python
# # Session 8.5 — Pandas માં Handling Missing Values
# # આજે આપણે Data Science માં ખૂબ જ મહત્વનો વિષય શીખીશું: Missing Values એટલે કે ડેટામાં ખાલી અથવા ઉપલબ્ધ ન હોય તેવી કિંમતોને ઓળખવી, દૂર કરવી અને યોગ્ય રીતે ભરવી.

# # આ Session તમે VS Code + Jupyter Notebook માં પ્રેક્ટિસ કરી શકશો અને વિદ્યાર્થીઓને પણ સમજાવી શકશો.

# # 8.5.1 — Missing Values શોધવા
# # isnull(), isna(), notnull()

# # 8.5.2 — Missing Values દૂર કરવા
# # dropna(), axis, how, subset

# # 8.5.3 — Missing Values ભરવા
# # fillna(), Mean, Median, Mode

# # 8.5.4 — Practical + Assessment
# # College dataset cleaning project

# # 1. Missing Values એટલે શું?
# # English Definition: Missing values are data entries that are absent, unknown, or not recorded in a dataset.

# # ગુજરાતીમાં: જ્યારે Dataset ના કોઈ Cell માં જરૂરી માહિતી ઉપલબ્ધ ન હોય, ત્યારે તેને Missing Value કહેવાય છે.

# Student	Marks	Attendance
# Amit	85	90
# Neha	NaN	95
# Ravi	76	NaN
# Pooja	88	85

# અહીં NaN એટલે Not a Number — Pandas માં Missing Value દર્શાવવા માટે વપરાતી સામાન્ય નિશાની.

# Missing values થવાનાં કારણો:
# - વિદ્યાર્થીએ માહિતી આપી ન હોય.
# - Data Entry કરતી વખતે ભૂલ થઈ હોય.
# - CSV અથવા Excel ફાઇલમાં Cell ખાલી હોય.
# - બે અલગ Dataset Merge કરતી વખતે માહિતી ન મળતી હોય.
# ધ્યાન રાખો: ખાલી Value ને દરેક વખતે 0 માનવી યોગ્ય નથી. જો Marks ઉપલબ્ધ ન હોય તો તેનો અર્થ 0 Marks થયો એવું નથી.
```

```python
# 2. Practical માટે Dataset બનાવો
# VS Code માં તમારી Jupyter Notebook ખોલીને નીચેનો Code ચલાવો.

import pandas as pd
import numpy as np

data = {
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira"],
    "Department": ["IT", "CS", "IT", None, "IT", "CS"],
    "Marks": [85, np.nan, 76, 88, np.nan, 95],
    "Attendance": [90, 95, np.nan, 85, 70, 98],
    "Fees": [15000, 18000, 15000, np.nan, 16000, 18000]
}

df = pd.DataFrame(data)

print(df)
```

```python
# આ Dataset માં Marks, Attendance, Department અને Fees માં કેટલીક Values ખૂટે છે. હવે આપણે તેને ઓળખતાં અને સાફ કરતાં શીખીશું.
# 3. Missing Values શોધવી
# 3.1 isnull() અને isna()
# બંને Functions Missing Values ઓળખે છે. આ બંને એકસરખી રીતે કામ કરે છે.

print(df.isnull())

# આઉટપુટમાં:
# - True એટલે Value Missing છે.
# - False એટલે Value હાજર છે.
# ઉદાહરણ તરીકે, Marks Column માટે:
```

```python
print(df["Marks"].isnull())
```

```python
# 3.2 દરેક Column માં કેટલી Missing Values છે?
# all ma ketli total value che

print(df.isnull().sum())

# સમજણ:
# - isnull() દરેક Cell તપાસે છે.
# - sum() દરેક Column માં True ની સંખ્યા ગણે છે.
# - તેથી આપણને ખબર પડે છે કે કયા Column માં કેટલી Values ખૂટે છે.
```

```python
# 3.3 Dataset માં કુલ કેટલી Missing Values છે?
#all sum find karse

print(df.isnull().sum().sum())
```

```python
# 3.4 Missing Values ની ટકાવારી શોધવી

missing_percent = (df.isnull().sum() / len(df)) *100

print(missing_percent)
```

```python
# 3.5 notnull() નો ઉપયોગ
# notnull() હાજર Values માટે True અને Missing Values માટે False આપે છે.

print(df["Marks"].notnull())
```

```python
# માત્ર Marks ઉપલબ્ધ હોય તેવા Students જોવા:

print(df[df["Marks"].notnull()])
```

```python
# 4. dropna() — Missing Values દૂર કરવી
# જ્યારે Missing Value ધરાવતી Row અથવા Column દૂર કરવી યોગ્ય હોય ત્યારે dropna() ઉપયોગી છે.
# 4.1 Missing Value ધરાવતી Rows દૂર કરવી

clean_df = df.dropna()

print(clean_df)

# શું થશે? જે Row માં એક પણ Missing Value હશે તે Row દૂર થશે. આપણા Dataset માં દરેક Row માં બધી Values ઉપલબ્ધ નથી, એટલે ઘણી Rows દૂર થઈ શકે છે
```

```python
# 4.2 how="all" — બધી Values Missing હોય ત્યારે જ Row દૂર કરવી

clean_df = df.dropna(how="all")

print(clean_df)
```

```python
# 4.3 ચોક્કસ Column ના આધારે Row દૂર કરવી
# ધારો કે College Ranking માટે Marks ફરજિયાત છે. તો માત્ર Marks Missing હોય તેવી Rows દૂર કરી શકાય.

marks_df = df.dropna(subset=["Marks"])

print(marks_df)

# marks ma nan val na hoy ae show thase
```

```python
# 4.4 axis નો ઉપયોગ
# Code	અર્થ
# df.dropna(axis=0)	Missing Value ધરાવતી Rows દૂર કરે
# df.dropna(axis=1)	Missing Value ધરાવતા Columns દૂર કરે
# df.dropna(how="all")	બધી Values Missing હોય તેવી Rows દૂર કરે
# df.dropna(subset=["Marks"])	Marks Missing હોય તેવી Rows દૂર કરે

```

```python
# 5. fillna() — Missing Values ભરવી
# ઘણી વખત Data દૂર કરવાને બદલે યોગ્ય Value વડે ભરવો વધુ સારું હોય છે.

df_filled = df.fillna(0)

print(df_filled)

# all missing value ne 0 bhari dese
```

```python
# 5.2 ચોક્કસ Column માં Value ભરવી
# deaprtment mising hoy aema unknwon nakhi sakahy

df["Department"] = df["Department"].fillna("Unknown")

print(df)
```

```python
# 5.3 Mean વડે Missing Values ભરવી
# Mean = સરેરાશ

# mean = all value sum / value ni total ketali 

mean_marks = df["Marks"].mean()

df["Marks"] = df["Marks"].fillna(mean_marks)
# આ Dataset માં Missing Marks ને 86 થી ભરવામાં આવશે.

print("Mean Marks : ",mean_marks)

print(df[["Name","Marks"]])
```

```python
# 5.4 Median વડે Missing Values ભરવી
# Median = ગોઠવેલી Values ની મધ્યની Value.

# ઉદાહરણ: 10, 20, 30, 1000

# Mean = 265, પરંતુ Median = 25.

# અહીં 1000 જેવી મોટી Value Mean ને વધારે અસર કરે છે, જ્યારે Median વધુ યોગ્ય મધ્યસ્થ Value આપી શકે છે.

median_marks = df["Marks"].median()

df["Marks"] = df["Marks"].fillna(median_marks)

print("Median Marks : ",median_marks)
```

```python
# 5.5 Mode વડે Missing Values ભરવી
# Mode = સૌથી વધુ વાર આવતી Value.

# Categorical Data માટે તે ઉપયોગી છે, જેમ કે Department અથવા City.

mode_department = df["Department"].mode()[0]

df["Department"] = df["Department"].fillna(mode_department)

print(df[["Name","Department"]])
```

```python
# 6. Mean, Median અને Mode વચ્ચેનો તફાવત
# Method	ગુજરાતી અર્થ	ક્યારે ઉપયોગી?
# Mean	સરેરાશ	Numeric Data માં Outliers નો પ્રભાવ ઓછો હોય ત્યારે
# Median	મધ્યસ્થ Value	Outliers અથવા Skewed Data હોય ત્યારે
# Mode	સૌથી વધુ આવતી Value	Category અથવા વારંવાર આવતી Value માટે
# Constant	નક્કી Value, જેમ કે "Unknown"	Missing Category ને સ્પષ્ટ રીતે દર્શાવવા

```

```python
# 7. Real-World Practical — College Dataset Clean કરવો
# આ પ્રેક્ટિકલમાં આપણે મૂળ df ને બદલ્યા વગર એક Clean Dataset બનાવીએ છીએ.

clean_df = df.copy()

# 1. Department ની Missing Values ભરો
clean_df["Department"] = clean_df["Department"].fillna("Unknown")

# 2. Marks ને Median વડે ભરો
clean_df["Marks"] = clean_df["Marks"].fillna(
    clean_df["Marks"].median()
)

# 3. Attendance ને Median વડે ભરો
clean_df["Attendance"] = clean_df["Attendance"].fillna(
    clean_df["Attendance"].median()
)

# 4. Fees ને Median વડે ભરો
clean_df["Fees"] = clean_df["Fees"].fillna(
    clean_df["Fees"].median()
)

# 5. Missing Values ફરી તપાસો
print("Missing Values:")
print(clean_df.isnull().sum())

print("\nClean Dataset:")
print(clean_df)

# અપેક્ષિત પરિણામ: બધા Columns માં Missing Values ની સંખ્યા 0 હશે.

# આ પ્રક્રિયાને Data Cleaning કહેવાય છે. પરંતુ વાસ્તવિક પ્રોજેક્ટમાં Missing Values ભરવાની રીત નક્કી કરતી વખતે Business Rules અને Data નો અર્થ પણ ધ્યાનમાં લેવા જોઈએ.
```
