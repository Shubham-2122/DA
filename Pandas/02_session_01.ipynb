```python
# 5. Series શું છે?
# English Definition: A Pandas Series is a one-dimensional labeled array that can store values of different data types.

# ગુજરાતી વ્યાખ્યા: Series એ એક dimension ધરાવતું data structure છે, જેમાં values સાથે labels અથવા index હોય છે.
# સામાન્ય રીતે Series ને એક જ column જેવા સમજી શકાય, પરંતુ તેમાં દરેક value માટે index પણ હોય છે.
# Example — Student Marks
# Index	Marks
# 0	85
# 1	92
# 2	76

# અહીં 0, 1, 2 એ index છે અને 85, 92, 76 એ values છે.
```

```python
import pandas as pd

marks = pd.Series([85,92,76,88])

print(marks)

# aama mane phela index aapse and pachi value 
# numpy as array 
# padas as series kam karse

# સમજણ:
# - pd.Series() નવી Series બનાવે છે.
# - [85, 92, 76, 88] values ની list છે.
# - Pandas આપમેળે 0, 1, 2, 3 index આપે છે.
# - int64 દર્શાવે છે કે values integer પ્રકારની છે.
```

```python
marks = pd.Series(
    [85,92,76],
    index=["shubham","sujal","ram"] # index name tame aapi sako cho
)

print(marks)
print("Shubha marks : ",marks["shubham"])
print("Aveerage : ",marks.mean())


# Trainer tip: વિદ્યાર્થીઓને પૂછો: “જો Neha ના marks શોધવા માટે marks["Neha"] લખી શકીએ, તો index ને નામ આપવાથી શું ફાયદો થયો?”

# જવાબ: numeric position યાદ રાખવાની જગ્યાએ ઓળખી શકાય તેવા label દ્વારા value મેળવી શકીએ છીએ.

# ધ્યાન રાખો: custom index આપ્યા પછી marks[0] જેવા positional access કરતાં marks.iloc[0] વાપરવું વધુ સ્પષ્ટ છે; label માટે marks.loc["Amit"] વાપરી શકાય છે.
```

```python
# 6. DataFrame શું છે?
# English Definition: A Pandas DataFrame is a two-dimensional labeled data structure with rows and columns. Different columns can contain different data types.

# ગુજરાતી વ્યાખ્યા: DataFrame એ rows અને columns ધરાવતું બે dimension વાળું data structure છે. તેમાં એક column માં નામ, બીજા column માં integer marks અને ત્રીજા column માં attendance જેવી values રાખી શકાય છે.


# Feature	Series	        DataFrame
# Dimension	1D	            2D
# Structure	Single labeled sequence	Rows અને columns
# Example	Marks	        Student name, marks, city
# Creation	pd.Series()	    pd.DataFrame()
```

```python
import pandas as pd

data = {
    "name" : ["Amit","Neha","Ravi","Pooja"],
    "Age":[20,21,19,22],
    "Marks":[85,92,76,88],
    "City":["Ahmedabad","Surat","Rajkot","Ahmedabad"]
}

df = pd.DataFrame(data)

print(df)

# આ ઉદાહરણમાં:
# - દરેક dictionary key એક column બને છે.
# - દરેક list ની value એક row સાથે સંબંધિત છે.
# - બધા columns ની lists ની length સમાન હોવી જોઈએ.
# - Pandas આપમેળે row index બનાવે છે.
```

```python
# Practical 4 — DataFrame વિશે માહિતી મેળવો

print(df.shape)
print(df.columns) #index
print(df.dtypes) 
print(df["Marks"])
print(df[["name","Marks"]]) 

# સમજણ:
# Code	ઉપયોગ
# df.shape	rows અને columns ની સંખ્યા
# df.columns	column names
# df.dtypes	દરેક column નો data type
# df["Marks"]	એક column — Series
# df[["Name", "Marks"]]	બે columns — DataFrame
```

```python
# 7. NumPy અને Pandas નો સંબંધ
# NumPy અને Pandas એકબીજાના વિકલ્પ નથી; બંને અલગ-અલગ કામ માટે ઉપયોગી છે.

import numpy as np
import pandas as pd

marks = np.array([85,92,76,88])

df = pd.DataFrame({
    "Name" : ["Amit","Neha","Ravi","Pooja"],
    "Marks":marks
})

print(df)
print("Average nmarks:",np.mean(df["Marks"]))

# અહીં NumPy array માં numerical values રાખી છે અને Pandas DataFrame માં નામ સાથે જોડીને data manage કર્યો છે.
```
