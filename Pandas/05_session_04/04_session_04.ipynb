```python
# ession 8.4 — Pandas માં Sorting & Reindexing
# આજે આપણે શીખીશું કે DataFrame ના records ને ascending અથવા descending order માં કેવી રીતે ગોઠવવા, index પ્રમાણે sort કેવી રીતે કરવા, જૂના index ને ફરીથી કેવી રીતે ગોઠવવા અને custom index કેવી રીતે લાગુ કરવો.

# આ concepts Data Analysis માં ખૂબ ઉપયોગી છે. ઉદાહરણ તરીકે, કોલેજના સૌથી વધુ marks ધરાવતા વિદ્યાર્થીઓ, સૌથી ઓછી attendance ધરાવતા વિદ્યાર્થીઓ અથવા sales માં સૌથી વધુ કમાણી કરનારા products શોધવા માટે sorting વપરાય છે.
# 1. Learning Objectives
# આ session પૂર્ણ થયા પછી તમે:
# - sort_values() વડે column ની values પ્રમાણે data ગોઠવી શકશો.
# - Ascending અને descending sorting કરી શકશો.
# - Multiple columns પ્રમાણે sorting કરી શકશો.
# - sort_index() વડે index પ્રમાણે data ગોઠવી શકશો.
# - reset_index() વડે index ફરીથી બનાવી શકશો.
# - reindex() વડે rows અથવા columns ને જરૂરી order માં ગોઠવી શકશો.
# - Sorting અને reindexing વચ્ચેનો તફાવત સમજી શકશો.
# 2. Definitions — English અને ગુજરાતી
# Sorting

# English: Sorting is the process of arranging data in a specified order, such as ascending or descending order.

# ગુજરાતી: Data ને નાનીથી મોટી, મોટીથી નાની અથવા અન્ય નિર્ધારિત order માં ગોઠવવાની પ્રક્રિયાને Sorting કહે છે.

# Reindexing

# English: Reindexing is the process of changing or aligning the row or column labels of a DataFrame to a specified index.

# ગુજરાતી: DataFrame ના row અથવા column labels ને જરૂરી labels અને order પ્રમાણે ફરીથી ગોઠવવાની પ્રક્રિયાને Reindexing કહે છે.

# Index Resetting

# English: Resetting an index replaces the existing row index with a new default integer index, unless another index is specified.

# ગુજરાતી: હાલના row index ને બદલીને ફરીથી 0, 1, 2, ... જેવો default index બનાવવાની પ્રક્રિયાને Index Resetting કહે છે.
```

```python
import pandas as pd

data = {
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira"],
    "Department": ["IT", "CS", "IT", "CS", "IT", "CS"],
    "Marks": [85, 92, 76, 88, 67, 95],
    "Attendance": [90, 95, 80, 85, 70, 98]
}

df = pd.DataFrame(data)

print(df)
```

```python
# # આ dataset માં Marks અને Attendance જેવા numerical columns છે. હવે આપણે તેમને અલગ અલગ રીતે sort કરીશું.

# 4. sort_values() — Values પ્રમાણે Sorting
# English Definition: DataFrame.sort_values() sorts rows according to the values in one or more columns.

# ગુજરાતી સમજણ: DataFrame ના એક અથવા વધુ columns ની values ના આધારે rows ને ascending અથવા descending order માં ગોઠવવા માટે sort_values() વપરાય છે.

result =  df.sort_values("Marks") # marks vise ascceding order

print(result[["Name","Marks"]])
```

```python
# Default રીતે ascending=True હોય છે, એટલે નાની value પહેલાં આવે છે.

result = df.sort_values("Marks", ascending=False) # descing order ma karva

print(result[["Name", "Marks"]])
```

```python
# અહીં સૌથી વધુ marks ધરાવતી Mira પ્રથમ આવે છે.
# Parameter	અર્થ
# by	કયા column પ્રમાણે sort કરવું
# ascending=True	નાનીથી મોટી value
# ascending=False	મોટીથી નાની value
# inplace=True	મૂળ DataFrame માં ફેરફાર કરવો
# na_position	Missing values શરૂઆતમાં કે અંતે રાખવી

# Practical 3 — Attendance પ્રમાણે sort કરો
# by ma column ne karvu hoy lakhi sakay

result = df.sort_values(
    by="Attendance",
    ascending=True
)

print(result[["Name","Attendance"]]) 
```

```python
# Trainer tip: વિદ્યાર્થીઓને પૂછો: “જો principal ને સૌથી ઓછી attendance ધરાવતા 3 વિદ્યાર્થીઓની યાદી જોઈએ તો શું કરશો?”
# head thi aapre occhi jeni hase ae malse

result = df.sort_values(
    "Attendance",
    ascending=True
).head(3)

print(result[["Name", "Attendance"]]) 
```

```python
# 5. Multiple Columns પ્રમાણે Sorting
# વાસ્તવિક dataset માં એક column પ્રમાણે sorting પૂરતું ન હોય. ઉદાહરણ તરીકે, પહેલાં Department પ્રમાણે અને પછી દરેક Department માં Marks પ્રમાણે વિદ્યાર્થીઓ ગોઠવવા હોય.
# Practical 4 — Department અને Marks

result = df.sort_values(
    by=["Department","Marks"],
    ascending=[True,False]
)

print(result[["Name","Department","Marks"]])

# સમજણ:
# - Department માટે True: alphabetical order માં CS પછી IT.
# - Marks માટે False: દરેક department માં સૌથી વધુ marks પહેલાં.
# - ascending ની list માં દરેક column માટે અલગ setting આપી શકાય છે.
```

```python
result = df.sort_values(
    by=["Attendance","Marks"],
    ascending=[False,True]
)

print(result[["Name","Attendance","Marks"]])
```

```python
# Trainer challenge: એકસરખા marks ધરાવતા વિદ્યાર્થીઓ હોય તો tie-breaker તરીકે Attendance નો ઉપયોગ કરો.

result = df.sort_values(
    by=["Marks","Attendance"],
    ascending=[False,False]
)

print(result[["Name","Marks","Attendance"]])
```

```python
# 6. sort_index() — Index પ્રમાણે Sorting
# English Definition: DataFrame.sort_index() sorts the rows or columns according to their index labels.

# ગુજરાતી સમજણ: DataFrame ના index labels ને ascending અથવા descending order માં ગોઠવવા sort_index() વપરાય છે. આ values ને નહીં, પરંતુ index ને sort કરે છે.
# Practical 5 — Index descending
# પહેલાં DataFrame ને અલગ ક્રમમાં ગોઠવીએ:

temp = df.sort_values("Marks",ascending=False)

print(temp[["Name","Marks"]])
```

```python
result = temp.sort_index()

print(result[["Name", "Marks"]]) # asceding
```

```python
result = temp.sort_index(ascending=False) # descing order
print(result[["Name","Marks"]])
```

```python
# axis નો ઉપયોગ

## Rows ને index પ્રમાણે sort કરવી
df.sort_index(axis=0)

# columns labels ne sort karva
df.sort_index(axis=1)

# axis=0: rows ના index પ્રમાણે.
# axis=1: column labels પ્રમાણે.

# મહત્વપૂર્ણ તફાવત:
# - sort_values("Marks") — Marks ના આધારે rows ગોઠવે છે.
# - sort_index() — index labels ના આધારે rows ગોઠવે છે.
```

```python
# 7. reset_index() — Index ફરીથી બનાવવો
# English Definition: reset_index() resets the DataFrame's row index to the default integer index and can optionally retain the old index as a column.

# ગુજરાતી સમજણ: Filtering અથવા sorting પછી index 0, 2, 5 જેવો દેખાતો હોય, તો reset_index() વડે નવો continuous index બનાવી શકાય છે.
# Practical 6 — Filtering પછી reset

reult = df[df["Marks"] >= 85]

print(result[["Name","Marks"]])
```

```python
result = result.reset_index(drop=True)

print(result[["Name", "Marks"]])
```

```python
filtered = df[df["Marks"] >= 85]

# જૂનો index column તરીકે જાળવો
result1 = filtered.reset_index(drop=False)
print(result1)

# જૂનો index કાઢીને નવો index બનાવો
result2 = filtered.reset_index(drop=True)
print(result2)

# drop=False default છે: જૂનો index index નામના column તરીકે આવી શકે છે.
# drop=True: જૂનો index દૂર થાય છે અને નવો default index મળે છે.
```

```python
# inplace=True ક્યારે વાપરવું?
df.reset_index(drop=True, inplace=True)
```

```python
# 8. reindex() — નવો Index અથવા Column Order
# English Definition: reindex() conforms a DataFrame to a new set of row or column labels, reordering existing labels and introducing missing values for labels that are not present.

# ગુજરાતી સમજણ: DataFrame ને આપેલા નવા index labels પ્રમાણે ગોઠવવા reindex() વપરાય છે. નવા label માટે મૂળ data ન હોય, તો ત્યાં સામાન્ય રીતે NaN આવે છે.
# Practical 7 — Rows ને custom order માં ગોઠવો
# અહીં rows ને [3, 1, 5, 0] પ્રમાણે ગોઠવવામાં આવી છે.

result = df.reindex([3,1,5,0])

print(result[["Name","Marks"]])
```

```python
# # Practical 8 — નવો index label ઉમેરો

# result = df.reindex([0,1,2,3,4,5,6])

# print(result[["Name","Marks"]])

# Index 6 મૂળ DataFrame માં નથી, તેથી તે row માં NaN આવશે.

# આ ખૂબ મહત્વનું છે: reindex() નવો label આપીને આપમેળે નવા student ની માહિતી શોધતું નથી. Data ઉપલબ્ધ ન હોય તો values missing રહે છે.
```

```python
# Practical 9 — Columns ને custom order માં ગોઠવો

result = df.reindex(
    columns=["Name","Marks","Department","Attendance"]
)

print(result.head())
```

```python
# અહીં columns આપેલા ક્રમમાં આવશે. StudentID અને અન્ય સૂચિમાં ન હોય તેવા columns result માં સામેલ નહીં થાય.
# reindex() અને reset_index() વચ્ચેનો તફાવત
# Feature	reset_index()	reindex()
# મુખ્ય હેતુ	Default row index ફરી બનાવવો	ચોક્કસ labels અને order પ્રમાણે ગોઠવવું
# Default index	0, 1, 2, ...	તમે આપેલા labels
# નવો label	સામાન્ય રીતે લાગુ પડતું નથી	નવો label હોય તો missing values આવી શકે
# Example	df.reset_index(drop=True)	df.reindex([2, 0, 1])
```

```python
# 9. Real-World Case Study — Student Ranking Report
# ધારો કે કોલેજને highest marks, lowest attendance અને department-wise student ranking નો report બનાવવો છે.

import pandas as pd
data = {
    "StudentID": [101, 102, 103, 104, 105, 106],
    "Name": ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira"],
    "Department": ["IT", "CS", "IT", "CS", "IT", "CS"],
    "Marks": [85, 92, 76, 88, 67, 95],
    "Attendance": [90, 95, 80, 85, 70, 98]
}

df = pd.DataFrame(data)

# 1 top 3 Students 
top_students = df.sort_values(
    "Marks",ascending=False
).head(3)

print("TOP 3 Studensts")
print(top_students[["Name","Marks"]])

# 2. Lowest attendance
low_attendance = df.sort_values(
    "Attendance", ascending=True
).head(3)

print("\nLOWEST ATTENDANCE")
print(low_attendance[["Name", "Attendance"]])

# 3. Department-wise ranking
department_ranking = df.sort_values(
    by=["Department", "Marks"],
    ascending=[True, False]
)

print("\nDEPARTMENT-WISE RANKING")
print(department_ranking[["Name", "Department", "Marks"]])

# 4. Export report
department_ranking.to_excel(
    "department_ranking.xlsx",
    index=False
)
```
