```python
from selenium import webdriver

from selenium.webdriver.common.by import By

driver = webdriver.Chrome()

try:

    driver.get(
        "https://the-internet.herokuapp.com/add_remove_elements/"
    )

    add_button = driver.find_element(
        By.XPATH,
        "//button[text()='Add Element']"
    )

    add_button.click()

    print("Button Clicked Successfully")

finally:

    driver.quit()
```

```python
# Practical – Search Using Enter Key
from selenium import webdriver

from selenium.webdriver.common.by import By

from selenium.webdriver.common.keys import Keys

driver = webdriver.Chrome()

try:

    driver.get("https://www.google.com")

    search_box = driver.find_element(
        By.NAME,
        "q"
    )

    search_box.send_keys("Python Web Scraping")

    search_box.send_keys(Keys.ENTER)

    print("Search Submitted")

finally:

    driver.quit()
```

```python
# javascript alerat 
# Step 1: Open Website
from selenium import webdriver

from selenium.webdriver.common.by import By

driver = webdriver.Chrome()

driver.get(
    "https://the-internet.herokuapp.com/javascript_alerts"
)
```

```python
alert_button = driver.find_element(
    By.XPATH,
    "//button[text()='Click for JS Alert']"
)

alert_button.click()
```

```python
# Step 3: Switch to Alert
alert = driver.switch_to.alert
```

```python
# Step 4: Read Alert Text
print(alert.text)
```

```python
alert.accept()
```

```python
# B. Confirmation Popup
button = driver.find_element(
    By.XPATH,
    "//button[text()='Click for JS Confirm']"
)

button.click()
```

```python
alert = driver.switch_to.alert

alert.accept()
```

```python
alert.dismiss()
```
