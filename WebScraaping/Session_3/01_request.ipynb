```python
#get request
import requests

url = "https://books.toscrape.com/"

response = requests.get(url)

print(response.status_code)
print(response.url)
print(response.text[:500])
```

```python
# #Post method 
# જ્યારે આપણે website ના server ને information મોકલવી હોય ત્યારે POST request નો ઉપયોગ થાય છે.

# Real-world Example:

# ધારો કે website પર Contact Form છે.

# તેમાં:
# - Name
# - Email
# - Message

import requests 

url ="https://httpbin.org/post"

data = {
    "name":"Shubham",
    "email":"shubham@example.com",
    "course":"DA/DS"
}

response = requests.post(url,data=data)

print(response.status_code)
print(response.json())
```

```python
#user - agent : clinet or software ni information aape  

import requests

url = "https://httpbin.org/user-agent"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(
    url,
    headers=headers
)

print(response.json())
```

```python
# custom Header 
import requests

url = "https://httpbin.org/headers"

headers = {
    "User-Agent": "MyPythonScraper/1.0",
    "Accept": "application/json",
    "Accept-Language": "en-US"
}

response = requests.get(
    url,
    headers=headers
)

print(response.json())
```

```python
#cookies How to store   

import requests

url = "https://httpbin.org/cookies/set"
res = requests.get(
    url,
    params = {"username":"shubham"}
)
print(res.status_code)
print(res.url)
```

```python
# have cookies ne ye joei ye 
print(res.cookies.get_dict())
```

```python
#manually cookies set kari ye 
import requests

url = "https://httpbin.org/cookies"

cookies = {
    "username": "Rahul",
    "course": "Python"
}

response = requests.get(
    url,
    cookies=cookies
)

print(response.json())
```

```python
# without session 
import requests

url = "https://httpbin.org/get"

response1 = requests.get(url)

response2 = requests.get(url)

print(response1.status_code)
print(response2.status_code)
```

```python
#with session   
import requests

session = requests.Session()

session.headers.update({
    "User-Agent": "MyPythonScraper/1.0"
})

response1 = session.get(
    "https://httpbin.org/get"
)

response2 = session.get(
    "https://httpbin.org/headers"
)

print(response1.status_code)
print(response2.status_code)
```

```python
# session allow cookies maintain 
import requests

session = requests.Session()

# First Request: Set Cookie
response1 = session.get(
    "https://httpbin.org/cookies/set",
    params={"username": "Rahul"}
)

print("First Request:")
print(response1.status_code)

# Second Request: Check Cookie
response2 = session.get(
    "https://httpbin.org/cookies"
)

print("\nSecond Request:")
print(response2.json())
```

```python
#with session allow  che 
import requests

with requests.Session() as session:

    response = session.get(
        "https://httpbin.org/get",
        timeout=20
    )

    print(response.status_code)
```

```python
# Request Headers vs Response Headers
# Request Headers	Response Headers
# Client મોકલે છે	    Server મોકલે છે
# User-Agent	    Content-Type
# Accept	        Content-Length
# Authorization	    Set-Cookie
```
