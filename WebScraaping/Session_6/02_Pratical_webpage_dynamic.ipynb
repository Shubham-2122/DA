```python
from selenium import webdriver

from selenium.webdriver.common.by import By

from selenium.webdriver.support.ui import WebDriverWait

from selenium.webdriver.support import expected_conditions as EC
```

```python
# Step 2: Open Chrome Browser

driver = webdriver.Chrome()

driver.get(
    "https://the-internet.herokuapp.com/dynamic_loading/1"
)
```

```python
# Step 3: Find Start Button
start_button = driver.find_element(
    By.CSS_SELECTOR,
    "#start button"
)

print(start_button.text)
```

```python
#start button direact click karse
start_button.click()
```

```python
# Step 5: Use Explicit Wait

wait = WebDriverWait(driver, 10)

result = wait.until(
    EC.visibility_of_element_located(
        (By.CSS_SELECTOR, "#finish h4")
    )
)

print(result.text)
```

```python
# Step 6: Close Browser
driver.quit()
```

```python
# webdriver.Chrome()	Browser launch
# driver.get()	Website open
# find_element()	HTML element find
# .click()	Button click
# WebDriverWait()	Wait for condition
# visibility_of_element_located()	Wait until element visible
# .text	Text extraction
# driver.quit()	Browser close

```

```python
# new particel
from selenium import webdriver

from selenium.webdriver.common.by import By

from selenium.webdriver.support.ui import WebDriverWait

from selenium.webdriver.support import expected_conditions as EC


driver = webdriver.Chrome()

all_quotes = []

try:

    driver.get("https://quotes.toscrape.com/js/")

    wait = WebDriverWait(driver, 15)

    quotes = wait.until(
        EC.presence_of_all_elements_located(
            (By.CSS_SELECTOR, ".quote")
        )
    )

    for quote in quotes:

        quote_text = quote.find_element(
            By.CSS_SELECTOR,
            ".text"
        ).text

        author = quote.find_element(
            By.CSS_SELECTOR,
            ".author"
        ).text

        tags = quote.find_elements(
            By.CSS_SELECTOR,
            ".tags .tag"
        )

        tag_names = [tag.text for tag in tags]

        quote_data = {
            "Quote": quote_text,
            "Author": author,
            "Tags": tag_names
        }

        all_quotes.append(quote_data)

    print("Total Quotes:", len(all_quotes))

    for item in all_quotes:
        print(item)

finally:

    driver.quit()
```

```python
# Step 7: Export to CSV Using Pandas

import pandas as pd

df = pd.DataFrame(all_quotes)

df["Tags"] = df["Tags"].apply(
    lambda tags: ", ".join(tags)
)

df.to_csv(
    "javascript_quotes.csv",
    index=False
)

print("CSV Created Successfully!")
```

```python
# 13. Selenium + BeautifulSoup Together
from selenium import webdriver
from bs4 import BeautifulSoup

driver = webdriver.Chrome()

try:

    driver.get("https://quotes.toscrape.com/js/")

    html = driver.page_source

    soup = BeautifulSoup(html, "html.parser")

    quotes = soup.select(".quote")

    for quote in quotes:

        text = quote.select_one(".text").get_text(strip=True)

        author = quote.select_one(".author").get_text(strip=True)

        print("Quote:", text)
        print("Author:", author)

finally:

    driver.quit()
```
