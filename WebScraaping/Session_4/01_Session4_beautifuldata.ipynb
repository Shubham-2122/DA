```python
import requests
from bs4 import BeautifulSoup

#find method use : website mathi elements joei ye 
html = """
<html>
<body>

    <h1>Welcome to Web Scraping</h1>

    <p>Python is easy.</p>

    <p>BeautifulSoup is useful.</p>

</body>
</html>
"""

soup = BeautifulSoup(html, "html.parser")

heading = soup.find("h1")

print(heading.text)
```

```python
# find a pragraph
# find method for first paragrph aapse
paragraph = soup.find("p")

print(paragraph.text)
```

```python
# find by class


price = soup.find("p", class_="price")

print(price.text)
```

```python
# finde by id
heading = soup.find(id="main-heading")

print(heading.text)
```

```python
# find by all method

html = """
<html>
<body>

    <p>Python</p>
    <p>Java</p>
    <p>JavaScript</p>

</body>
</html>
"""

soup = BeautifulSoup(html, "html.parser")

paragraphs = soup.find_all("p")

for p in paragraphs:
    print(p.text)
```

```python
html = """
<h1>Web Scraping</h1>
<h2>Python</h2>
<h2>BeautifulSoup</h2>
<h3>Practical</h3>
"""

soup = BeautifulSoup(html, "html.parser")

headings = soup.find_all(["h1", "h2", "h3"])

for heading in headings:
    print(heading.text)
```

```python
html = """
<h1 id="main-heading">Welcome</h1>
<p class="student">Rahul</p>
<p class="student">Amit</p>
<p class="student">Priya</p>
"""

soup = BeautifulSoup(html, "html.parser")

students = soup.find_all(
    "p",
    class_="student"
)

for student in students:
    print(student.text)
```

```python
# find()	find_all()
# First matching element આપે	બધા matching elements આપે
# Single Tag અથવા None	ResultSet (list-like collection)
# First element માટે	Multiple elements માટે
# soup.find("p")	soup.find_all("p")
```

```python
# First paragraph
first = soup.find("p")
print(first.text)

# All paragraphs
all_paragraphs = soup.find_all("p")

for p in all_paragraphs:
    print(p.text)
```

```python
students = soup.select(".student")

for student in students:
    print(student.text)
```

```python
heading = soup.select("#main-heading")

print(heading[0].text)
```

```python
# Selector	Meaning
# p	All paragraph tags
# .price	Elements with price class
# #main	Element with main ID
# div.product	div with product class
# a[href]	Anchor with href attribute
# .product .price	Descendant price inside product
# div > p	Direct child paragraph of div
# h1, h2	h1 and h2 elements
```

```python
# find() vs CSS select()
# find()	select()
# BeautifulSoup search method	CSS selector-based search
# Tag, class_, id arguments	CSS selector string
# First match	All matches
# find("p", class_="price")	select(".price")
```

```python
links = soup.find_all("a")

for link in links:

    text = link.get_text(strip=True)

    href = link.get("href")

    print("Link Text:", text)
    print("URL:", href)
```

```python
# Important Attribute Methods:
# Method	Purpose
# element.get("href")	href value
# element.get("src")	src value
# element.get("alt")	alt value
# element.attrs	All attributes dictionary
# element["href"]	Direct attribute access
```
