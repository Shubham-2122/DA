```python
# Session 9.4 — Pie Chart & Donut Chart in Matplotlib
# આજે આપણે શીખીશું કે કુલ data માં દરેક category નો કેટલો હિસ્સો છે તે Pie Chart અને Donut Chart દ્વારા કેવી રીતે દર્શાવવો.

# ઉદાહરણ તરીકે, college માં IT, Computer Science અને Mechanical Engineering ના વિદ્યાર્થીઓનું percentage દર્શાવવું હોય અથવા company ની કુલ sales માં દરેક product નો હિસ્સો જાણવો હોય, ત્યારે આ charts ઉપયોગી થઈ શકે છે.
# 1. Learning objectives
# આ session પૂર્ણ થયા પછી તમે:
# - plt.pie() થી Pie Chart બનાવી શકશો.
# - autopct વડે percentage બતાવી શકશો.
# - labels, colors, startangle અને explode નો ઉપયોગ કરી શકશો.
# - Donut Chart બનાવી શકશો.
# - યોગ્ય data માટે Pie Chart અને Bar Chart વચ્ચે પસંદગી કરી શકશો.
# 2. Pie Chart શું છે?
# English definition: A pie chart represents how individual categories contribute to a whole, using slices of a circle.

# ગુજરાતીમાં, કુલ data ને એક વર્તુળ તરીકે દર્શાવી, દરેક category નો હિસ્સો અલગ slice માં બતાવતા chart ને Pie Chart કહેવામાં આવે છે.

# pie cahrt slice 

# percentage = category value / total value * 100 
```

```python
# 3. Step 1 — Basic Pie Chart બનાવો

import matplotlib.pyplot as plt 

departments = ["IT", "CS", "Mechanical"]
students = [40, 35, 25]

plt.pie(students)

plt.title("Department-wise Students")
plt.show()

# આ code ત્રણ slices ધરાવતો Pie Chart બનાવશે.
```

```python
# 4. Step 2 — Labels અને Percentage ઉમેરો

import matplotlib.pyplot as plt

departments = ["IT", "CS", "Mechanical"]
students = [40, 35, 25]

plt.figure(figsize=(7, 7))

plt.pie(
    students,
    labels=departments,
    autopct="%1.1f%%" # percetage wise jova
)

plt.title("Department-wise Student Distribution")
plt.show()
```

```python
# હત્વના parameters:
# Parameter	ઉપયોગ
# labels	દરેક slice નું નામ
# autopct	દરેક slice નું percentage
# startangle	Chart કયા angle થી શરૂ થાય તે
# colors	Slice ના રંગ
# explode	પસંદ કરેલી slice ને બહાર ખસેડવી
# shadow	Slice પાછળ shadow બતાવવો
# counterclock	Slices ઘડિયાળની દિશામાં કે વિરુદ્ધ દિશામાં દોરવા

# %1.1f percentage ને એક decimal place સુધી દર્શાવે છે.

# છેલ્લું %% output માં % symbol દર્શાવે છે.

# autopct="%1.0f%%"  # 40%
# autopct="%1.2f%%"  # 40.00%

```

```python
# 5. Step 3 — Colors અને Start Angle

import matplotlib.pyplot as plt

departments = ["IT", "CS", "Mechanical"]
students = [40, 35, 25]

colors = ["skyblue", "orange", "lightgreen"]

plt.figure(figsize=(7,7))

plt.pie(
    students,
    labels=departments,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Department-wise Student Distrubution")
plt.show()

# olors દરેક slice નો રંગ નક્કી કરે છે.
# startangle=90 chart ના પ્રથમ slice ની શરૂઆતનો angle બદલે છે.
# startangle બદલવાથી percentages બદલાતા નથી; માત્ર chart ની orientation બદલાય છે.
```

```python
# 6. Step 4 — explode વડે Slice Highlight કરો
# કોઈ એક category પર ધ્યાન ખેંચવું હોય ત્યારે explode વાપરી શકાય છે.

import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet"]
sales = [50, 30, 20]

explode = (0.1, 0, 0) # vadhre bhag door kari loko ae tarf jovu

plt.figure(figsize=(7, 7))

plt.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    explode=explode,
    startangle=90,
    shadow=True
)

plt.title("Product Sales Share")
plt.show()

# explode=(0.1, 0, 0) પ્રથમ slice ને કેન્દ્રથી થોડું દૂર ખસેડે છે.
# shadow=True shadow effect ઉમેરે છે.
```

```python
# 7. Step 5 — Donut Chart બનાવો
# Donut Chart મૂળભૂત રીતે મધ્યમાં ખાલી જગ્યા ધરાવતો Pie Chart છે.

# Matplotlib માં wedgeprops નો ઉપયોગ કરીને આ ખાલી જગ્યા બનાવી શકાય છે.

import matplotlib.pyplot as plt

departments = ["IT", "CS", "Mechanical"]
students = [40, 35, 25]

plt.figure(figsize=(7, 7))

plt.pie(
    students,
    labels=departments,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.4}
)

plt.title("Department-wise Student Distribution")
plt.show()

wedgeprops={"width": 0.4} slices ની જાડાઈ નક્કી કરે છે.



# મધ્યમાં ખાલી જગ્યા દેખાય છે.
# width નું મૂલ્ય બદલવાથી donut ની જાડાઈ બદલાય છે.
```

```python
# Donut ના કેન્દ્રમાં Total બતાવવું

import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet"]
sales = [50000, 30000, 20000]

fig, ax = plt.subplots(figsize=(7, 7))

ax.pie(
    sales,
    labels=products,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.4}
)

total = sum(sales)

ax.text(
    0, 0,
    f"Total\n₹{total:,}",
    ha="center",
    va="center",
    fontsize=13,
    fontweight="bold"
)

ax.set_title("Product Sales Distribution")
plt.show()
```

```python
# 8. Step 6 — Pandas DataFrame સાથે Pie Chart
# ચાલો, Pandas માંથી department-wise student count કાઢીને Pie Chart બનાવીએ.

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Department": ["IT", "CS", "IT", "Mechanical",
                   "CS", "IT", "Mechanical", "CS"],
    "StudentID": [101, 102, 103, 104, 105, 106, 107, 108]
})

dept_counts = df["Department"].value_counts()

print(dept_counts)

plt.figure(figsize=(7, 7))

plt.pie(
    dept_counts.values,
    labels=dept_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Department-wise Student Share")
plt.show()

# આ code માં:
# - value_counts() દરેક department ના records ગણે છે.
# - dept_counts.index department names આપે છે.
# - dept_counts.values તેની સંખ્યા આપે છે.
```

```python
# . Pie Chart ક્યારે વાપરવો?
# પરિસ્થિતિ	યોગ્ય પસંદગી
# કુલ sales માં product નો percentage	Pie અથવા Donut Chart
# 3–5 categories ના proportions	Pie Chart
# અનેક categories ની સરખામણી	Bar Chart
# બહુ નાનાં percentage differences	Bar Chart સામાન્ય રીતે વધુ સ્પષ્ટ
# સમય પ્રમાણે sales નો trend	Line Chart
```
