```python
# Session 9.7 — Subplots & Multi-Panel Figures in Matplotlib
# આજે આપણે શીખીશું કે એક જ Figure માં એક કરતાં વધુ charts કેવી રીતે બતાવવા. Data Science માં આ ખૂબ ઉપયોગી છે, કારણ કે એક જ dataset નું અલગ-અલગ charts વડે analysis કરી શકાય છે.

# ઉદાહરણ તરીકે, એક જ Sales dataset માટે:
# - Line Chart — મહિના પ્રમાણે Sales નો trend.
# - Bar Chart — Category પ્રમાણે Sales.
# - Histogram — Sales values નું distribution.
# - Scatter Plot — Advertising અને Sales વચ્ચેનો સંબંધ.
# 1. Learning Objectives
# આ Session પૂર્ણ થયા પછી તમે:
# - plt.subplot() અને plt.subplots() નો ઉપયોગ કરી શકશો.
# - એક Figure માં 2, 3 કે 4 charts બનાવી શકશો.
# - figsize, sharex, sharey અને tight_layout() સમજી શકશો.
# - ax[0], ax[1] દ્વારા અલગ charts ને control કરી શકશો.
# - એક નાનું multi-chart dashboard બનાવી શકશો.

# Figure: આખું canvas અથવા સંપૂર્ણ chart window.
# Axes: Figure ની અંદરનો એક plot વિસ્તાર, જેમાં data દોરવામાં આવે છે.
# Axis: દરેક Axes માં X-axis અને Y-axis હોય છે.
```

```python
# 3. Method 1 — plt.subplot()
# plt.subplot(rows, columns, position) દ્વારા Figure માં chart ની જગ્યા પસંદ થાય છે.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]
profit = [20, 35, 25, 45]

# First chart: 1 row, 2 columns, position 1
plt.subplot(1, 2, 1)
plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

# Second chart: 1 row, 2 columns, position 2
plt.subplot(1, 2, 2)
plt.bar(months, profit)
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")

plt.tight_layout()
plt.show()


# plt.subplot(1, 2, 1) નો અર્થ
# Value	અર્થ
# 1	1 row
# 2	2 columns
# 1	પ્રથમ chart ની position

# આ જ રીતે plt.subplot(2, 2, 3) નો અર્થ છે: 2 rows, 2 columns અને ત્રીજી position.

# plt.subplot() નાના examples માટે ઉપયોગી છે, પરંતુ વધુ charts અને customization માટે plt.subplots() વધુ અનુકૂળ છે.
```

```python
# 4. Method 2 — plt.subplots() (Recommended)
# આ પદ્ધતિમાં Figure અને Axes બંને એકસાથે મળે છે.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]
profit = [20, 35, 25, 45]

fig,ax = plt.subplots(1, 2, figsize=(10, 4))

ax[0].plot(months,sales,marker="o")
ax[0].set_title("Monthly Sales")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")

ax[1].bar(months, profit)
ax[1].set_title("Monthly Profit")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Profit")

fig.suptitle("Business Performance Dashboard")
fig.tight_layout()
plt.show()


# Code	અર્થ
# fig, ax = plt.subplots()	Figure અને Axes બનાવે છે
# 1, 2	1 row અને 2 columns
# figsize=(10, 4)	Figure ની પહોળાઈ 10 અને ઊંચાઈ 4 inches
# ax[0]	પ્રથમ chart
# ax[1]	બીજો chart
# ax[0].set_title()	પ્રથમ chart નું title
# fig.suptitle()	આખી Figure નું મુખ્ય title
# fig.tight_layout()	Charts વચ્ચેનું spacing ગોઠવે છે


```

```python
# 5. 2 × 2 Dashboard — ચાર Charts એકસાથે
# હવે એક જ dataset પરથી Line, Bar, Histogram અને Scatter Plot બનાવીએ.

import matplotlib.pyplot as plt
import numpy as np

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]
profit = [20, 25, 22, 35, 40, 48]
advertising = [10, 15, 12, 20, 25, 30]

fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# 1. Line Chart
ax[0, 0].plot(months, sales, marker="o")
ax[0, 0].set_title("Monthly Sales Trend")
ax[0, 0].set_ylabel("Sales")
ax[0, 0].grid(True, alpha=0.3)

# 2. Bar Chart
ax[0, 1].bar(months, profit)
ax[0, 1].set_title("Monthly Profit")
ax[0, 1].set_ylabel("Profit")

# 3. Histogram
ax[1, 0].hist(sales, bins=5, edgecolor="black")
ax[1, 0].set_title("Sales Distribution")
ax[1, 0].set_xlabel("Sales")
ax[1, 0].set_ylabel("Frequency")

# 4. Scatter Plot
ax[1, 1].scatter(advertising, sales, s=70)
ax[1, 1].set_title("Advertising vs Sales")
ax[1, 1].set_xlabel("Advertising")
ax[1, 1].set_ylabel("Sales")

fig.suptitle("Sales Analysis Dashboard", fontsize=16)
fig.tight_layout()
plt.show()
```

```python
# 6. sharex અને sharey શું છે?
# કેટલાક charts માં એક જ X-axis અથવા Y-axis scale share કરવાની જરૂર પડે છે.

import matplotlib.pyplot as plt

months = [1, 2, 3, 4, 5, 6]
sales_2025 = [100, 120, 140, 130, 160, 180]
sales_2026 = [110, 125, 150, 145, 175, 200]

fig, ax = plt.subplots(
    2, 1,
    figsize=(8, 6),
    sharex=True,
    sharey=True
)

ax[0].plot(months, sales_2025, marker="o")
ax[0].set_title("Sales 2025")
ax[0].set_ylabel("Sales")

ax[1].plot(months, sales_2026, marker="o")
ax[1].set_title("Sales 2026")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Sales")

fig.tight_layout()
plt.show()
```

```python
# . Subplots માં ઉપયોગી Parameters
# Parameter	ઉપયોગ	Example
# nrows	Rows ની સંખ્યા	2
# ncols	Columns ની સંખ્યા	2
# figsize	Figure નું કદ	(12, 8)
# sharex	X-axis share કરવો	True
# sharey	Y-axis share કરવો	True
# tight_layout()	Spacing સુધારવું	fig.tight_layout()
# suptitle()	સંપૂર્ણ Figure નું title	fig.suptitle("Report")

fig, ax = plt.subplots(1, 2, squeeze=False)

ax[0, 0].plot([1, 2, 3], [10, 20, 15])
ax[0, 1].bar(["A", "B", "C"], [10, 20, 15])

fig.tight_layout()
plt.show()
```
