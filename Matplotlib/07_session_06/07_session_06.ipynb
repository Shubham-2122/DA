```python
# # Session 9.6 — Scatter Plot & Trend Line in Matplotlib
# # આજે આપણે Data Visualization માં એક ખૂબ જ ઉપયોગી chart શીખીશું: Scatter Plot. તેનો ઉપયોગ બે numerical variables વચ્ચેનો સંબંધ સમજવા માટે થાય છે. ત્યારબાદ NumPy ની મદદથી Trend Line બનાવીશું, જેથી data માં દેખાતો overall pattern સમજી શકાય.

# # ઉદાહરણ તરીકે:
# # - વિદ્યાર્થી જેટલા વધુ કલાક અભ્યાસ કરે છે, તેના marks વધે છે કે નહીં?
# # - Advertising ખર્ચ વધે ત્યારે Sales વધે છે કે નહીં?
# # - Temperature વધે ત્યારે Ice Cream Sales વધે છે કે નહીં?
# # 1. Learning Objectives
# # આ Session પૂર્ણ થયા પછી તમે:
# # - plt.scatter() વડે Scatter Plot બનાવી શકશો.
# # - X અને Y variables વચ્ચેનો સંબંધ સમજાવી શકશો.
# # - Positive, Negative અને No Correlation ઓળખી શકશો.
# # - color, marker, s અને alpha parameters વાપરી શકશો.
# # - np.polyfit() વડે Linear Trend Line બનાવી શકશો.
# # - Scatter Plot દ્વારા practical data analysis કરી શકશો.
# # 2. Scatter Plot શું છે?
# # Scatter Plot માં દરેક observation ને એક dot તરીકે દર્શાવવામાં આવે છે.

# # માની લો કે આપણી પાસે પાંચ વિદ્યાર્થીઓના Study Hours અને Marks છે:

# Student	Study Hours (X)	Marks (Y)
# Amit	1	35
# Neha	2	42
# Ravi	3	50
# Pooja	4	65
# Karan	5	78

# અહીં:
# - X-axis: અભ્યાસના કલાકો.
# - Y-axis: પરીક્ષાના marks.
# - દરેક dot: એક વિદ્યાર્થીનું observation.
# જો અભ્યાસના કલાકો વધવાની સાથે marks પણ વધે, તો બંને variables વચ્ચે Positive Relationship હોવાનો સંકેત મળે છે.
```

```python
# 3. Basic Scatter Plot — plt.scatter()
# VS Code ની Jupyter Notebook માં આ code ચલાવો.

import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [30, 35, 45, 50, 60, 65, 75, 85]

plt.scatter(study_hours,marks)

plt.title("Study Hours vs Exam Marks")

plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")

plt.grid(True)

plt.show()

# Code	અર્થ
# plt.scatter(x, y)	X અને Y values માટે dots દોરે છે
```

```python
# 4. Scatter Plot ના Parameters
# હવે chart ને વધુ સુંદર અને સમજવામાં સરળ બનાવીએ.

import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [30, 35, 45, 50, 60, 65, 75, 85]

plt.scatter(
    study_hours,
    marks,
    color="royalblue",
    marker="o",
    s=100, # size vadar va dot
    alpha=0.8,
    edgecolor="black"
)

plt.title("Study Hours vs Exam Marks")
plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# # 5. Correlation — બે variables વચ્ચેનો સંબંધ
# # Scatter Plot માં dots કઈ દિશામાં ફેલાયેલા છે તેના આધારે variables વચ્ચેનો સંબંધ સમજી શકાય છે.

# Positive Correlation

# X વધે ત્યારે Y સામાન્ય રીતે વધે છે.

# ઉદાહરણ: Study Hours વધે ત્યારે Marks વધવાની શક્યતા.


# યાદ રાખો: Correlation એટલે સંબંધ; તે કારણ અને પરિણામ (causation) સાબિત કરતું નથી.
```

```python
# 6. Correlation Coefficient — np.corrcoef()
# Correlation ને આંકડાકીય રીતે માપવા માટે NumPy નો ઉપયોગ કરી શકીએ છીએ.

import numpy as np

study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [30, 35, 45, 50, 60, 65, 75, 85]

correlation = np.corrcoef(study_hours,marks)[0,1]

print("Corrleation : ",round(correlation,2))

# Correlation coefficient સામાન્ય રીતે \(-1\) થી \(+1\) વચ્ચે હોય છે.
# Value	સામાન્ય અર્થ
# +1 ની નજીક	મજબૂત Positive Linear Relationship
# 0 ની નજીક	નબળો અથવા કોઈ Linear Relationship નહીં
# -1 ની નજીક	મજબૂત Negative Linear Relationship
```

```python
# 7. Trend Line શું છે?
# Trend Line એ data points ની overall direction દર્શાવતી સીધી રેખા છે. Linear Trend Line નો ઉપયોગ કરીને આપણે જોઈ શકીએ છીએ કે data સામાન્ય રીતે ઉપર જાય છે કે નીચે.

# આપણે NumPy ના np.polyfit() function વડે Trend Line બનાવીશું.

# Step 1 — Data તૈયાર કરો

import numpy as np
import matplotlib.pyplot as plt

study_hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
marks = np.array([30, 35, 45, 50, 60, 65, 75, 85])

# Step 2 — Regression Line ની ગણતરી કરો

slope, intercept = np.polyfit(study_hours, marks, 1)

print("Slope:", round(slope, 2))
print("Intercept:", round(intercept, 2))

# x: Independent variable, અહીં Study Hours.
# y: Dependent variable, અહીં Marks.
# 1: Linear polynomial, એટલે સીધી રેખા.
# slope: રેખાનો ઢાળ.
# intercept: X = 0 હોય ત્યારે રેખાની અંદાજિત Y value.

# y = mx + b
```

```python
# Step 3 — Trend Line બનાવો

trend_marks = slope * study_hours + intercept

plt.scatter(
    study_hours,
    marks,
    color="royalblue",
    s=80,
    label="Actual Marks"
)

plt.plot(
    study_hours,
    trend_marks,
    color="red",
    linestyle="--",
    linewidth=2,
    label="Trend Line"
)

plt.title("Study Hours vs Marks with Trend Line")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()

# વાદળી dots = Actual Marks.
# લાલ dashed line = Linear Trend Line.
# plt.scatter() actual observations દર્શાવે છે.
# plt.plot() Trend Line દોરે છે.
# plt.legend() બંનેને ઓળખવામાં મદદ કરે છે.
```

```python
# 8. Real-World Practical — Advertising અને Sales
# માની લો કે એક કંપની જુદા-જુદા દિવસોમાં Advertising પર અલગ ખર્ચ કરે છે અને Sales નોંધે છે.

import numpy as np
import matplotlib.pyplot as plt

advertising = np.array([10, 20, 30, 40, 50, 60, 70, 80])
sales = np.array([15, 22, 28, 35, 45, 48, 60, 72])

# Correlation
corr = np.corrcoef(advertising, sales)[0, 1]

# Trend Line
slope, intercept = np.polyfit(advertising, sales, 1)

# Sorted X values for a clean line
x_line = np.linspace(advertising.min(), advertising.max(), 100)
y_line = slope * x_line + intercept

# Visualization
plt.scatter(
    advertising,
    sales,
    color="green",
    s=80,
    label="Actual Sales"
)

plt.plot(
    x_line,
    y_line,
    color="red",
    linestyle="--",
    label="Linear Trend"
)

plt.title(f"Advertising vs Sales (r = {corr:.2f})")
plt.xlabel("Advertising Spend (units)")
plt.ylabel("Sales (units)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

print("Correlation:", round(corr, 2))
print("Slope:", round(slope, 2))


# આ પરિણામમાંથી શું શીખી શકાય?
# 1. Dots નો pattern ઉપર જતો હોય તો Positive Relationship નો સંકેત મળે છે.
# 2. Correlation coefficient સંબંધની દિશા અને linear strength સમજવામાં મદદ કરે છે.
# 3. Trend Line data નો overall linear trend દર્શાવે છે.
# 4. આ ઉદાહરણથી Advertising જ Sales વધારવાનું કારણ છે એવું સાબિત થતું નથી; અન્ય factors પણ અસર કરી શકે છે.
```
