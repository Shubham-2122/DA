```python
# 6. Step 2 — તમારો પહેલો chart
# આપણે એક સરળ line chart બનાવીએ.

import matplotlib.pyplot as plt 

months = ["Jan","Feb","Mar","Apr","May"]
sales = [10000,15000,12000,20000,25000]

plt.plot(months,sales) #line chart banavse

plt.show() # chart show karse

# - X-axis પર months છે.
# - Y-axis પર sales છે.
# - Line દર્શાવે છે કે sales સમય પ્રમાણે કેવી રીતે બદલાઈ.
# plt.plot() line chart બનાવે છે અને plt.show() chart ને display કરે છે.
```

```python
# 7. Step 3 — Chart ના title, labels અને grid ઉમેરો
# હવે chart ને વધુ professional બનાવીએ.

import matplotlib.pyplot as plt 

months = ["Jan","Feb","Mar","Apr","May"]
sales = [10000,15000,12000,20000,25000]

plt.figure(figsize=(8,4)) # chart ni figure set

plt.plot(months,sales,marker="o") # point par dot aapva

plt.title("Monthly Sales Analysis") # titel
plt.xlabel("Month") # x par name
plt.ylabel("Sales ($)") # y par name

plt.grid(True) # background ma grid line show karse

plt.show()

# Function	ઉપયોગ
# plt.figure(figsize=(8, 4))	Chart માટે figure નું કદ નક્કી કરે છે
# plt.title()	Chart નું title આપે છે
# plt.xlabel()	X-axis નું label આપે છે
# plt.ylabel()	Y-axis નું label આપે છે
# plt.grid(True)	Background માં grid lines બતાવે છે
# marker="o"	દરેક data point પર ગોળ marker બતાવે છે
# plt.show()	Chart display કરે છે
```

```python
# # 8. Figure, Axes અને Axis વચ્ચેનો તફાવત
# # આ Matplotlib નો ખૂબ જ મહત્વપૂર્ણ concept છે.

# 1. Figure
# આખું canvas અથવા container, જેમાં એક અથવા વધુ charts હોઈ શકે છે.

# 2. Axes
# Chart નો actual plotting area, જેમાં data, title, labels, legend અને grid હોય છે. એક Figure માં અનેક Axes હોઈ શકે છે.

# 3. Axis
# X-axis અને Y-axis જેવી coordinate lines, ticks અને tick labels જે data નું માપ દર્શાવે છે.

# યાદ રાખવાની trick:
# - Figure = આખું canvas
# - Axes = chart નું plotting area
# - Axis = X અથવા Y માપ દર્શાવતી દિશા
# Object-oriented રીત
# Matplotlib માં chart બનાવવા માટે આ રીત પણ ખૂબ ઉપયોગી છે:

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [10000, 15000, 12000, 20000, 25000]

fig,ax = plt.subplots(figsize=(8,4))

ax.plot(months,sales,marker="o")
ax.set_title("Monthly Sales Analysis")
ax.set_xlabel("Month")
ax.set_ylabel("Sales $")
ax.grid(True)

plt.show()
```

```python
# 10. Hands-on Practical
# હવે નીચેનો code જાતે લખો અને ચલાવો.

import matplotlib.pyplot as plt

subjects = ["Python", "SQL", "Pandas", "NumPy", "Matplotlib"]
marks = [80, 75, 90, 85, 95]

plt.figure(figsize=(9,5))

plt.plot(
    subjects,
    marks,
    marker="o",
    linewidth=2,
    label="Student Marks"
)

plt.title("Student-wise Perfromance")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
plt.grid(True)
plt.show()
```
