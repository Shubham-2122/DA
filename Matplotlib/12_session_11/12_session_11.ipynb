```python
# Session 9.11 — Exporting & Saving Plots
# આજે આપણે Matplotlib માં બનાવેલા charts ને image અથવા document file તરીકે save કરવાનું શીખીશું, જેથી તમે તેને PowerPoint presentation, PDF report, project documentation અને business dashboard માં ઉપયોગ કરી શકો.
# 1. Learning objectives
# આ session પૂર્ણ થયા પછી તમે શીખી શકશો:
# - plt.savefig() અને fig.savefig() નો ઉપયોગ.
# - PNG, JPG, SVG અને PDF માં chart save કરવો.
# - dpi વડે image resolution નિયંત્રિત કરવું.
# - bbox_inches="tight" અને pad_inches સમજવા.
# - White background અને transparent background સાથે chart export કરવો.
# - VS Code/Jupyter માં file ક્યાં save થઈ છે તે શોધવું.
```

```python
# 2. Basic chart save કરવો
# સૌપ્રથમ એક સરળ chart બનાવીએ અને તેને PNG file તરીકે save કરીએ

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]


fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(months, sales, marker="o", label="Sales")
ax.set_title("Monthly Sales Report")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.legend()
ax.grid(alpha=0.3)

# Chart ને file તરીકે save કરો
fig.savefig("monthly_sales.png")

plt.show()
```

```python
# 3. plt.savefig() અને fig.savefig() વચ્ચેનો તફાવત
# Function	સમજણ
# plt.savefig("chart.png")	હાલની active figure ને save કરે છે
# fig.savefig("chart.png")	ચોક્કસ fig object ને save કરે છે

fig.savefig("sales_report.png", dpi=300)
fig.savefig("sales_report.jpg", dpi=300)
fig.savefig("sales_report.svg")
fig.savefig("sales_report.pdf")
```

```python
# 5. dpi — Image ની resolution નિયંત્રિત કરવી
# dpi નો અર્થ Dots Per Inch છે. Matplotlib માં raster image જેમ કે PNG અથવા JPG કેટલી detail સાથે render થાય છે તે નિયંત્રિત કરવામાં તેનો ઉપયોગ થાય છે.

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 25, 18, 35, 45]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(x, y, marker="o")
ax.set_title("Performance Report")
ax.set_xlabel("Week")
ax.set_ylabel("Performance")

fig.savefig("chart_low.png", dpi=100)
fig.savefig("chart_high.png", dpi=300)

plt.show()
```

```python
# Setting	સામાન્ય ઉપયોગ
# dpi=72 અથવા 100	ઝડપી preview અથવા સામાન્ય ઉપયોગ
# dpi=150	સામાન્ય report
# dpi=300	High-resolution report અને print માટે

```

```python
# 6. bbox_inches="tight" — વધારાની ખાલી જગ્યા દૂર કરવી
# ક્યારેક chart save કરતી વખતે title, labels અથવા legend ની આસપાસ વધારે જગ્યા આવી જાય છે. તેને નિયંત્રિત કરવા bbox_inches="tight" ઉપયોગી છે.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 150, 130, 190, 220]

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(months, sales, marker="o", label="Sales")
ax.set_title("Monthly Sales Performance")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

ax.legend(
    loc="upper left",
    bbox_to_anchor=(1.02, 1)
)

fig.savefig(
    "sales_tight.png",
    dpi=300,
    bbox_inches="tight",
    pad_inches=0.2
)

plt.show()

# અહીં:
# - bbox_inches="tight" — chart ની બહારના જરૂરી elements સહિત figure ની boundary ગોઠવે છે.
# - pad_inches=0.2 — boundary ની આસપાસ થોડી વધારાની જગ્યા રાખે છે.
# Legend ને chart ની બહાર મૂક્યું હોય ત્યારે આ ખાસ ઉપયોગી છે
```

```python
# 7. Transparent background સાથે save કરવું
# ક્યારેક chart ને રંગીન PowerPoint slide અથવા website પર મૂકવો હોય અને chart ની પાછળનો સફેદ background દૂર રાખવો હોય છે.
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    [1, 2, 3, 4],
    [10, 25, 20, 40],
    marker="o",
    label="Growth"
)

ax.set_title("Growth Report")
ax.legend()

fig.savefig(
    "growth_transparent.png",
    dpi=300,
    transparent=True,
    bbox_inches="tight"
)

plt.show()
```

```python
# 9. Complete practical — Professional Sales Report
# હવે annotation, KPI, target l
# ine અને export settings ને એક સાથે જોડીએ

```

```python
import os
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]
target = 180

x = list(range(len(months)))
max_idx = sales.index(max(sales))

# Output folder બનાવો
os.makedirs("reports", exist_ok=True)

fig, ax = plt.subplots(figsize=(10, 6))

# Sales chart
ax.plot(
    x, sales,
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

# Target line
ax.axhline(
    y=target,
    linestyle="--",
    label=f"Target = {target}"
)

# Peak sales highlight
ax.scatter(
    max_idx, sales[max_idx],
    s=140,
    label="Peak Sales"
)

# Annotation
ax.annotate(
    f"Peak: {sales[max_idx]}",
    xy=(max_idx, sales[max_idx]),
    xytext=(max_idx - 1.5, sales[max_idx] + 18),
    arrowprops=dict(arrowstyle="->"),
    ha="center"
)

# KPI box
ax.text(
    0.03, 0.95,
    f"Total Sales: {sum(sales)}\nAverage: {sum(sales) / len(sales):.1f}",
    transform=ax.transAxes,
    va="top",
    bbox=dict(boxstyle="round", facecolor="white", alpha=0.9)
)

ax.set_xticks(x, months)
ax.set_title("Monthly Sales Performance Report")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.legend(loc="lower right")
ax.grid(alpha=0.3)

fig.tight_layout()

# અલગ formats માં save કરો
fig.savefig(
    "reports/sales_report.png",
    dpi=300,
    bbox_inches="tight"
)

fig.savefig(
    "reports/sales_report.pdf",
    bbox_inches="tight"
)

fig.savefig(
    "reports/sales_report.svg",
    bbox_inches="tight"
)

plt.show()

print("Chart files saved in the reports folder.")
print("Current working directory:", os.getcwd())
```
