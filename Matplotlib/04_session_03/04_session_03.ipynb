```python
# Session 9.3 — Bar Charts (Vertical + Horizontal) in Matplotlib
# આજે આપણે Matplotlib માં Vertical Bar Chart અને Horizontal Bar Chart બનાવતાં શીખીશું. આ charts ખાસ કરીને અલગ-અલગ categories વચ્ચે comparison કરવા માટે ઉપયોગી છે.
# 1. Learning objectives
# આ session પૂર્ણ થયા પછી તમે:
# - plt.bar() થી vertical bar chart બનાવી શકશો.
# - plt.barh() થી horizontal bar chart બનાવી શકશો.
# - Bar નો રંગ, width અને height customize કરી શકશો.
# - દરેક bar ઉપર તેની value બતાવી શકશો.
# - Pandas DataFrame માંથી category-wise analysis chart માં દર્શાવી શકશો.
# 2. Bar Chart શું છે?
# English definition: A bar chart represents categorical data using rectangular bars, where the bar length or height corresponds to a numerical value.

# ગુજરાતીમાં, અલગ-અલગ categories ની numerical values ની સરખામણી લંબચોરસ bars વડે દર્શાવતા chart ને Bar Chart કહેવામાં આવે છે.

# ઉદાહરણ તરીકે, પાંચ વિષયોમાં વિદ્યાર્થીના marks:
# Subject	Marks
# Python	80
# SQL	75
# Pandas	90
# NumPy	85
# Matplotlib	95

# આ data માં કયા subject માં સૌથી વધુ marks છે તે Bar Chart વડે સરળતાથી જોઈ શકાય છે
```

```python
# 3. Step 1 — Vertical Bar Chart: plt.bar()
# plt.bar() vertical bars બનાવે છે.

import matplotlib.pyplot as plt

subjects = ["Python", "SQL", "Pandas", "NumPy", "Matplotlib"]
marks = [80, 75, 90, 85, 95]

plt.figure(figsize=(9, 5))

plt.bar(subjects,marks)

plt.title("Subject-wise Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")

plt.show()

# X-axis પર subjects છે.
# Y-axis પર marks છે.
# Bar જેટલો ઊંચો, તેની value એટલી વધારે.
```

```python
# 4. Step 2 — Bar Chart ના મહત્વના parameters
# Parameter	ઉપયોગ	Example
# color	Bar નો રંગ	color="green"
# width	Bar ની પહોળાઈ	width=0.6
# edgecolor	Bar ની border નો રંગ	edgecolor="black"
# linewidth	Border ની જાડાઈ	linewidth=1
# label	Legend માં નામ	label="Marks"
# alpha	પારદર્શકતા	alpha=0.8


plt.bar(
    subjects,
    marks,
    color="skyblue",
    width=0.6,
    edgecolor="black",
    linewidth=1,
    alpha=0.9
)

plt.title("Customized Bar Chart")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.show()

# width=0.6 નો અર્થ category ના ઉપલબ્ધ plotting space ની સરખામણીમાં bar ની પહોળાઈ નક્કી કરવી. એક જ chart માં ઘણી categories હોય ત્યારે width ઘટાડવાથી bars વચ્ચે વધુ જગ્યા મળે છે
```

```python
# 5. Step 3 — દરેક Bar ઉપર Value બતાવવી
# Business report અને college performance report માં દરેક bar ની value બતાવવાથી chart વાંચવો સરળ બને છે.

import matplotlib.pyplot as plt

subjects = ["Python", "SQL", "Pandas", "NumPy", "Matplotlib"]
marks = [80, 75, 90, 85, 95]

plt.figure(figsize=(9, 5))

bars = plt.bar(
    subjects,
    marks,
    color="cornflowerblue",
    edgecolor="black"
)

plt.bar_label(bars,padding=3)

plt.title("Subject-wise Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.ylim(0, 110)

plt.show()

# ars માં દરેક bar નો reference મળે છે.
# plt.bar_label(bars, padding=3) દરેક bar ઉપર value લખે છે.
# plt.ylim(0, 110) values અને labels માટે ઉપર જગ્યા રાખે છે.
```

```python
# 6. Step 4 — Horizontal Bar Chart: plt.barh()
# જ્યારે category names લાંબા હોય, ત્યારે horizontal bar chart વધુ વાંચી શકાય એવો બને છે.

import matplotlib.pyplot as plt

departments = [
    "Computer Engineering",
    "Information Technology",
    "Mechanical Engineering",
    "Civil Engineering"
]

students = [120, 100, 80, 60]

plt.figure(figsize=(9, 5))

bars = plt.barh(
    departments,
    students,
    color="mediumseagreen"
)

plt.bar_label(bars,padding=4)

plt.title("Students by Department")
plt.xlabel("Number of Students")
plt.ylabel("Department")

plt.show()

# plt.barh() horizontal bars બનાવે છે.
# Department names Y-axis પર આવે છે.
# Student count X-axis પર આવે છે.
# plt.bar_label() દરેક bar ની value બતાવે છે.
```

```python
# Vertical અને Horizontal વચ્ચેનો તફાવત
# Feature	    Vertical	        Horizontal
# Function	    plt.bar()	        plt.barh()
# Categories	X-axis	            Y-axis
# Values	    Y-axis	            X-axis
# શ્રેષ્ઠ ઉપયોગ	    ટૂંકાં category namesલાંબાં category names
```

```python
# 7. Step 5 — Bars ને Sort કરવી
# ચાલો, department ના students ને સૌથી વધુથી ઓછા ક્રમમાં બતાવીએ

import matplotlib.pyplot as plt

departments = ["IT", "CS", "Mechanical", "Civil"]
students = [100, 150, 80, 60]

# વધુ students ધરાવતા department પહેલ
sorted_data = sorted(
    zip(departments, students),
    key=lambda item: item[1],
    reverse=True
)

sorted_departments = [item[0] for item in sorted_data]
sorted_students = [item[1] for item in sorted_data]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    sorted_departments,
    sorted_students,
    color="orange"
)

plt.bar_label(bars, padding=3)
plt.title("Department-wise Student Count")
plt.xlabel("Department")
plt.ylabel("Students")
plt.show()
```

```python
# 8. Step 6 — Pandas DataFrame સાથે Bar Chart
# હવે Pandas ના groupby() નો ઉપયોગ કરીને department-wise average marks કાઢીએ અને chart બનાવીએ.

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Department": ["IT", "CS", "IT", "CS", "IT", "CS"],
    "Student": ["Amit", "Neha", "Ravi", "Pooja", "Karan", "Mira"],
    "Marks": [70, 85, 65, 95, 80, 75]
})

dept_summary = (
    df.groupby("Department", as_index=False)
      .agg(
          Average_Marks=("Marks", "mean"),
          Student_Count=("Student", "count")
      )
)

print(dept_summary)

plt.figure(figsize=(7, 4))

bars = plt.bar(
    dept_summary["Department"],
    dept_summary["Average_Marks"],
    color=["steelblue", "coral"]
)

plt.bar_label(bars, fmt="%.1f", padding=3)

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.ylim(0, 110)

plt.show()
```

```python
# 9. Step 7 — બે પ્રકારના Bar Charts એક સાથે
# ક્યારેક આપણે department પ્રમાણે student count અને average marks બંને બતાવવા માંગીએ છીએ.

import matplotlib.pyplot as plt
import numpy as np

departments = ["IT", "CS", "Mechanical", "Civil"]
students = [100, 150, 80, 60]
avg_marks = [72, 85, 68, 75]

x = np.arange(len(departments))
width = 0.35

plt.figure(figsize=(9, 5))

plt.bar(
    x - width / 2,
    students,
    width=width,
    label="Students"
)

plt.bar(
    x + width / 2,
    avg_marks,
    width=width,
    label="Average Marks"
)

plt.xticks(x, departments)
plt.title("Department Comparison")
plt.xlabel("Department")
plt.ylabel("Value")
plt.legend()
plt.show()
```
