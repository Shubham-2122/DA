```python
# Session 9.10 — Annotations, Legends & Highlighting
# આજના session માં આપણે Matplotlib chart ને વધુ professional અને business-report friendly બનાવતાં શીખીશું. માત્ર chart બનાવવો પૂરતો નથી; chart માં સૌથી વધુ sales કયા મહિને થઈ, કયો point મહત્વનો છે અને કઈ માહિતી viewer એ તરત સમજવી જોઈએ તે પણ બતાવવું જરૂરી છે.
# 1. Learning objectives
# આ session પૂર્ણ થયા પછી તમે શીખી શકશો:
# - ax.annotate() થી કોઈ ખાસ data point પર label અને arrow બતાવવા.
# - ax.text() થી chart પર text અને KPI box મૂકવા.
# - ax.legend() થી legend ની position અને layout control કરવા.
# - ax.scatter() વડે important point highlight કરવા.
# - ax.axhline() અને ax.axvline() વડે reference line ઉમેરવી.
# - Business report માટે insights highlight કરતો chart બનાવવો.
# 2. Annotation શું છે?
# Annotation એટલે chart માં કોઈ ખાસ data point વિશે વધારાની માહિતી આપવી. ઉદાહરણ તરીકે, sales chart માં maximum sales થયેલા મહિના પર arrow સાથે “Peak Sales” લખવું.

# ax.annotate()

# કોઈ ચોક્કસ data point તરફ arrow સાથે message બતાવે છે.



# ax.text()

# Chart પર text, note અથવા KPI box મૂકવા માટે.



# ax.legend()

# Chart માં દર્શાવેલી અલગ-अलग lines અથવા data series ઓળખવા માટે.
```

```python
# 3. ax.annotate() — Peak Sales highlight કરવી
# સૌપ્રથમ VS Code માં તમારી Jupyter Notebook ખોલો અને આ code run કરો.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]

fig, ax = plt.subplots(figsize=(10, 5))

x = list(range(len(months)))

ax.plot(
    x, sales,
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

# Maximum sales શોધો
max_idx = sales.index(max(sales))

# Peak point highlight કરો
ax.annotate(
    f"Peak Sales: {sales[max_idx]}",
    xy=(max_idx, sales[max_idx]),
    xytext=(max_idx - 2, sales[max_idx] + 15),
    arrowprops=dict(arrowstyle="->", linewidth=1.5),
    fontsize=11,
    ha="center"
)

ax.set_xticks(x, months)
ax.set_title("Monthly Sales Performance")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.grid(alpha=0.3)
ax.legend()

plt.show()


# Parameter	ઉપયોગ
# xy	જ્યાં actual data point છે તેનું location
# xytext	જ્યાં annotation નો text મૂકવો છે
# arrowprops	Arrow નો દેખાવ control કરે છે
# fontsize	Text નો size
# ha="center"	Text ને horizontal center align કરે છે
```

```python
# 4. ax.text() — KPI box બનાવવું
# Business dashboard માં Total Sales, Profit, Average Sales જેવા KPIs chart ની અંદર બતાવી શકાય છે.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]

fig, ax = plt.subplots(figsize=(10, 5))

x = list(range(len(months)))

ax.plot(x, sales, marker="o", label="Sales")

total_sales = sum(sales)
average_sales = sum(sales) / len(sales)

ax.text(
    0.03, 0.95,
    f"Total Sales: {total_sales}\nAverage: {average_sales:.1f}",
    transform=ax.transAxes,
    va="top",
    fontsize=11,
    bbox=dict(
        boxstyle="round,pad=0.5",
        facecolor="white",
        edgecolor="gray",
        alpha=0.9
    )
)

ax.set_xticks(x, months)
ax.set_title("Sales Dashboard")
ax.set_ylabel("Sales")
ax.legend()
ax.grid(alpha=0.3)

plt.show()

# અહીં ખાસ શું સમજવાનું છે?
# - transform=ax.transAxes: text ની position chart ના axes પ્રમાણે નક્કી કરે છે.
# - 0.03, 0.95: chart ની અંદર ડાબી બાજુથી 3% અને નીચેથી 95% જેટલી position.
# - va="top": text ની ઉપરની ધારને આપેલી position પર રાખે છે.
# - bbox: text ની પાછળ box બનાવે છે.
# - alpha=0.9: box લગભગ opaque રાખે છે.

# ax.transAxes માં સામાન્ય રીતે x અને y ની range 0 થી 1 હોય છે. આ રીતે data values બદલાય તો પણ KPI box chart ની અંદર નિશ્ચિત જગ્યાએ રહી શકે છે.
```

```python
# 5. ax.legend() — Legend ની position control કરવી
# જ્યારે chart માં એકથી વધુ lines હોય, ત્યારે legend બતાવે છે કે કઈ line કયા data ને represent કરે છે.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales_2025 = [100, 120, 140, 130, 160, 180]
sales_2026 = [120, 150, 135, 180, 200, 220]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(months, sales_2025, marker="o", label="Sales 2025")
ax.plot(months, sales_2026, marker="s", label="Sales 2026")

ax.set_title("Sales Comparison: 2025 vs 2026")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

ax.legend(
    loc="upper left",
    title="Sales Year",
    frameon=True,
    ncol=1
)

ax.grid(alpha=0.3)
plt.show()

# Legend ના ઉપયોગી parameters
# Parameter	સમજણ
# loc="best"	Matplotlib યોગ્ય જગ્યા પસંદ કરવાનો પ્રયાસ કરે છે
# loc="upper left"	ઉપર ડાબી બાજુ
# loc="lower right"	નીચે જમણી બાજુ
# title="Sales Year"	Legend માટે heading
# ncol=2	Legend ને બે columns માં બતાવે છે
# frameon=False	Legend ની આસપાસનું box દૂર કરે છે
# bbox_to_anchor	Legend ને chart ની બહાર જેવી ચોક્કસ position પર મૂકવામાં મદદ કરે છે
```

```python
# Legend ને chart ની બહાર મૂકવું
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(months, sales_2025, marker="o", label="Sales 2025")
ax.plot(months, sales_2026, marker="s", label="Sales 2026")

ax.legend(
    loc="upper left",
    bbox_to_anchor=(1.02, 1)
)

fig.subplots_adjust(right=0.78)
plt.show()

# bbox_to_anchor=(1.02, 1) legend ને axes ની જમણી બાજુ મૂકવામાં મદદ કરે છે. subplots_adjust() legend માટે જગ્યા બનાવે છે.

```

```python
# 6. Important data point ને અલગથી highlight કરવો
# ધારો કે આપણે maximum sales થયેલા મહિને એક મોટો marker બતાવવા માંગીએ છીએ.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]

x = list(range(len(months)))
max_idx = sales.index(max(sales))

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(x, sales, marker="o", label="Monthly Sales")

# Maximum point ને અલગથી highlight કરો
ax.scatter(
    max_idx,
    sales[max_idx],
    s=180,
    label="Highest Sales"
)

# Reference line
ax.axhline(
    y=180,
    linestyle="--",
    linewidth=1,
    label="Target Sales = 180"
)

ax.annotate(
    "Target Achieved",
    xy=(max_idx, sales[max_idx]),
    xytext=(max_idx - 2, sales[max_idx] + 15),
    arrowprops=dict(arrowstyle="->"),
    ha="center"
)

ax.set_xticks(x, months)
ax.set_title("Sales Target Analysis")
ax.set_ylabel("Sales")
ax.legend()
ax.grid(alpha=0.3)

plt.show()
```

```python
# 7. Business report માટે સંપૂર્ણ practical
# હવે આ બધું એક જ chart માં જોડીએ: monthly sales, target line, peak sales annotation, KPI box અને legend.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]
target = 180

x = list(range(len(months)))
max_idx = sales.index(max(sales))

fig, ax = plt.subplots(figsize=(11, 6))

# Sales line
ax.plot(
    x, sales,
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

# Target line
ax.axhline(
    y=target,
    linestyle="--",
    label=f"Target = {target}"
)

# Peak point
ax.scatter(
    max_idx,
    sales[max_idx],
    s=150,
    label="Peak Sales"
)

# Annotation
ax.annotate(
    f"Peak: {sales[max_idx]}",
    xy=(max_idx, sales[max_idx]),
    xytext=(max_idx - 1.5, sales[max_idx] + 18),
    arrowprops=dict(arrowstyle="->"),
    ha="center"
)

# KPI box
ax.text(
    0.03, 0.95,
    f"Total Sales: {sum(sales)}\nTarget: {target}",
    transform=ax.transAxes,
    va="top",
    bbox=dict(boxstyle="round", facecolor="white", alpha=0.9)
)

ax.set_xticks(x, months)
ax.set_title("Monthly Sales Performance Report")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.legend(loc="lower right")
ax.grid(alpha=0.3)

fig.tight_layout()
plt.show()
```
