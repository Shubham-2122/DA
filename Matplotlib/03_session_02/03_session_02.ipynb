```python
# # Session 9.2 — Basic Plotting: Line Charts in Matplotlib
# # આજે આપણે plt.plot() ને ઊંડાણપૂર્વક શીખીશું. Session 9.1 માં તમે Matplotlib નો પરિચય, Figure, Axes અને Axis શીખ્યા હતા. હવે આપણે line charts ને professional રીતે બનાવતાં શીખીશું.
# # 1. Learning objectives
# # આ session પૂર્ણ થયા પછી તમે:
# # - plt.plot() ના મહત્વના parameters સમજી શકશો.
# # - marker, color, linewidth અને linestyle વાપરી શકશો.
# # - Chart માં title, labels, legend અને grid ઉમેરશો.
# # - એક જ chart માં multiple lines દોરી શકશો.
# # - Real-world sales અને student performance ના line charts બનાવી શકશો.
# # 2. Line Chart શું છે?
# # English definition: A line chart is a data visualization that connects data points with lines to show changes or trends, often over time.

# # ગુજરાતીમાં, અલગ-અલગ data points ને lines વડે જોડીને સમય પ્રમાણે અથવા ક્રમ પ્રમાણે થતો ફેરફાર દર્શાવતા chart ને Line Chart કહે છે.

# Line chart ક્યારે ઉપયોગી છે?
# - મહિના પ્રમાણે sales માં ફેરફાર.
# - દિવસ પ્રમાણે website visitors.
# - વર્ષ પ્રમાણે company revenue.
# - વિદ્યાર્થીના weekly test marks.
# - સમય પ્રમાણે temperature.
# જો તમારે અલગ-अलग categories ની સરખામણી કરવી હોય, તો Bar Chart પણ ઘણી વાર યોગ્ય પસંદગી હોય છે.
```

```python
# 3. Step 1 — Basic plt.plot() syntax

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x,y)

plt.show()
```

```python
# 4. Step 2 — plt.plot() ના મુખ્ય parameters
# Parameter	ઉપયોગ	Example
# color	Line નો રંગ	color="red"
# marker	Data point નું symbol	marker="o"
# linewidth	Line ની જાડાઈ	linewidth=3
# linestyle	Line નો પ્રકાર	linestyle="--"
# label	Legend માં નામ	label="Sales"
# alpha	પારદર્શકતા	alpha=0.7

# 5. Step 3 — Colors
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [10000, 15000, 12000, 20000, 25000]

plt.plot(months,sales,color="red")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales ($)")

plt.show()

# color="blue"
# color="green"
# color="orange"
# color="purple"
# color="#FF5733"
```

```python
# 6. Step 4 — Markers
# Marker એટલે chart પરના individual data points દર્શાવતું symbol.

plt.plot(
    months,
    sales,
    marker="o"
)

plt.show()

# Marker	અર્થ
# "o"	Circle
# "s"	Square
# "^"	Triangle
# "D"	Diamond
# "*"	Star
# "x"	Cross
```

```python
# વધુ customization માટે markersize અને markerfacecolor પણ વાપરી શકાય છે.

plt.plot(
    months,
    sales,
    marker="o",
    markersize=8,
    markerfacecolor="yellow",
    markeredgecolor="black"
)

plt.show()
```

```python
# 7. Step 5 — Linewidth અને Linestyle
# Linewidth
# Line કેટલી જાડી દેખાય તે નિયંત્રિત કરે છે

plt.plot(months,sales,linewidth=3)

plt.show()
```

```python
# inestyle
# Line solid, dashed કે dotted હશે તે નક્કી કરે છે.

plt.plot(months,sales,linestyle="--")
plt.show()

# Value	Line style
# "-"	Solid
# "--"	Dashed
# "-."	Dash-dot
# ":"	Dotted
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [10000, 15000, 12000, 20000, 25000]

plt.figure(figsize=(9,5))

plt.plot(
    months,
    sales,
    color="green",
    marker="o",
    markersize=8,
    linewidth=3,
    linestyle="--",
    label="Monthly Sales"
)

plt.title("Monthly Sales Analysis",fontsize=16)
plt.xlabel("Month",fontsize=12)
plt.ylabel("Sales $",fontsize=12)

plt.grid(True,alpha=0.3)
plt.legend()
plt.show()
```

```python
# 8. Step 6 — Multiple Lines એક જ Chart માં
# હવે કલ્પના કરો કે એક company પાસે બે products છે. તમારે બંનેની sales સરખાવવી છે.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

product_a = [10000, 15000, 12000, 20000, 25000]
product_b = [8000, 12000, 16000, 18000, 22000]

plt.figure(figsize=(9, 5))

plt.plot(
    months,
    product_a,
    marker="o",
    linewidth=2,
    label="Product A"
)

plt.plot(
    months,
    product_b,
    marker="s",
    linewidth=2,
    linestyle="--",
    label="Product B"
)

plt.title("Product Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales ($)")

plt.legend()
plt.grid(True,alpha=0.3)

plt.show()

# આ chart માં:
# - પહેલી line Product A ની sales બતાવે છે.
# - બીજી line Product B ની sales બતાવે છે.
# - label દરેક line ને નામ આપે છે.
# - plt.legend() line કયા product ની છે તે સમજાવે છે.
# Business insight: માત્ર chart જોઈને આપણે બંને products ની monthly sales સરખાવી શકીએ છીએ. ચોક્કસ business conclusion માટે actual values અને સમયગાળો પણ તપાસવો જોઈએ.
```

```python
# 9. Step 7 — Axis limits અને ticks
# ક્યારેક chart ના axis પર values કેવી રીતે દેખાય તે નિયંત્રિત કરવું પડે છે

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [10000, 15000, 12000, 20000, 25000]

plt.plot(months, sales, marker="o")

plt.ylim(0,30000)
plt.yticks([0,10000,20000,30000])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")

plt.grid(True, alpha=0.3)
plt.show()

# - plt.ylim(0, 30000) Y-axis ની range નક્કી કરે છે.
# - plt.yticks(...) Y-axis પર કઈ values ના tick labels દેખાશે તે નક્કી કરે છે.
# Axis limit પસંદ કરતી વખતે ધ્યાન રાખો: અયોગ્ય રીતે axis range કાપવાથી data માં રહેલો ફેરફાર વાસ્તવિક કરતાં વધારે કે ઓછો દેખાઈ શકે છે.

```

```python
# 10. Step 8 — Pandas DataFrame માંથી Line Chart
# તમે અગાઉ Pandas શીખ્યા છો, એટલે હવે DataFrame સીધું chart માં વાપરીએ.

import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [10000, 15000, 12000, 20000, 25000]
})

plt.figure(figsize=(8,4))

plt.plot(
    df["Month"],
    df["Sales"],
    marker="o",
    linewidth=2,
    label="Sales"
)
plt.title("Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
