```python
# # Session 9.5 — Histogram (Distribution Plot) in Matplotlib
# # આજે આપણે Python ની Matplotlib લાઇબ્રેરીમાં Histogram શીખીશું. Data Science માં Histogram નો ઉપયોગ marks, salary, age, product prices અને અન્ય numerical data નું distribution સમજવા માટે થાય છે.
# # 1. Learning Objectives
# # આ Session પૂર્ણ થયા પછી તમે:
# # - Histogram શું છે અને તેનો ઉપયોગ ક્યારે થાય છે તે સમજી શકશો.
# # - plt.hist() દ્વારા Histogram બનાવી શકશો.
# # - bins, range, edgecolor, alpha જેવા parameters વાપરી શકશો.
# # - Frequency અને Density વચ્ચેનો તફાવત સમજી શકશો.
# # - Pandas DataFrame માંથી Histogram બનાવી શકશો.
# # - Mean અને Median દર્શાવતી vertical lines ઉમેરી શકશો.
# # 2. Histogram શું છે?
# # Histogram એ numerical data કઈ range માં કેટલો છે તે દર્શાવતો chart છે.

# # ઉદાહરણ તરીકે, એક ક્લાસના વિદ્યાર્થીઓના marks નીચે મુજબ છે:

# # 35, 42, 45, 48, 50, 52, 55, 60, 62, 65, 70, 72, 75, 78, 80, 85, 90, 95

# # આ marks ને અલગ-અલગ intervals માં વહેંચી શકાય છે. જેમ કે:
# # - 30–49 marks
# # - 50–69 marks
# # - 70–89 marks
# # - 90–109 marks
# # પછી દરેક interval માં કેટલા વિદ્યાર્થીઓ છે તે ગણવામાં આવે છે. આ જ Histogram નો મૂળ વિચાર છે.

# મુદ્દો	Histogram	Bar Chart
# ઉપયોગ	Numerical data નું distribution	Categories ની સરખામણી
# X-axis	Numerical intervals અથવા bins	Categories
# Bars	સામાન્ય રીતે જોડાયેલા	સામાન્ય રીતે અલગ
# ઉદાહરણ	વિદ્યાર્થીઓના marks નું distribution	દરેક subject ના average marks

# યાદ રાખો: Marks ની range માં કેટલા વિદ્યાર્થીઓ છે તે જાણવા Histogram વાપરો. Maths, Science અને English ના marks સરખાવવા Bar Chart વાપરો.
```

```python
# 4. Basic Histogram — plt.hist()
import matplotlib.pyplot as plt

marks = [
    35, 42, 45, 48, 50, 52,
    55, 60, 62, 65, 70, 72,
    75, 78, 80, 85, 90, 95
]

plt.hist(marks,bins=5,edgecolor="black") # 5 bar banvi ya aevu dekahy

plt.title("Distrubution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()

# Code	અર્થ
# plt.hist(marks)	Marks નો Histogram બનાવે છે
# bins=5	Data ને 5 intervals માં વહેંચે છે
# edgecolor="black"	દરેક bar ની boundary કાળી કરે છે
```

```python
# 5. bins શું છે?
# bins એટલે data ને કેટલા intervals અથવા ભાગોમાં વહેંચવો.

plt.hist(marks, bins=5, edgecolor="black")
plt.title("Histogram with 5 Bins")
plt.show()
```

```python
plt.hist(marks, bins=10, edgecolor="black")
plt.title("Histogram with 10 Bins")
plt.show()

# બંને charts માં data એક જ છે, પરંતુ intervals ની સંખ્યા બદલાઈ છે.
# - ઓછા bins: Distribution નો મોટો pattern સરળતાથી દેખાય છે, પરંતુ કેટલીક વિગતો છુપાઈ શકે છે.
# - વધુ bins: Data ની નાની વિગતો દેખાય છે, પરંતુ chart વધુ noisy લાગી શકે છે.
```

```python
# 6. Histogram ના ઉપયોગી Parameters
# Parameter	ઉપયોગ	Example
# bins	Intervals ની સંખ્યા	bins=8
# range	કઈ numerical range દર્શાવવી	range=(0, 100)
# color	Bars નો રંગ	color="skyblue"
# edgecolor	Bars ની boundary નો રંગ	edgecolor="black"
# alpha	Transparency, 0 થી 1	alpha=0.7
# histtype	Histogram નો પ્રકાર	histtype="bar"
# density	Frequency બદલે density દર્શાવવી	density=True
plt.hist(
    marks,
    bins=8,
    range=(0, 100),
    color="skyblue",
    edgecolor="black",
    alpha=0.8
)

plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
# 7. Frequency અને Density વચ્ચેનો તફાવત
# આ Histogram નો ખૂબ જ મહત્વપૂર્ણ concept છે.
# A. Frequency — density=False

plt.hist(marks, bins=5, density=False, edgecolor="black")

plt.title("Histogram - Frequency")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()

# Frequency એટલે દરેક bin માં કેટલા data points છે.
# density=False default setting છે.
# Y-axis પર count અથવા વિદ્યાર્થીઓની સંખ્યા જોવા મળે છે.
```

```python
# B. Density — density=True

plt.hist(marks, bins=5, density=True, edgecolor="black")

plt.title("Histogram - Density")
plt.xlabel("Marks")
plt.ylabel("Probability Density")
plt.show()

# ensity data distribution નો normalized આકાર દર્શાવે છે.
# Density Histogram માં બધા bars ની કુલ area 1 થાય છે; bars ની heights નો સરવાળો જરૂરી નથી કે 1 થાય.
# અલગ-અલગ sample sizes ધરાવતા datasets ના distributions સરખાવવામાં તે ઉપયોગી છે.
```

```python
# મુદ્દો	        density=False	                density=True
# Y-axis	    requency / Count	            Probability Density
# મુખ્ય ઉપયોગ	 કેટલા observations છે તે જાણવા	    Distribution સરખાવવા
# Default	    હા	                             ના
```

```python
# 8. Pandas DataFrame માંથી Histogram
# Data Science માં data ઘણી વખત CSV અથવા Excel ફાઇલમાંથી આવે છે. તેથી Pandas સાથે Histogram બનાવતા શીખીએ.
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Student": [
        "Amit", "Neha", "Ravi", "Pooja", "Karan",
        "Meera", "Jay", "Priya", "Raj", "Anjali"
    ],
    "Marks": [35, 42, 48, 55, 60, 65, 72, 78, 85, 92]
})

print(df)

plt.hist(df["Marks"], bins=5, edgecolor="black")

plt.title("Student Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
# 9. Histogram માં Mean અને Median દર્શાવવું
# હવે Histogram ને વધુ informative બનાવીએ. Mean અને Median ની location chart પર vertical lines વડે દર્શાવીશું.

import matplotlib.pyplot as plt
import numpy as np

marks = [
    35, 42, 45, 48, 50, 52,
    55, 60, 62, 65, 70, 72,
    75, 78, 80, 85, 90, 95
]

mean_marks = np.mean(marks)
median_marks = np.median(marks)

plt.hist(
    marks,
    bins=5,
    edgecolor="black",
    alpha=0.7,
    label="Students' Marks"
)

plt.axvline(
    mean_marks,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean_marks:.1f}"
)

plt.axvline(
    median_marks,
    linestyle=":",
    linewidth=2,
    label=f"Median = {median_marks:.1f}"
)

plt.title("Marks Distribution with Mean and Median")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()
plt.show()

# - np.mean(marks) સરેરાશ શોધે છે.
# - np.median(marks) મધ્યસ્થ value શોધે છે.
# - plt.axvline() આપેલી X-axis value પર vertical line દોરે છે.
# - label અને plt.legend() chart માં દરેક line ઓળખવામાં મદદ કરે છે.
# Mean અને Median એકબીજાની નજીક હોય તો તે distribution ના આકાર વિશે સંકેત આપી શકે છે. પરંતુ માત્ર આ બંને values પરથી distribution વિશે સંપૂર્ણ નિર્ણય ન લેવો.
```
