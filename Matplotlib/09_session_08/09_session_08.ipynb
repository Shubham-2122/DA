```python
# Session 9.8 — Charts Customization & Style Management in Matplotlib
# આજે આપણે શીખીશું કે Matplotlib માં બનાવેલા charts ને professional, attractive અને presentation-ready કેવી રીતે બનાવવા.

# Data Science માં માત્ર chart બનાવવો પૂરતો નથી. Chart ના colors, font size, labels, grid, figure size અને layout પણ યોગ્ય હોવા જોઈએ, જેથી data સરળતાથી સમજી શકાય.
# 1. Learning Objectives
# આ Session પૂર્ણ થયા પછી તમે:
# - figsize અને dpi દ્વારા chart નું કદ અને resolution નિયંત્રિત કરી શકશો.
# - color, linestyle, linewidth અને marker customize કરી શકશો.
# - plt.style.use() વડે અલગ chart styles લાગુ કરી શકશો.
# - alpha, grid, legend અને tick_params() વાપરી શકશો.
# - tight_layout() અને savefig() સાથે presentation-ready charts તૈયાર કરી શકશો.

```

```python
# 2. Basic Chart ને Professional બનાવો
# સૌપ્રથમ સામાન્ય Line Chart બનાવીએ.
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
# aama style aapi che aetle loko jove

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    months,
    sales,
    color="royalblue",
    linestyle="-",
    linewidth=2.5,
    marker="o",
    markersize=7,
    markerfacecolor="orange",
    markeredgecolor="black",
    label="Monthly Sales"
)

ax.set_title("Monthly Sales Report", fontsize=16, fontweight="bold")
ax.set_xlabel("Month", fontsize=12)
ax.set_ylabel("Sales (Units)", fontsize=12)

ax.tick_params(axis="both", labelsize=10)
ax.grid(True, linestyle="--", alpha=0.3)
ax.legend()

fig.tight_layout()
plt.show()



# કયા parameters શું કરે છે?
# Parameter	ઉપયોગ	Example
# figsize	Figure નું કદ	(10, 5)
# color	Line નો રંગ	"royalblue"
# linestyle	Line નો પ્રકાર	"--", ":", "-."
# linewidth	Line ની જાડાઈ	2.5
# marker	Data point નો આકાર	"o", "s", "^"
# markersize	Marker નું કદ	7
# fontsize	Text નું કદ	16
# fontweight	Text ની જાડાઈ	"bold"
# alpha	Transparency, 0 થી 1	0.3
# label	Legend માં નામ	"Monthly Sales"
```

```python
# 3. Figure Size અને DPI
# figsize શું છે?
# figsize=(width, height) Figure નું કદ inches માં નક્કી કરે છે.

import matplotlib.pyplot as plt

sales = [120, 150, 135, 180, 200, 220]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(months, sales, marker="o")
ax.set_title("Wide Sales Chart")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 5), dpi=100)

ax.plot(months, sales, marker="o")
ax.set_title("Sales Chart")

plt.show()
```

```python
# 4. Line Styles અને Markers
# એક જ data પર અલગ line styles અજમાવીએ.

import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [10, 20, 25, 35, 40]
y2 = [12, 18, 28, 30, 45]
y3 = [8, 15, 20, 25, 32]

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(x, y1, linestyle="-", marker="o", label="Solid Line")
ax.plot(x, y2, linestyle="--", marker="s", label="Dashed Line")
ax.plot(x, y3, linestyle=":", marker="^", label="Dotted Line")

ax.set_title("Line Style Comparison")
ax.set_xlabel("X Values")
ax.set_ylabel("Y Values")
ax.legend()
ax.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
```

```python
# 5. Matplotlib Styles — plt.style.use()
# Matplotlib માં પહેલેથી ઉપલબ્ધ styles વડે charts નો દેખાવ બદલી શકાય છે.

import matplotlib.pyplot as plt

print(plt.style.available)
```

```python
import matplotlib.pyplot as plt

plt.style.use("ggplot")

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 135, 180, 200, 220]

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(months, sales, marker="o", linewidth=2)
ax.set_title("Monthly Sales — ggplot Style")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")

plt.show()
```

```python
# એક જ data માટે styles ની સરખામણી
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 130, 125, 170, 200]

styles = ["default", "ggplot", "bmh"]

for style in styles:
    with plt.style.context(style):
        fig, ax = plt.subplots(figsize=(7, 3))
        ax.plot(months, sales, marker="o")
        ax.set_title(f"Style: {style}")
        ax.set_xlabel("Month")
        ax.set_ylabel("Sales")
        fig.tight_layout()
        plt.show()
```

```python
# 6. Colors અને Transparency
# Colors chart ને સમજવામાં મદદરૂપ થવા જોઈએ. માત્ર સુંદર દેખાવ માટે વધારે રંગો ઉમેરવા કરતાં જરૂરી તફાવત સ્પષ્ટ કરવો વધુ સારો અભિગમ છે.

import matplotlib.pyplot as plt

categories = ["Electronics", "Fashion", "Books", "Furniture"]
sales = [450, 300, 150, 250]

fig, ax = plt.subplots(figsize=(9, 5))

bars = ax.bar(
    categories,
    sales,
    color=["steelblue", "orange", "seagreen", "mediumpurple"],
    edgecolor="black",
    alpha=0.85
)

ax.bar_label(bars, padding=3)
ax.set_title("Category-wise Sales", fontsize=15)
ax.set_xlabel("Category")
ax.set_ylabel("Sales")
ax.grid(axis="y", alpha=0.25)

fig.tight_layout()
plt.show()
```

```python
# 7. Tick Labels અને Rotation
# કેટલીક categories ના નામ લાંબા હોય છે, જેના કારણે labels એકબીજા સાથે અથડાય છે. તેને સુધારવા માટે rotation વાપરી શકાય છે.

import matplotlib.pyplot as plt

categories = [
    "Mobile Phones",
    "Home Appliances",
    "Computer Accessories",
    "Fashion Products",
    "Sports Equipment"
]

sales = [300, 250, 180, 220, 160]

fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(categories, sales)

ax.set_title("Product Category Sales")
ax.set_xlabel("Category")
ax.set_ylabel("Sales")

ax.tick_params(axis="x", labelrotation=25)
fig.tight_layout()

plt.show()
```

```python
# 8. Professional Customization — Complete Example
# હવે અગાઉ શીખેલા concepts જોડીને એક report-style chart બનાવીએ.

import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales_2025 = [100, 125, 115, 150, 170, 190]
sales_2026 = [120, 140, 135, 175, 200, 230]

fig, ax = plt.subplots(figsize=(11, 6), dpi=100)

ax.plot(
    months,
    sales_2025,
    color="slategray",
    linestyle="--",
    marker="s",
    linewidth=2,
    label="2025 Sales"
)

ax.plot(
    months,
    sales_2026,
    color="royalblue",
    linestyle="-",
    marker="o",
    linewidth=2.5,
    label="2026 Sales"
)

ax.set_title("Monthly Sales Comparison", fontsize=16, fontweight="bold")
ax.set_xlabel("Month", fontsize=12)
ax.set_ylabel("Sales (Units)", fontsize=12)

ax.tick_params(axis="both", labelsize=10)
ax.grid(axis="y", linestyle=":", alpha=0.4)
ax.legend(loc="upper left", frameon=True)

fig.tight_layout()
plt.show()
```
