```python
# # 5. Step 4 — Complete Sales Dashboard બનાવો

# Monthly Sales — Line Chart
# Category Revenue — Bar Chart
# Sales Distribution — Histogram
# Advertising vs Sales — Scatter Plot અને Trend Line
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# -------------------------------
# 1. Dataset
# -------------------------------
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 135, 180, 200, 220],
    "Profit": [24, 32, 27, 40, 45, 52],
    "Advertising": [15, 20, 18, 25, 30, 35],
    "Orders": [40, 48, 44, 58, 65, 72]
}

df = pd.DataFrame(data)

category_sales = pd.DataFrame({
    "Category": [
        "Electronics", "Furniture", "Clothing",
        "Books", "Beauty"
    ],
    "Revenue": [350, 220, 280, 100, 180]
})

# -------------------------------
# 2. KPI calculations
# -------------------------------
total_sales = df["Sales"].sum()
average_sales = df["Sales"].mean()
total_profit = df["Profit"].sum()

best_idx = df["Sales"].idxmax()
best_month = df.loc[best_idx, "Month"]
best_sales = df.loc[best_idx, "Sales"]

# -------------------------------
# 3. Create dashboard
# -------------------------------
fig, ax = plt.subplots(2, 2, figsize=(14, 9))

fig.suptitle(
    "Sales Performance Dashboard",
    fontsize=18
)

# Chart 1: Line Chart
ax[0, 0].plot(
    df["Month"],
    df["Sales"],
    marker="o",
    linewidth=2,
    label="Monthly Sales"
)

ax[0, 0].annotate(
    f"Peak: {best_sales}",
    xy=(best_month, best_sales),
    xytext=(-45, -30),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

ax[0, 0].set_title("Monthly Sales Trend")
ax[0, 0].set_xlabel("Month")
ax[0, 0].set_ylabel("Sales")
ax[0, 0].legend()
ax[0, 0].grid(alpha=0.3)

# Chart 2: Bar Chart
ax[0, 1].bar(
    category_sales["Category"],
    category_sales["Revenue"],
    label="Revenue"
)

ax[0, 1].set_title("Revenue by Category")
ax[0, 1].set_xlabel("Category")
ax[0, 1].set_ylabel("Revenue")
ax[0, 1].tick_params(axis="x", rotation=25)
ax[0, 1].legend()

# Chart 3: Histogram
ax[1, 0].hist(
    df["Sales"],
    bins=5,
    edgecolor="black"
)

ax[1, 0].axvline(
    average_sales,
    linestyle="--",
    label=f"Mean: {average_sales:.1f}"
)

ax[1, 0].set_title("Sales Distribution")
ax[1, 0].set_xlabel("Sales")
ax[1, 0].set_ylabel("Frequency")
ax[1, 0].legend()

# Chart 4: Scatter Plot + Trend Line
ax[1, 1].scatter(
    df["Advertising"],
    df["Sales"],
    label="Monthly Data"
)

# Linear trend line
slope, intercept = np.polyfit(
    df["Advertising"],
    df["Sales"],
    1
)

trend_y = (
    slope * df["Advertising"] + intercept
)

ax[1, 1].plot(
    df["Advertising"],
    trend_y,
    linestyle="--",
    label="Trend Line"
)

ax[1, 1].set_title("Advertising vs Sales")
ax[1, 1].set_xlabel("Advertising")
ax[1, 1].set_ylabel("Sales")
ax[1, 1].legend()
ax[1, 1].grid(alpha=0.3)

# -------------------------------
# 4. Add KPI box
# -------------------------------
fig.text(
    0.5, 0.925,
    f"Total Sales: {total_sales}  |  "
    f"Average: {average_sales:.1f}  |  "
    f"Total Profit: {total_profit}",
    ha="center",
    fontsize=12
)

# Layout adjustment
fig.tight_layout(rect=[0, 0, 1, 0.90])

# -------------------------------
# 5. Save dashboard
# -------------------------------
os.makedirs("reports", exist_ok=True)

fig.savefig(
    "reports/sales_dashboard.png",
    dpi=300,
    bbox_inches="tight"
)

fig.savefig(
    "reports/sales_dashboard.pdf",
    bbox_inches="tight"
)

plt.show()

print("Dashboard saved successfully.")
print("Best Month:", best_month)
print("Best Sales:", best_sales)
```
